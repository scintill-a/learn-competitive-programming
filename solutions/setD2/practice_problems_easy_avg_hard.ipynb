# 🟢 Problem 1: Employee Payroll & Net Pay Breakdown (Easy — Sequential/Payroll Math)

### 📖 Problem Statement
An accounting automation script needs to compute the bi-weekly paycheck summary for hourly employees. Given an employee's **hourly wage**, **hours worked**, **tax deduction percentage**, and **fixed medical insurance deduction** on a single line:
1. Compute the **Gross Pay** (`hourly_wage * hours_worked`).
2. Compute the **Tax Deduction** (`gross_pay * (tax_pct / 100)`).
3. Compute **Total Deductions** (`tax_deduction + medical_insurance`).
4. Compute the final **Net Pay** (`gross_pay - total_deductions`).

Format all financial values with dollar signs and exactly 2 decimal places (`$XX.XX`).

### 📥 Input Format
A single line containing four space-separated numbers:
`hourly_wage` (float) `hours_worked` (float) `tax_pct` (float) `medical_insurance` (float)

### 📤 Output Format
`Gross: $XX.XX | Deductions: $XX.XX | Net: $XX.XX`

### 🔒 Constraints
- 0.00 <= hourly_wage <= 1,000.00
- 0.00 <= hours_worked <= 168.00
- 0.00 <= tax_pct <= 50.00
- 0.00 <= medical_insurance <= 500.00

### 💡 Example 1
```text
Input:
25.50 40 12.0 45.00

Output:
Gross: $1020.00 | Deductions: $167.40 | Net: $852.60
```
*Explanation:*
- Gross = 25.50 × 40 = 1020.00
- Tax (12%) = 1020.00 × 0.12 = 122.40
- Deductions = 122.40 + 45.00 = 167.40
- Net = 1020.00 - 167.40 = 852.60

### 💡 Example 2
```text
Input:
18.00 35 10.0 0.00

Output:
Gross: $630.00 | Deductions: $63.00 | Net: $567.00
```


````python
# 💡 Problem 1 Solution: Employee Payroll & Net Pay Breakdown

def calculate_payroll(hourly_wage: float, hours_worked: float, tax_pct: float, medical_insurance: float) -> str:
    gross_pay = hourly_wage * hours_worked
    tax_deduction = gross_pay * (tax_pct / 100.0)
    total_deductions = tax_deduction + medical_insurance
    net_pay = gross_pay - total_deductions
    
    return f"Gross: ${gross_pay:.2f} | Deductions: ${total_deductions:.2f} | Net: ${net_pay:.2f}"

def solve_problem_1():
    print("--- Running Problem 1 (Input/Output) ---")
    try:
        raw = input("Enter wage, hours, tax_pct, medical: ").split()
        if raw:
            w, h, t, m = float(raw[0]), float(raw[1]), float(raw[2]), float(raw[3])
            print(calculate_payroll(w, h, t, m))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((25.50, 40.0, 12.0, 45.00), "Gross: $1020.00 | Deductions: $167.40 | Net: $852.60"),
        ((18.00, 35.0, 10.0, 0.00), "Gross: $630.00 | Deductions: $63.00 | Net: $567.00"),
        ((50.00, 20.0, 20.0, 100.00), "Gross: $1000.00 | Deductions: $300.00 | Net: $700.00"),
    ]
    for args, expected in cases:
        actual = calculate_payroll(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Fuel Economy & Road Trip Cost Splitter (Easy — Sequential/Ratio Math)

### 📖 Problem Statement
A group of friends goes on a road trip. Given the **total trip distance** in kilometers, the car's **fuel consumption rate** in liters per 100 km (L/100km), the **current fuel price per liter**, and the **number of passengers** sharing the fuel expense equally:
1. Compute the **total liters of fuel needed** (`(distance / 100) * consumption_rate`).
2. Compute the **total fuel cost** (`liters_needed * fuel_price`).
3. Compute the **cost per passenger** (`total_cost / passengers`).

Format liters to 2 decimal places (`<liters:.2f> L`) and all currency values to 2 decimal places (`$XX.XX`).

### 📥 Input Format
A single line containing four space-separated values:
`distance_km` (float) `consumption_l_per_100km` (float) `fuel_price_per_l` (float) `passengers` (int)

### 📤 Output Format
`Fuel Needed: <liters:.2f> L | Total Cost: $XX.XX | Per Person: $XX.XX`

### 🔒 Constraints
- 1.0 <= distance_km <= 20,000.0
- 1.0 <= consumption_l_per_100km <= 50.0
- 0.10 <= fuel_price_per_l <= 10.00
- 1 <= passengers <= 50

### 💡 Example 1
```text
Input:
450.0 8.0 1.65 3

Output:
Fuel Needed: 36.00 L | Total Cost: $59.40 | Per Person: $19.80
```
*Explanation:*
- Fuel Needed = (450.0 / 100) × 8.0 = 36.00 L
- Total Cost = 36.00 × 1.65 = 59.40
- Per Person = 59.40 / 3 = 19.80

### 💡 Example 2
```text
Input:
125.5 6.4 1.80 2

Output:
Fuel Needed: 8.03 L | Total Cost: $14.46 | Per Person: $7.23
```


````python
# 💡 Problem 2 Solution: Fuel Economy & Road Trip Cost Splitter

def calculate_trip_fuel(distance_km: float, consumption: float, price_per_l: float, passengers: int) -> str:
    liters_needed = (distance_km / 100.0) * consumption
    total_cost = liters_needed * price_per_l
    per_person = total_cost / passengers if passengers > 0 else 0.0
    
    return f"Fuel Needed: {liters_needed:.2f} L | Total Cost: ${total_cost:.2f} | Per Person: ${per_person:.2f}"

def solve_problem_2():
    print("--- Running Problem 2 (Input/Output) ---")
    try:
        raw = input("Enter distance, consumption, price, passengers: ").split()
        if raw:
            d, c, p, n = float(raw[0]), float(raw[1]), float(raw[2]), int(raw[3])
            print(calculate_trip_fuel(d, c, p, n))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        ((450.0, 8.0, 1.65, 3), "Fuel Needed: 36.00 L | Total Cost: $59.40 | Per Person: $19.80"),
        ((125.5, 6.4, 1.80, 2), "Fuel Needed: 8.03 L | Total Cost: $14.46 | Per Person: $7.23"),
        ((1000.0, 5.5, 1.50, 4), "Fuel Needed: 55.00 L | Total Cost: $82.50 | Per Person: $20.62"),
    ]
    for args, expected in cases:
        actual = calculate_trip_fuel(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Package Freight & Dimensional Weight Classifier (Average — Selection)

### 📖 Problem Statement
A courier service charges customers based on a package's **Billable Weight**, which is the greater of its **Actual Weight (kg)** or its **Dimensional Weight (kg)**:
```text
Dimensional Weight = (Length_cm × Width_cm × Height_cm) / 5000.0
Billable Weight = max(Actual Weight, Dimensional Weight)
```

**Weight Limit Rule**:
If `Billable Weight > 50.0 kg`, the parcel cannot be processed by standard couriers. Output: `REJECTED: Overweight/Oversized`.

**Pricing Schedule & Categories**:
- `Billable Weight <= 2.0 kg`: Tier `Small Parcel` | Base Cost = $12.00 flat.
- `2.0 kg < Billable Weight <= 10.0 kg`: Tier `Medium Package` | Base Cost = $12.00 + (Billable Weight - 2.0) × $3.50.
- `10.0 kg < Billable Weight <= 30.0 kg`: Tier `Heavy Cargo` | Base Cost = $40.00 + (Billable Weight - 10.0) × $2.50.
- `30.0 kg < Billable Weight <= 50.0 kg`: Tier `Freight` | Base Cost = $90.00 + (Billable Weight - 30.0) × $2.00.

**Shipping Mode Surcharge**:
- If `shipping_mode` is `EXPRESS`, add a 25% surcharge to the base cost (`base_cost * 1.25`).
- If `shipping_mode` is `STANDARD`, no surcharge applies.

### 📥 Input Format
A single line containing:
`length_cm` (float) `width_cm` (float) `height_cm` (float) `actual_weight_kg` (float) `shipping_mode` (string: `STANDARD` or `EXPRESS`)

### 📤 Output Format
If rejected:
`REJECTED: Overweight/Oversized`
Otherwise:
`Billable Weight: <billable_weight:.2f> kg | Tier: <Tier> | Total Cost: $XX.XX`

### 🔒 Constraints
- Dimensions and weight > 0.0
- shipping_mode in {"STANDARD", "EXPRESS"}

### 💡 Example 1
```text
Input:
30 20 15 1.5 STANDARD

Output:
Billable Weight: 1.80 kg | Tier: Small Parcel | Total Cost: $12.00
```

### 💡 Example 2
```text
Input:
50 40 30 8.0 EXPRESS

Output:
Billable Weight: 12.00 kg | Tier: Heavy Cargo | Total Cost: $56.25
```


````python
# 💡 Problem 3 Solution: Package Freight & Dimensional Weight Classifier

def calculate_shipping(length_cm: float, width_cm: float, height_cm: float, actual_weight: float, mode: str) -> str:
    dim_weight = (length_cm * width_cm * height_cm) / 5000.0
    billable_weight = max(actual_weight, dim_weight)
    
    if billable_weight > 50.0:
        return "REJECTED: Overweight/Oversized"
        
    if billable_weight <= 2.0:
        tier = "Small Parcel"
        base_cost = 12.00
    elif billable_weight <= 10.0:
        tier = "Medium Package"
        base_cost = 12.00 + (billable_weight - 2.0) * 3.50
    elif billable_weight <= 30.0:
        tier = "Heavy Cargo"
        base_cost = 40.00 + (billable_weight - 10.0) * 2.50
    else:
        tier = "Freight"
        base_cost = 90.00 + (billable_weight - 30.0) * 2.00
        
    if mode.upper() == "EXPRESS":
        total_cost = base_cost * 1.25
    else:
        total_cost = base_cost
        
    return f"Billable Weight: {billable_weight:.2f} kg | Tier: {tier} | Total Cost: ${total_cost:.2f}"

def solve_problem_3():
    print("--- Running Problem 3 (Decision / Selection) ---")
    try:
        raw = input("Enter length, width, height, weight, mode: ").split()
        if raw:
            l, w, h, wt, m = float(raw[0]), float(raw[1]), float(raw[2]), float(raw[3]), raw[4]
            print(calculate_shipping(l, w, h, wt, m))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((30.0, 20.0, 15.0, 1.5, "STANDARD"), "Billable Weight: 1.80 kg | Tier: Small Parcel | Total Cost: $12.00"),
        ((50.0, 40.0, 30.0, 8.0, "EXPRESS"), "Billable Weight: 12.00 kg | Tier: Heavy Cargo | Total Cost: $56.25"),
        ((60.0, 60.0, 60.0, 40.0, "STANDARD"), "Billable Weight: 43.20 kg | Tier: Freight | Total Cost: $116.40"),
        ((100.0, 80.0, 50.0, 10.0, "STANDARD"), "REJECTED: Overweight/Oversized"),
    ]
    for args, expected in cases:
        actual = calculate_shipping(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Academic Honors & Grade Curve Classifier (Average — Selection)

### 📖 Problem Statement
A university department evaluates final student standings based on **Exam Score** (0.0 to 100.0), **Class Attendance** (0.0 to 100.0%), and a **Disciplinary Violation Flag** (`Y` or `N`).

**1. Attendance Curve Adjustment**:
- If `raw_score < 50.0` and `attendance_pct >= 85.0`, the student receives a curve boost of +5.0 points to help diligent students pass. However, the boosted score is capped at a maximum of 50.0.
- Otherwise, `final_score = raw_score`.

**2. Base Award Classification**:
- `final_score >= 90.0`:
  - If `attendance_pct >= 80.0`: `High Distinction`
  - Otherwise: `Distinction` (downgraded due to low attendance)
- `80.0 <= final_score < 90.0`: `Distinction`
- `70.0 <= final_score < 80.0`: `Merit`
- `50.0 <= final_score < 70.0`: `Pass`
- `final_score < 50.0`: `Fail`

**3. Disciplinary Penalty**:
If `violation_flag` is `Y`:
- Any honor standing (`High Distinction`, `Distinction`, `Merit`) is revoked and demoted to `Pass (Honors Forfeited)`.
- A `Pass` standing becomes `Pass (Probation)`.
- A `Fail` remains `Fail`.

### 📥 Input Format
A single line containing:
`raw_score` (float) `attendance_pct` (float) `violation_flag` (string: `Y` or `N`)

### 📤 Output Format
`Final Score: <final_score:.1f> | Award: <award_string>`

### 🔒 Constraints
- 0.0 <= raw_score <= 100.0
- 0.0 <= attendance_pct <= 100.0
- violation_flag in {"Y", "N"}

### 💡 Example 1
```text
Input:
92.5 85.0 N

Output:
Final Score: 92.5 | Award: High Distinction
```

### 💡 Example 2
```text
Input:
85.0 90.0 Y

Output:
Final Score: 85.0 | Award: Pass (Honors Forfeited)
```


````python
# 💡 Problem 4 Solution: Academic Honors & Grade Curve Classifier

def classify_student(raw_score: float, attendance_pct: float, violation_flag: str) -> str:
    # Curve adjustment
    if raw_score < 50.0 and attendance_pct >= 85.0:
        final_score = min(50.0, raw_score + 5.0)
    else:
        final_score = raw_score
        
    # Base award determination
    if final_score >= 90.0:
        if attendance_pct >= 80.0:
            base_award = "High Distinction"
        else:
            base_award = "Distinction"
    elif final_score >= 80.0:
        base_award = "Distinction"
    elif final_score >= 70.0:
        base_award = "Merit"
    elif final_score >= 50.0:
        base_award = "Pass"
    else:
        base_award = "Fail"
        
    # Disciplinary penalty
    if violation_flag.upper() == "Y":
        if base_award in ("High Distinction", "Distinction", "Merit"):
            award = "Pass (Honors Forfeited)"
        elif base_award == "Pass":
            award = "Pass (Probation)"
        else:
            award = "Fail"
    else:
        award = base_award
        
    return f"Final Score: {final_score:.1f} | Award: {award}"

def solve_problem_4():
    print("--- Running Problem 4 (Decision / Selection) ---")
    try:
        raw = input("Enter raw_score, attendance, violation (e.g. 92.5 85.0 N): ").split()
        if raw:
            s, a, v = float(raw[0]), float(raw[1]), raw[2]
            print(classify_student(s, a, v))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ((92.5, 85.0, "N"), "Final Score: 92.5 | Award: High Distinction"),
        ((94.0, 75.0, "N"), "Final Score: 94.0 | Award: Distinction"),
        ((85.0, 90.0, "Y"), "Final Score: 85.0 | Award: Pass (Honors Forfeited)"),
        ((46.0, 88.0, "N"), "Final Score: 50.0 | Award: Pass"),
        ((42.0, 70.0, "N"), "Final Score: 42.0 | Award: Fail"),
        ((55.0, 95.0, "Y"), "Final Score: 55.0 | Award: Pass (Probation)"),
    ]
    for args, expected in cases:
        actual = classify_student(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Smart Thermostat Anomaly & Cycle Detector (Difficult — Loop)

### 📖 Problem Statement
An environmental control unit monitors temperatures around a target setpoint. Given `target_temp` and allowed `tolerance`, evaluate a stream of consecutive temperature readings:
- **Operating Bounds**:
  - `Upper Bound = target_temp + tolerance`
  - `Lower Bound = target_temp - tolerance`
- **Operating Modes**:
  - If `reading > Upper Bound`: mode is `COOLING`.
  - If `reading < Lower Bound`: mode is `HEATING`.
  - If `Lower Bound <= reading <= Upper Bound`: mode is `IDLE`.
- **Mode Transition Cycles**:
  - A cycle is counted whenever the system transitions from one mode to a distinct mode (e.g. `IDLE` -> `COOLING`, `COOLING` -> `IDLE`, etc.). The initial reading sets the starting mode and does not count as a transition.
- **Severe Anomalies**:
  - A reading is flagged as a severe anomaly if `reading >= target_temp + 2 * tolerance` OR `reading <= target_temp - 2 * tolerance`.
- **Maximum Consecutive Streaks**:
  - Track the maximum length of consecutive readings spent in `COOLING` mode.
  - Track the maximum length of consecutive readings spent in `HEATING` mode.

### 📥 Input Format
- Line 1: `target_temp` (float) `tolerance` (float)
- Line 2: Sequence of float readings ending with `END`

### 📤 Output Format
`Cycles: <cycle_count> | Max Cooling Streak: <max_cooling> | Max Heating Streak: <max_heating> | Severe Anomalies: <spike_count>`

### 🔒 Constraints
- tolerance > 0.0
- Stream length between 1 and 100,000 readings

### 💡 Example 1
```text
Input:
22.0 2.0
22.5 23.0 25.0 26.5 24.5 22.0 19.0 17.5 21.0 END

Output:
Cycles: 4 | Max Cooling Streak: 3 | Max Heating Streak: 2 | Severe Anomalies: 2
```

### 💡 Example 2
```text
Input:
20.0 1.0
20.5 20.2 20.8 END

Output:
Cycles: 0 | Max Cooling Streak: 0 | Max Heating Streak: 0 | Severe Anomalies: 0
```


````python
# 💡 Problem 5 Solution: Smart Thermostat Anomaly & Cycle Detector

def analyze_thermostat(target_temp: float, tolerance: float, readings: list) -> str:
    if not readings:
        return "Cycles: 0 | Max Cooling Streak: 0 | Max Heating Streak: 0 | Severe Anomalies: 0"
        
    upper_bound = target_temp + tolerance
    lower_bound = target_temp - tolerance
    spike_high = target_temp + 2.0 * tolerance
    spike_low = target_temp - 2.0 * tolerance
    
    cycles = 0
    severe_anomalies = 0
    
    current_cooling = 0
    max_cooling = 0
    current_heating = 0
    max_heating = 0
    
    prev_state = None
    
    for r in readings:
        if r >= spike_high or r <= spike_low:
            severe_anomalies += 1
            
        if r > upper_bound:
            state = "COOLING"
        elif r < lower_bound:
            state = "HEATING"
        else:
            state = "IDLE"
            
        if prev_state is not None and state != prev_state:
            cycles += 1
        prev_state = state
        
        if state == "COOLING":
            current_cooling += 1
            if current_cooling > max_cooling:
                max_cooling = current_cooling
            current_heating = 0
        elif state == "HEATING":
            current_heating += 1
            if current_heating > max_heating:
                max_heating = current_heating
            current_cooling = 0
        else:
            current_cooling = 0
            current_heating = 0
            
    return f"Cycles: {cycles} | Max Cooling Streak: {max_cooling} | Max Heating Streak: {max_heating} | Severe Anomalies: {severe_anomalies}"

def solve_problem_5():
    print("--- Running Problem 5 (Loop / State Tracking) ---")
    try:
        raw_header = input("Enter target_temp and tolerance: ").split()
        if raw_header:
            t_target = float(raw_header[0])
            tol = float(raw_header[1])
            tokens = input("Enter readings ending with END: ").split()
            readings = [float(x) for x in tokens if x != "END"]
            print(analyze_thermostat(t_target, tol, readings))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        ((22.0, 2.0, [22.5, 23.0, 25.0, 26.5, 24.5, 22.0, 19.0, 17.5, 21.0]),
         "Cycles: 4 | Max Cooling Streak: 3 | Max Heating Streak: 2 | Severe Anomalies: 2"),
        ((20.0, 1.0, [20.5, 20.2, 20.8]),
         "Cycles: 0 | Max Cooling Streak: 0 | Max Heating Streak: 0 | Severe Anomalies: 0"),
        ((25.0, 3.0, [32.0, 33.0, 25.0, 15.0, 14.0]),
         "Cycles: 2 | Max Cooling Streak: 2 | Max Heating Streak: 2 | Severe Anomalies: 4"),
    ]
    for (target, tol, readings), expected in cases:
        actual = analyze_thermostat(target, tol, readings)
        assert actual == expected, f"Failed: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: target={target}, tol={tol} -> {actual}")

test_problem_5()
````
