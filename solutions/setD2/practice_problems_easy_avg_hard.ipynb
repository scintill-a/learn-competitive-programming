# 🟢 Problem 1: Right Triangle Hypotenuse & Geometry (Easy — Sequential/IO)

### 📖 Problem Statement
Given the lengths of the two legs `a` and `b` of a right triangle on a single line, calculate its hypotenuse `c`, perimeter `P`, and area `A`.

Formulas:
c = √(a² + b²)
Perimeter = a + b + c
Area = 0.5 × a × b

Format all output numbers with exactly 2 decimal places.

### 📥 Input Format
A single line containing two space-separated positive numbers:
`a` (float) `b` (float)

### 📤 Output Format
Print a single line formatted as:
`Hypotenuse: <c:.2f> | Perimeter: <P:.2f> | Area: <A:.2f>`

### 🔒 Constraints
- 0.01 <= a, b <= 10,000.00

### 💡 Example 1
```text
Input:
3.0 4.0

Output:
Hypotenuse: 5.00 | Perimeter: 12.00 | Area: 6.00

Explanation: c = √(9 + 16) = 5.00. Perimeter = 3 + 4 + 5 = 12.00. Area = 0.5 × 3 × 4 = 6.00.
```

### 💡 Example 2
```text
Input:
5.0 12.0

Output:
Hypotenuse: 13.00 | Perimeter: 30.00 | Area: 30.00

Explanation: c = √(25 + 144) = 13.00. Perimeter = 5 + 12 + 13 = 30.00. Area = 0.5 × 5 × 12 = 30.00.
```


````python
# 💡 Problem 1 Solution: Right Triangle Hypotenuse & Geometry
import math

def right_triangle_geometry(a: float, b: float) -> str:
    c = math.sqrt(a**2 + b**2)
    p = a + b + c
    area = 0.5 * a * b
    return f"Hypotenuse: {c:.2f} | Perimeter: {p:.2f} | Area: {area:.2f}"

def solve_problem_1():
    print("--- Running Problem 1 (Right Triangle) ---")
    try:
        raw = input("Enter legs a and b: ").split()
        if raw:
            a, b = float(raw[0]), float(raw[1])
            print(right_triangle_geometry(a, b))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((3.0, 4.0), "Hypotenuse: 5.00 | Perimeter: 12.00 | Area: 6.00"),
        ((5.0, 12.0), "Hypotenuse: 13.00 | Perimeter: 30.00 | Area: 30.00"),
        ((6.0, 8.0), "Hypotenuse: 10.00 | Perimeter: 24.00 | Area: 24.00"),
    ]
    for args, expected in cases:
        actual = right_triangle_geometry(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Road Trip Fuel Cost & Mileage Splitter (Easy — Sequential/IO)

### 📖 Problem Statement
A driver is planning a road trip. Given the total trip distance in miles, the car's fuel efficiency in miles per gallon (MPG), and the fuel price per gallon on a single line:
1. Calculate the total gallons of fuel required (`distance / mpg`).
2. Calculate the total fuel cost (`gallons × price_per_gal`).
3. Calculate the fuel cost per mile (`total_cost / distance`).

Format gallons and all dollar amounts to exactly 2 decimal places.

### 📥 Input Format
A single line containing three space-separated numbers:
`distance_miles` (float) `mpg` (float) `price_per_gal` (float)

### 📤 Output Format
Print a single line formatted as:
`Gallons: <gallons:.2f> | Total Cost: $XX.XX | Cost Per Mile: $XX.XX`

### 🔒 Constraints
- 1.0 <= distance_miles <= 10,000.0
- 5.0 <= mpg <= 100.0
- 0.50 <= price_per_gal <= 20.00

### 💡 Example 1
```text
Input:
300.0 25.0 3.50

Output:
Gallons: 12.00 | Total Cost: $42.00 | Cost Per Mile: $0.14

Explanation: Gallons = 300 / 25 = 12.00. Total Cost = 12 × 3.50 = 42.00. Cost Per Mile = 42 / 300 = 0.14.
```

### 💡 Example 2
```text
Input:
150.0 30.0 4.00

Output:
Gallons: 5.00 | Total Cost: $20.00 | Cost Per Mile: $0.13

Explanation: Gallons = 150 / 30 = 5.00. Total Cost = 5 × 4.00 = 20.00. Cost Per Mile = 20 / 150 = 0.13.
```


````python
# 💡 Problem 2 Solution: Road Trip Fuel Cost & Mileage Splitter

def fuel_cost_calc(distance_miles: float, mpg: float, price_per_gal: float) -> str:
    gallons = distance_miles / mpg
    total_cost = gallons * price_per_gal
    cost_per_mile = total_cost / distance_miles
    return f"Gallons: {gallons:.2f} | Total Cost: ${total_cost:.2f} | Cost Per Mile: ${cost_per_mile:.2f}"

def solve_problem_2():
    print("--- Running Problem 2 (Fuel Cost) ---")
    try:
        raw = input("Enter distance_miles, mpg, price_per_gal: ").split()
        if raw:
            d, m, p = float(raw[0]), float(raw[1]), float(raw[2])
            print(fuel_cost_calc(d, m, p))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        ((300.0, 25.0, 3.50), "Gallons: 12.00 | Total Cost: $42.00 | Cost Per Mile: $0.14"),
        ((150.0, 30.0, 4.00), "Gallons: 5.00 | Total Cost: $20.00 | Cost Per Mile: $0.13"),
        ((500.0, 20.0, 3.00), "Gallons: 25.00 | Total Cost: $75.00 | Cost Per Mile: $0.15"),
    ]
    for args, expected in cases:
        actual = fuel_cost_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Taxi Meter Fare & Surcharge Classifier (Average — Selection)

### 📖 Problem Statement
A taxi service calculates passenger fare based on distance traveled in kilometers and whether the trip took place during the `DAY` or `NIGHT` shift:
- **Base Flag-Drop Fee**: $4.00 flat (covers up to the first 1.0 km).
- **Distance beyond 1.0 km**:
  - If `DAY`: $1.50 per additional kilometer.
  - If `NIGHT`: $2.25 per additional kilometer.

If total distance is less than or equal to 1.0 km, only the $4.00 base fee is charged.

### 📥 Input Format
A single line containing:
`distance_km` (float) `shift` (string: `DAY` or `NIGHT`)

### 📤 Output Format
Print a single line formatted as:
`Shift: <DAY/NIGHT> | Distance: <dist:.2f> km | Total Fare: $XX.XX`

### 🔒 Constraints
- 0.1 <= distance_km <= 1000.0
- shift in {"DAY", "NIGHT"}

### 💡 Example 1
```text
Input:
5.5 DAY

Output:
Shift: DAY | Distance: 5.50 km | Total Fare: $10.75

Explanation: Base 1.0 km = $4.00. Excess 4.5 km × $1.50 = $6.75. Total Fare = $10.75.
```

### 💡 Example 2
```text
Input:
0.8 NIGHT

Output:
Shift: NIGHT | Distance: 0.80 km | Total Fare: $4.00

Explanation: Distance <= 1.0 km, so only the $4.00 flag-drop fee applies.
```


````python
# 💡 Problem 3 Solution: Taxi Meter Fare & Surcharge Classifier

def taxi_fare_calc(distance_km: float, shift: str) -> str:
    if distance_km <= 1.0:
        fare = 4.00
    else:
        excess = distance_km - 1.0
        rate = 1.50 if shift == "DAY" else 2.25
        fare = 4.00 + excess * rate
    return f"Shift: {shift} | Distance: {distance_km:.2f} km | Total Fare: ${fare:.2f}"

def solve_problem_3():
    print("--- Running Problem 3 (Taxi Meter) ---")
    try:
        raw = input("Enter distance_km and shift (DAY/NIGHT): ").split()
        if raw:
            d, s = float(raw[0]), raw[1].upper()
            print(taxi_fare_calc(d, s))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((5.5, "DAY"), "Shift: DAY | Distance: 5.50 km | Total Fare: $10.75"),
        ((0.8, "NIGHT"), "Shift: NIGHT | Distance: 0.80 km | Total Fare: $4.00"),
        ((10.0, "NIGHT"), "Shift: NIGHT | Distance: 10.00 km | Total Fare: $24.25"),
        ((1.0, "DAY"), "Shift: DAY | Distance: 1.00 km | Total Fare: $4.00"),
    ]
    for args, expected in cases:
        actual = taxi_fare_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: 2D Bounding Box Point Classifier (Average — Selection)

### 📖 Problem Statement
An axis-aligned rectangular bounding box is defined by its bottom-left corner `(x1, y1)` and top-right corner `(x2, y2)` with `x1 < x2` and `y1 < y2`. Given a query point `(px, py)`, classify its spatial position relative to the box:
- `INSIDE`: The point is strictly inside (`x1 < px < x2` and `y1 < py < y2`).
- `BORDER`: The point lies on the box boundary (`x1 <= px <= x2` and `y1 <= py <= y2`, and at least one coordinate equals a boundary).
- `OUTSIDE`: The point lies outside the box.

### 📥 Input Format
A single line containing six space-separated numbers:
`x1` `y1` `x2` `y2` `px` `py` (all floats, with x1 < x2 and y1 < y2).

### 📤 Output Format
Print a single line formatted as:
`Position: <INSIDE/BORDER/OUTSIDE>`

### 🔒 Constraints
- -10,000.0 <= coordinates <= 10,000.0, with x1 < x2 and y1 < y2

### 💡 Example 1
```text
Input:
0.0 0.0 10.0 10.0 5.0 5.0

Output:
Position: INSIDE

Explanation: Point (5, 5) is strictly between the coordinates (0, 0) and (10, 10).
```

### 💡 Example 2
```text
Input:
0.0 0.0 10.0 10.0 0.0 5.0

Output:
Position: BORDER

Explanation: px = 0.0 touches the left boundary x1, and py is between y1 and y2.
```


````python
# 💡 Problem 4 Solution: 2D Bounding Box Point Classifier

def point_box_classifier(x1: float, y1: float, x2: float, y2: float, px: float, py: float) -> str:
    if x1 < px < x2 and y1 < py < y2:
        return "Position: INSIDE"
    elif x1 <= px <= x2 and y1 <= py <= y2:
        return "Position: BORDER"
    else:
        return "Position: OUTSIDE"

def solve_problem_4():
    print("--- Running Problem 4 (Bounding Box) ---")
    try:
        raw = input("Enter x1 y1 x2 y2 px py: ").split()
        if raw:
            vals = list(map(float, raw))
            print(point_box_classifier(*vals))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ((0.0, 0.0, 10.0, 10.0, 5.0, 5.0), "Position: INSIDE"),
        ((0.0, 0.0, 10.0, 10.0, 0.0, 5.0), "Position: BORDER"),
        ((0.0, 0.0, 10.0, 10.0, 12.0, 5.0), "Position: OUTSIDE"),
        ((0.0, 0.0, 10.0, 10.0, 10.0, 10.0), "Position: BORDER"),
        ((2.0, 2.0, 8.0, 8.0, 1.0, 5.0), "Position: OUTSIDE"),
    ]
    for args, expected in cases:
        actual = point_box_classifier(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Longest Strictly Increasing Streak (Difficult — Loops)

### 📖 Problem Statement
Given a sequence of space-separated integers on a single line, iterate through the numbers using a loop to find the length of the longest contiguous subsegment (streak) of numbers where each element is strictly greater than the preceding element (`nums[i] > nums[i-1]`). A single element counts as a streak of length 1.

### 📥 Input Format
A single line containing space-separated integers.

### 📤 Output Format
Print a single line formatted as:
`Max Streak: <max_length>`

### 🔒 Constraints
- 1 <= count of numbers <= 100,000
- -1,000,000 <= number <= 1,000,000

### 💡 Example 1
```text
Input:
1 3 5 2 4 6 8 1 2

Output:
Max Streak: 4

Explanation: Streaks are [1, 3, 5] (len 3), [2, 4, 6, 8] (len 4), and [1, 2] (len 2). Max streak is 4.
```

### 💡 Example 2
```text
Input:
5 4 3 2 1

Output:
Max Streak: 1

Explanation: No element is strictly greater than its predecessor. Max streak is 1.
```


````python
# 💡 Problem 5 Solution: Longest Strictly Increasing Streak

def longest_increasing_streak(nums: list) -> str:
    if not nums:
        return "Max Streak: 0"
    max_streak = 1
    curr_streak = 1
    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            curr_streak += 1
            if curr_streak > max_streak:
                max_streak = curr_streak
        else:
            curr_streak = 1
    return f"Max Streak: {max_streak}"

def solve_problem_5():
    print("--- Running Problem 5 (Streak Loop) ---")
    try:
        raw = input("Enter space-separated integers: ").split()
        if raw:
            nums = list(map(int, raw))
            print(longest_increasing_streak(nums))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        (([1, 3, 5, 2, 4, 6, 8, 1, 2],), "Max Streak: 4"),
        (([5, 4, 3, 2, 1],), "Max Streak: 1"),
        (([10, 20, 20, 30, 40],), "Max Streak: 3"),
        (([7],), "Max Streak: 1"),
        (([1, 2, 3, 4, 5],), "Max Streak: 5"),
    ]
    for args, expected in cases:
        actual = longest_increasing_streak(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_5()
````
