# 🟢 Problem 1: Hemisphere Surface Area & Volume (Easy — Sequential/IO)

### 📖 Problem Statement
Given the radius `r` of a solid hemisphere (half of a sphere) on a single line, calculate its total surface area (including the flat circular base) and its volume:

Formulas:
Total Surface Area = 3 × π × r²
Volume = (2 / 3) × π × r³

Use `math.pi` for the value of π and format both results to 2 decimal places.

### 📥 Input Format
A single line containing a single positive number:
`radius` (float)

### 📤 Output Format
Print a single line formatted as:
`Total Surface Area: <area:.2f> | Volume: <vol:.2f>`

### 🔒 Constraints
- 0.01 <= radius <= 10,000.00

### 💡 Example 1
```text
Input:
3.0

Output:
Total Surface Area: 84.82 | Volume: 56.55

Explanation: Area = 3 × π × 9 = 84.82. Volume = (2/3) × π × 27 = 56.55.
```

### 💡 Example 2
```text
Input:
5.0

Output:
Total Surface Area: 235.62 | Volume: 261.80

Explanation: Area = 3 × π × 25 = 235.62. Volume = (2/3) × π × 125 = 261.80.
```


````python
# 💡 Problem 1 Solution: Hemisphere Surface Area & Volume
import math

def hemisphere_metrics_calc(radius: float) -> str:
    area = 3.0 * math.pi * (radius ** 2)
    vol = (2.0 / 3.0) * math.pi * (radius ** 3)
    return f"Total Surface Area: {area:.2f} | Volume: {vol:.2f}"

def solve_problem_1():
    print("--- Running Problem 1 (Hemisphere) ---")
    try:
        raw = input("Enter radius: ").strip()
        if raw:
            r = float(raw)
            print(hemisphere_metrics_calc(r))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((3.0,), "Total Surface Area: 84.82 | Volume: 56.55"),
        ((5.0,), "Total Surface Area: 235.62 | Volume: 261.80"),
        ((1.0,), "Total Surface Area: 9.42 | Volume: 2.09"),
    ]
    for args, expected in cases:
        actual = hemisphere_metrics_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Shared Work Rate & Completion Time (Easy — Sequential/IO)

### 📖 Problem Statement
Two workers A and B can finish an identical project individually in `days_A` and `days_B` days. When working cooperatively at the same time:
1. Calculate the total days required to complete the task together:
   `days_together = (days_A × days_B) / (days_A + days_B)`
2. Calculate the percentage of total work contributed by worker A:
   `worker_A_share = (days_together / days_A) × 100`

Format `days_together` to 2 decimal places and worker A's share percentage to 1 decimal place.

### 📥 Input Format
A single line containing two space-separated positive numbers:
`days_A` (float) `days_B` (float)

### 📤 Output Format
Print a single line formatted as:
`Days Together: <days:.2f> | Worker A Share: <pct:.1f>%`

### 🔒 Constraints
- 0.1 <= days_A, days_B <= 1,000.0

### 💡 Example 1
```text
Input:
6.0 12.0

Output:
Days Together: 4.00 | Worker A Share: 66.7%

Explanation: Days = (6 × 12) / (6 + 12) = 72 / 18 = 4.00. Share = (4 / 6) × 100 = 66.7%.
```

### 💡 Example 2
```text
Input:
10.0 10.0

Output:
Days Together: 5.00 | Worker A Share: 50.0%

Explanation: Equal speed, together takes 5.00 days, worker A does 50.0% of the work.
```


````python
# 💡 Problem 2 Solution: Shared Work Rate & Completion Time

def shared_work_calc(days_a: float, days_b: float) -> str:
    together = (days_a * days_b) / (days_a + days_b)
    share_a = (together / days_a) * 100.0
    return f"Days Together: {together:.2f} | Worker A Share: {share_a:.1f}%"

def solve_problem_2():
    print("--- Running Problem 2 (Work Rate) ---")
    try:
        raw = input("Enter days_A and days_B: ").split()
        if raw:
            da, db = float(raw[0]), float(raw[1])
            print(shared_work_calc(da, db))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        ((6.0, 12.0), "Days Together: 4.00 | Worker A Share: 66.7%"),
        ((10.0, 10.0), "Days Together: 5.00 | Worker A Share: 50.0%"),
        ((4.0, 6.0), "Days Together: 2.40 | Worker A Share: 60.0%"),
    ]
    for args, expected in cases:
        actual = shared_work_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Speeding Violation & Traffic Fine Classifier (Average — Selection)

### 📖 Problem Statement
A highway speed monitoring sensor evaluates a vehicle's speed relative to the posted speed limit:
- `speed <= limit`: `No Violation` | Fine = $0.00
- `1.0 <= excess <= 10.0`: `Minor Violation` | Fine = $50.00
- `10.1 <= excess <= 20.0`: `Moderate Violation` | Fine = $125.00
- `excess > 20.0`: `Reckless Speeding` | Fine = $250.00 + (excess - 20.0) × $10.00

Where `excess = speed - limit`.

### 📥 Input Format
A single line containing two space-separated numbers:
`speed` (float) `limit` (float)

### 📤 Output Format
Print a single line formatted as:
`Status: <status> | Excess: <excess:.1f> mph | Fine: $XX.XX`

### 🔒 Constraints
- 0.0 <= speed <= 300.0
- 10.0 <= limit <= 150.0

### 💡 Example 1
```text
Input:
78.0 65.0

Output:
Status: Moderate Violation | Excess: 13.0 mph | Fine: $125.00

Explanation: Excess = 78 - 65 = 13.0 mph (10.1 to 20.0 range). Fine is $125.00.
```

### 💡 Example 2
```text
Input:
62.0 65.0

Output:
Status: No Violation | Excess: 0.0 mph | Fine: $0.00

Explanation: Speed is within limit. No violation.
```


````python
# 💡 Problem 3 Solution: Speeding Violation & Traffic Fine Classifier

def speeding_fine_calc(speed: float, limit: float) -> str:
    if speed <= limit:
        return "Status: No Violation | Excess: 0.0 mph | Fine: $0.00"
    excess = speed - limit
    if excess <= 10.0:
        status = "Minor Violation"
        fine = 50.00
    elif excess <= 20.0:
        status = "Moderate Violation"
        fine = 125.00
    else:
        status = "Reckless Speeding"
        fine = 250.00 + (excess - 20.0) * 10.00
    return f"Status: {status} | Excess: {excess:.1f} mph | Fine: ${fine:.2f}"

def solve_problem_3():
    print("--- Running Problem 3 (Speed Fine) ---")
    try:
        raw = input("Enter speed and speed limit: ").split()
        if raw:
            s, l = float(raw[0]), float(raw[1])
            print(speeding_fine_calc(s, l))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((62.0, 65.0), "Status: No Violation | Excess: 0.0 mph | Fine: $0.00"),
        ((72.0, 65.0), "Status: Minor Violation | Excess: 7.0 mph | Fine: $50.00"),
        ((78.0, 65.0), "Status: Moderate Violation | Excess: 13.0 mph | Fine: $125.00"),
        ((95.0, 65.0), "Status: Reckless Speeding | Excess: 30.0 mph | Fine: $350.00"),
        ((65.0, 65.0), "Status: No Violation | Excess: 0.0 mph | Fine: $0.00"),
    ]
    for args, expected in cases:
        actual = speeding_fine_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: 2D Points Collinearity & Triangle Area (Average — Selection)

### 📖 Problem Statement
Given the 2D Cartesian coordinates of three distinct points `(x1, y1)`, `(x2, y2)`, and `(x3, y3)` on a single line, calculate the area formed by these points using the coordinate Shoelace formula:

Area = 0.5 × |x1 × (y2 - y3) + x2 × (y3 - y1) + x3 × (y1 - y2)|

Classify the geometry:
- If `Area < 0.001`: The three points lie on a single straight line. Output `Points are Collinear`.
- Otherwise: Output `Valid Triangle | Area: <area:.2f>`.

### 📥 Input Format
A single line containing six space-separated numbers:
`x1` `y1` `x2` `y2` `x3` `y3` (all floats)

### 📤 Output Format
If collinear: `Points are Collinear`
Otherwise: `Valid Triangle | Area: <area:.2f>`

### 🔒 Constraints
- -10,000.0 <= coordinates <= 10,000.0

### 💡 Example 1
```text
Input:
0.0 0.0 4.0 0.0 0.0 3.0

Output:
Valid Triangle | Area: 6.00

Explanation: Base = 4.0, height = 3.0. Triangle area = 0.5 × 4 × 3 = 6.00.
```

### 💡 Example 2
```text
Input:
1.0 1.0 2.0 2.0 3.0 3.0

Output:
Points are Collinear

Explanation: All three points lie along the line y = x. Area is 0.00.
```


````python
# 💡 Problem 4 Solution: 2D Points Collinearity & Triangle Area

def collinearity_triangle_calc(x1: float, y1: float, x2: float, y2: float, x3: float, y3: float) -> str:
    cross = x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)
    area = 0.5 * abs(cross)
    if area < 1e-3:
        return "Points are Collinear"
    return f"Valid Triangle | Area: {area:.2f}"

def solve_problem_4():
    print("--- Running Problem 4 (Collinearity) ---")
    try:
        raw = input("Enter x1 y1 x2 y2 x3 y3: ").split()
        if raw:
            vals = list(map(float, raw))
            print(collinearity_triangle_calc(*vals))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ((0.0, 0.0, 4.0, 0.0, 0.0, 3.0), "Valid Triangle | Area: 6.00"),
        ((1.0, 1.0, 2.0, 2.0, 3.0, 3.0), "Points are Collinear"),
        ((0.0, 0.0, 5.0, 0.0, 2.0, 4.0), "Valid Triangle | Area: 10.00"),
        ((0.0, 0.0, 0.0, 5.0, 0.0, 10.0), "Points are Collinear"),
    ]
    for args, expected in cases:
        actual = collinearity_triangle_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Prime Factor Decomposition & Largest Factor (Difficult — Loops)

### 📖 Problem Statement
Given an integer `n >= 2`, compute its prime factorization using trial division inside a loop.

Count the total number of prime factors (counting duplicates / multiplicity), and determine the largest prime factor of `n`.

### 📥 Input Format
A single line containing an integer:
`n`

### 📤 Output Format
Print a single line formatted as:
`Factor Count: <count> | Largest Factor: <largest>`

### 🔒 Constraints
- 2 <= n <= 10^12

### 💡 Example 1
```text
Input:
60

Output:
Factor Count: 4 | Largest Factor: 5

Explanation: 60 = 2 × 2 × 3 × 5. There are 4 prime factors, and the largest is 5.
```

### 💡 Example 2
```text
Input:
13

Output:
Factor Count: 1 | Largest Factor: 13

Explanation: 13 is prime. There is 1 factor, and the largest is 13.
```


````python
# 💡 Problem 5 Solution: Prime Factor Decomposition & Largest Factor

def prime_factors_tracker(n: int) -> str:
    count = 0
    largest = 0
    curr = n
    d = 2
    while d * d <= curr:
        while curr % d == 0:
            count += 1
            largest = d
            curr //= d
        d += 1
    if curr > 1:
        count += 1
        largest = curr
    return f"Factor Count: {count} | Largest Factor: {largest}"

def solve_problem_5():
    print("--- Running Problem 5 (Prime Factors) ---")
    try:
        raw = input("Enter integer n (>= 2): ").strip()
        if raw:
            print(prime_factors_tracker(int(raw)))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        ((60,), "Factor Count: 4 | Largest Factor: 5"),
        ((13,), "Factor Count: 1 | Largest Factor: 13"),
        ((100,), "Factor Count: 4 | Largest Factor: 5"),
        ((84,), "Factor Count: 4 | Largest Factor: 7"),
        ((2,), "Factor Count: 1 | Largest Factor: 2"),
    ]
    for args, expected in cases:
        actual = prime_factors_tracker(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_5()
````
