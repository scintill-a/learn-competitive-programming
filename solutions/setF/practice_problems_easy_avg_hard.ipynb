# 🟢 Problem 1: Find Pivot Equilibrium Index (Easy — Prefix Accumulation)

### 📖 Problem Statement
Given an array of integers `nums`, find the leftmost 0-based pivot index `i` where the sum of all numbers strictly to the left of index `i` equals the sum of all numbers strictly to its right:

Condition:
left_sum == total_sum - left_sum - nums[i]

If index is at the left edge (`i == 0`), left sum is 0. If index is at the right edge, right sum is 0. If no pivot index exists, return -1. Compute in O(N) time and O(1) space.

### 📥 Input Format
A single line containing space-separated integers.

### 📤 Output Format
If pivot exists: `Pivot Index: <idx> | Value: <val>`
If none exists: `Pivot Index: -1`

### 🔒 Constraints
- 1 <= len(nums) <= 100,000
- -1,000 <= nums[i] <= 1,000

### 💡 Example 1
```text
Input:
1 7 3 6 5 6

Output:
Pivot Index: 3 | Value: 6

Explanation: At index 3 (value 6), left sum = 1+7+3 = 11, right sum = 5+6 = 11.
```

### 💡 Example 2
```text
Input:
1 2 3

Output:
Pivot Index: -1

Explanation: No index satisfies the equilibrium condition.
```


````python
# 💡 Problem 1 Solution: Find Pivot Equilibrium Index

def find_pivot_index(nums: list) -> str:
    total = sum(nums)
    left = 0
    for i, x in enumerate(nums):
        if left == total - left - x:
            return f"Pivot Index: {i} | Value: {x}"
        left += x
    return "Pivot Index: -1"

def solve_problem_1():
    print("--- Running Problem 1 (Equilibrium Pivot) ---")
    try:
        nums = list(map(int, input("Enter space-separated integers: ").split()))
        print(find_pivot_index(nums))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        (([1, 7, 3, 6, 5, 6],), "Pivot Index: 3 | Value: 6"),
        (([1, 2, 3],), "Pivot Index: -1"),
        (([2, 1, -1],), "Pivot Index: 0 | Value: 2"),
    ]
    for args, expected in cases:
        actual = find_pivot_index(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Valid Palindrome with At Most One Removal (Easy — Two Pointers)

### 📖 Problem Statement
Given a lowercase string `s`, determine whether it can be made into a palindrome by deleting **at most one** character from it.

Use the two pointers method starting from both ends:
- Move inward while `s[left] == s[right]`.
- On the first mismatch, test whether deleting `s[left]` OR deleting `s[right]` yields a valid palindrome.
- Solve in strict O(N) time and O(1) extra space.

### 📥 Input Format
A single line containing the string `s`.

### 📤 Output Format
If valid: `Valid: True | Deleted: <char/None>`
If invalid: `Valid: False`

### 🔒 Constraints
- 1 <= len(s) <= 100,000

### 💡 Example 1
```text
Input:
abca

Output:
Valid: True | Deleted: c

Explanation: Deleting 'c' leaves "aba" which is a palindrome.
```

### 💡 Example 2
```text
Input:
racecar

Output:
Valid: True | Deleted: None

Explanation: Already a palindrome; 0 deletions needed.
```


````python
# 💡 Problem 2 Solution: Valid Palindrome with At Most One Removal

def valid_palindrome_one_removal(s: str) -> str:
    def is_pal(sub):
        return sub == sub[::-1]

    l, r = 0, len(s) - 1
    while l < r:
        if s[l] == s[r]:
            l += 1
            r -= 1
        else:
            if is_pal(s[l + 1:r + 1]):
                return f"Valid: True | Deleted: {s[l]}"
            elif is_pal(s[l:r]):
                return f"Valid: True | Deleted: {s[r]}"
            else:
                return "Valid: False"
    return "Valid: True | Deleted: None"

def solve_problem_2():
    print("--- Running Problem 2 (Palindrome Tolerance) ---")
    try:
        s = input("Enter string: ").strip()
        print(valid_palindrome_one_removal(s))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        (("racecar",), "Valid: True | Deleted: None"),
        (("abca",), ["Valid: True | Deleted: c", "Valid: True | Deleted: b"]),
        (("abc",), "Valid: False"),
    ]
    for args, expected in cases:
        actual = valid_palindrome_one_removal(*args)
        if isinstance(expected, list):
            assert actual in expected, f"Failed for {args}: expected one of {expected}, got '{actual}'"
        else:
            assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟢 Problem 3: Interval Overlap & Peak Concurrent Meetings (Easy — Greedy Sweep)

### 📖 Problem Statement
Given an integer `N` on Line 1 followed by `N` space-separated meeting intervals `[start, end]` on subsequent lines:
Find the maximum number of concurrent meetings active at the exact same moment (i.e. the minimum number of meeting rooms required).

Sort start and end times separately, then sweep through them using two pointers in O(N log N) time:
- If `start[i] < end[j]`: a meeting starts before the earliest finishes; increment room count and `i`.
- Otherwise: a meeting has concluded; decrement room count and `j`.

### 📥 Input Format
- Line 1: Integer `N`
- Next N lines: `start` `end` (integers)

### 📤 Output Format
Print a single line formatted as:
`Peak Concurrent: <count>`

### 🔒 Constraints
- 1 <= N <= 50,000
- 0 <= start < end <= 1,000,000

### 💡 Example 1
```text
Input:
3
0 30
5 10
15 20

Output:
Peak Concurrent: 2

Explanation: At time t=6, meeting [0, 30] and meeting [5, 10] are both active (2 rooms).
```

### 💡 Example 2
```text
Input:
2
7 10
2 4

Output:
Peak Concurrent: 1

Explanation: [2, 4] finishes completely before [7, 10] begins; 1 room suffices.
```


````python
# 💡 Problem 3 Solution: Interval Overlap & Peak Concurrent Meetings

def peak_concurrent_meetings(intervals: list) -> str:
    starts = sorted([iv[0] for iv in intervals])
    ends = sorted([iv[1] for iv in intervals])
    max_c = cur_c = 0
    i = j = 0
    n = len(intervals)
    while i < n and j < n:
        if starts[i] < ends[j]:
            cur_c += 1
            if cur_c > max_c:
                max_c = cur_c
            i += 1
        else:
            cur_c -= 1
            j += 1
    return f"Peak Concurrent: {max_c}"

def solve_problem_3():
    print("--- Running Problem 3 (Sweep Line Intervals) ---")
    try:
        n = int(input("Enter number of intervals N: ").strip())
        ivs = []
        for _ in range(n):
            s, e = map(int, input().split())
            ivs.append([s, e])
        print(peak_concurrent_meetings(ivs))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        (([[0, 30], [5, 10], [15, 20]],), "Peak Concurrent: 2"),
        (([[7, 10], [2, 4]],), "Peak Concurrent: 1"),
        (([[1, 4], [2, 5], [3, 6]],), "Peak Concurrent: 3"),
    ]
    for args, expected in cases:
        actual = peak_concurrent_meetings(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Find All Anagrams in String (Medium — Fixed Sliding Window & Frequency)

### 📖 Problem Statement
Given two lowercase strings `s` and `p`:
Find all 0-based start indices of `p`'s anagrams in `s`. An anagram is a permutation of the characters.

You must solve this in O(len(s)) time using a **Fixed Sliding Window of size len(p)**:
- Maintain character frequency counts of `p` and of the current window in `s`.
- Slide the window one character at a time, incrementing the new character's count and decrementing the exiting character's count.
- If the frequency profiles match, record the starting index.

### 📥 Input Format
- Line 1: String `s`
- Line 2: String `p`

### 📤 Output Format
If matches found: `Anagram Indices: <idx1>, <idx2>, ...`
If none found: `Anagram Indices: None`

### 🔒 Constraints
- 1 <= len(p) <= len(s) <= 100,000

### 💡 Example 1
```text
Input:
cbaebabacd
abc

Output:
Anagram Indices: 0, 6

Explanation: Substring at index 0 is "cba" (anagram of "abc"); at index 6 is "bac" (anagram of "abc").
```

### 💡 Example 2
```text
Input:
abab
ab

Output:
Anagram Indices: 0, 1, 2

Explanation: "ab" at 0, "ba" at 1, "ab" at 2.
```


````python
# 💡 Problem 4 Solution: Find All Anagrams in String
from collections import Counter

def find_all_anagrams(s: str, p: str) -> str:
    if len(s) < len(p):
        return "Anagram Indices: None"
    p_cnt = Counter(p)
    w_cnt = Counter(s[:len(p)])
    res = []
    if w_cnt == p_cnt:
        res.append(0)
    k = len(p)
    for i in range(k, len(s)):
        w_cnt[s[i]] += 1
        w_cnt[s[i - k]] -= 1
        if w_cnt[s[i - k]] == 0:
            del w_cnt[s[i - k]]
        if w_cnt == p_cnt:
            res.append(i - k + 1)
    if not res:
        return "Anagram Indices: None"
    idx_str = ", ".join(map(str, res))
    return f"Anagram Indices: {idx_str}"

def solve_problem_4():
    print("--- Running Problem 4 (Fixed Anagram Sliding Window) ---")
    try:
        s = input("Enter string s: ").strip()
        p = input("Enter pattern p: ").strip()
        print(find_all_anagrams(s, p))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        (("cbaebabacd", "abc"), "Anagram Indices: 0, 6"),
        (("abab", "ab"), "Anagram Indices: 0, 1, 2"),
        (("hello", "xyz"), "Anagram Indices: None"),
    ]
    for args, expected in cases:
        actual = find_all_anagrams(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🟡 Problem 5: 3Sum Zero Target Combinations (Medium — Two Pointers & Sort)

### 📖 Problem Statement
Given an integer array `nums`, find all unique triplets `[nums[i], nums[j], nums[k]]` such that `i ≠ j ≠ k` and `nums[i] + nums[j] + nums[k] == 0`.
The output list must not contain duplicate triplets.

Solve in O(N²) time:
1. Sort the array.
2. Iterate `i` from `0` to `N - 3`. If `nums[i] == nums[i-1]`, skip to avoid duplicates.
3. For each `i`, initialize two pointers `left = i + 1` and `right = N - 1` to find pairs that sum to `-nums[i]`. Skip duplicate values of `left` and `right`.

### 📥 Input Format
A single line containing space-separated integers.

### 📤 Output Format
If found: `Triplets: [a, b, c], [d, e, f]`
If none: `Triplets: None`

### 🔒 Constraints
- 3 <= len(nums) <= 3,000
- -100,000 <= nums[i] <= 100,000

### 💡 Example 1
```text
Input:
-1 0 1 2 -1 -4

Output:
Triplets: [-1, -1, 2], [-1, 0, 1]

Explanation: The distinct triplets summing to zero are [-1, -1, 2] and [-1, 0, 1].
```

### 💡 Example 2
```text
Input:
0 1 1

Output:
Triplets: None

Explanation: No triplet sums to 0.
```


````python
# 💡 Problem 5 Solution: 3Sum Zero Target Combinations

def three_sum_zero(nums: list) -> str:
    nums.sort()
    res = []
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue
        l, r = i + 1, len(nums) - 1
        while l < r:
            s = nums[i] + nums[l] + nums[r]
            if s == 0:
                res.append([nums[i], nums[l], nums[r]])
                while l < r and nums[l] == nums[l + 1]:
                    l += 1
                while l < r and nums[r] == nums[r - 1]:
                    r -= 1
                l += 1
                r -= 1
            elif s < 0:
                l += 1
            else:
                r -= 1
    if not res:
        return "Triplets: None"
    trip_str = ", ".join(str(t) for t in res)
    return f"Triplets: {trip_str}"

def solve_problem_5():
    print("--- Running Problem 5 (3Sum Search) ---")
    try:
        nums = list(map(int, input("Enter space-separated integers: ").split()))
        print(three_sum_zero(nums))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        (([-1, 0, 1, 2, -1, -4],), "Triplets: [-1, -1, 2], [-1, 0, 1]"),
        (([0, 1, 1],), "Triplets: None"),
        (([0, 0, 0],), "Triplets: [0, 0, 0]"),
    ]
    for args, expected in cases:
        actual = three_sum_zero(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_5()
````

# 🟡 Problem 6: Find Peak Element in O(log N) (Medium — Binary Search)

### 📖 Problem Statement
An element in an array is defined as a peak element if its value is strictly greater than its immediate neighbors (`nums[i] > nums[i-1]` and `nums[i] > nums[i+1]`, assuming virtual -∞ beyond array boundaries).

Find the 0-based index and value of any peak element in strictly **O(log N)** time using Binary Search on the discrete slope:
- If `nums[mid] < nums[mid + 1]`: an ascending slope guarantees a peak exists in the right half (`low = mid + 1`).
- Otherwise: a peak must exist at `mid` or in the left half (`high = mid`).

### 📥 Input Format
A single line containing space-separated integers where no two adjacent elements are equal.

### 📤 Output Format
Print a single line formatted as:
`Peak Index: <idx> | Value: <val>`

### 🔒 Constraints
- 1 <= len(nums) <= 100,000
- -1,000,000 <= nums[i] <= 1,000,000

### 💡 Example 1
```text
Input:
1 2 3 1

Output:
Peak Index: 2 | Value: 3

Explanation: Index 2 with value 3 is strictly greater than neighbors 2 and 1.
```

### 💡 Example 2
```text
Input:
1 2 1 3 5 6 4

Output:
Peak Index: 5 | Value: 6

Explanation: 6 is strictly greater than 5 and 4. (Index 1 is also a valid peak).
```


````python
# 💡 Problem 6 Solution: Find Peak Element in O(log N)

def find_peak_element(nums: list) -> str:
    low, high = 0, len(nums) - 1
    while low < high:
        mid = (low + high) // 2
        if nums[mid] < nums[mid + 1]:
            low = mid + 1
        else:
            high = mid
    return f"Peak Index: {low} | Value: {nums[low]}"

def solve_problem_6():
    print("--- Running Problem 6 (Logarithmic Peak Search) ---")
    try:
        nums = list(map(int, input("Enter space-separated integers: ").split()))
        print(find_peak_element(nums))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_6()
````

````python
# 🧪 Problem 6 Test Suite
def test_problem_6():
    cases = [
        (([1, 2, 3, 1],), "Peak Index: 2 | Value: 3"),
        (([1, 2],), "Peak Index: 1 | Value: 2"),
        (([5],), "Peak Index: 0 | Value: 5"),
    ]
    for args, expected in cases:
        actual = find_peak_element(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_6()
````

# 🔴 Problem 7: Longest Substring with At Most K Distinct Characters (Hard — Sliding Window & Dict)

### 📖 Problem Statement
Given a string `s` on Line 1 and a non-negative integer `K` on Line 2:
Find the length of the longest contiguous substring that contains **at most K distinct characters**.

Implement the **Variable Sliding Window with Frequency Map** pattern in O(N) time:
- Maintain window character counts in a hash map.
- Expand `right`. When `len(map) > K`, increment `left` and decrement character counts until `len(map) <= K`.
- Track the maximum window size and output the first substring achieving that length.

### 📥 Input Format
- Line 1: String `s`
- Line 2: Integer `K`

### 📤 Output Format
Print a single line formatted as:
`Max Length: <len> | Substring: <sub>`

### 🔒 Constraints
- 0 <= K <= 100,000
- 0 <= len(s) <= 100,000

### 💡 Example 1
```text
Input:
eceba
2

Output:
Max Length: 3 | Substring: ece

Explanation: "ece" has length 3 and contains 2 distinct characters ('e' and 'c').
```

### 💡 Example 2
```text
Input:
abaccc
2

Output:
Max Length: 4 | Substring: accc

Explanation: "accc" contains 2 distinct characters ('a' and 'c') and has length 4.
```


````python
# 💡 Problem 7 Solution: Longest Substring with At Most K Distinct Characters
from collections import Counter

def longest_k_distinct_substring(s: str, k: int) -> str:
    if k == 0 or not s:
        return "Max Length: 0 | Substring: "
    counts = Counter()
    l = 0
    max_len = 0
    best_sub = ""
    for r, ch in enumerate(s):
        counts[ch] += 1
        while len(counts) > k:
            left_ch = s[l]
            counts[left_ch] -= 1
            if counts[left_ch] == 0:
                del counts[left_ch]
            l += 1
        if (r - l + 1) > max_len:
            max_len = r - l + 1
            best_sub = s[l:r + 1]
    return f"Max Length: {max_len} | Substring: {best_sub}"

def solve_problem_7():
    print("--- Running Problem 7 (K Distinct Sliding Window) ---")
    try:
        s = input("Enter string: ").strip()
        k = int(input("Enter K: ").strip())
        print(longest_k_distinct_substring(s, k))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_7()
````

````python
# 🧪 Problem 7 Test Suite
def test_problem_7():
    cases = [
        (("eceba", 2), "Max Length: 3 | Substring: ece"),
        (("aa", 1), "Max Length: 2 | Substring: aa"),
        (("abaccc", 2), "Max Length: 4 | Substring: accc"),
        (("", 2), "Max Length: 0 | Substring: "),
    ]
    for args, expected in cases:
        actual = longest_k_distinct_substring(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_7()
````

# 🔴 Problem 8: Sliding Window Maximum (Hard — Monotonic Deque)

### 📖 Problem Statement
Given an array of integers on Line 1 and a sliding window size `K` on Line 2:
A window of size `K` slides from left to right along the array. At each step, output the maximum value visible in the window.

You must achieve strict **O(N) total runtime** using a **Monotonic Decreasing Deque** storing array indices:
- Before inserting index `i`, pop smaller or equal elements from the back of the deque.
- Remove indices that fall outside the current window (`deque[0] < i - K + 1`) from the front.
- For all `i >= K - 1`, the maximum element is at `deque[0]`.

### 📥 Input Format
- Line 1: Space-separated integers.
- Line 2: Window size `K`.

### 📤 Output Format
Print a single line formatted as:
`Window Maxima: <max1> <max2> ...`

### 🔒 Constraints
- 1 <= K <= len(nums) <= 100,000
- -10,000 <= nums[i] <= 10,000

### 💡 Example 1
```text
Input:
1 3 -1 -3 5 3 6 7
3

Output:
Window Maxima: 3 3 5 5 6 7

Explanation: Windows: [1,3,-1]->3, [3,-1,-3]->3, [-1,-3,5]->5, [-3,5,3]->5, [5,3,6]->6, [3,6,7]->7.
```

### 💡 Example 2
```text
Input:
9 11 8 5 7 10
3

Output:
Window Maxima: 11 11 8 10
```


````python
# 💡 Problem 8 Solution: Sliding Window Maximum
from collections import deque

def sliding_window_maximum(nums: list, k: int) -> str:
    dq = deque()
    res = []
    for i, x in enumerate(nums):
        while dq and dq[0] < i - k + 1:
            dq.popleft()
        while dq and nums[dq[-1]] <= x:
            dq.pop()
        dq.append(i)
        if i >= k - 1:
            res.append(nums[dq[0]])
    max_str = " ".join(map(str, res))
    return f"Window Maxima: {max_str}"

def solve_problem_8():
    print("--- Running Problem 8 (Monotonic Deque Window Maximum) ---")
    try:
        nums = list(map(int, input("Enter array: ").split()))
        k = int(input("Enter window size K: ").strip())
        print(sliding_window_maximum(nums, k))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_8()
````

````python
# 🧪 Problem 8 Test Suite
def test_problem_8():
    cases = [
        (([1, 3, -1, -3, 5, 3, 6, 7], 3), "Window Maxima: 3 3 5 5 6 7"),
        (([1], 1), "Window Maxima: 1"),
        (([9, 11, 8, 5, 7, 10], 3), "Window Maxima: 11 11 8 10"),
    ]
    for args, expected in cases:
        actual = sliding_window_maximum(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_8()
````

# 🔴 Problem 9: Trapping Rain Water (Hard — Two Pointers / Elevation Map)

### 📖 Problem Statement
Given `N` non-negative integers representing an elevation terrain map where the width of each bar is 1:
Compute the total units of water trapped after a rainfall.

Solve in optimal **O(N) time and O(1) space** using Two Pointers converging inward:
- Maintain `left_max` and `right_max`.
- If `height[left] < height[right]`: water at `left` is determined by `left_max - height[left]`; advance `left`.
- Otherwise: water at `right` is determined by `right_max - height[right]`; advance `right`.

### 📥 Input Format
A single line containing space-separated non-negative integers.

### 📤 Output Format
Print a single line formatted as:
`Trapped Water: <units> units`

### 🔒 Constraints
- 1 <= len(height) <= 100,000
- 0 <= height[i] <= 100,000

### 💡 Example 1
```text
Input:
0 1 0 2 1 0 1 3 2 1 2 1

Output:
Trapped Water: 6 units

Explanation: Water is trapped at elevation valleys: total = 1 + 1 + 2 + 1 + 1 = 6 units.
```

### 💡 Example 2
```text
Input:
4 2 0 3 2 5

Output:
Trapped Water: 9 units

Explanation: Water trapped between heights: 2 + 4 + 1 + 2 = 9 units.
```


````python
# 💡 Problem 9 Solution: Trapping Rain Water

def trap_rain_water(height: list) -> str:
    l, r = 0, len(height) - 1
    left_max = right_max = 0
    water = 0
    while l < r:
        if height[l] < height[r]:
            if height[l] >= left_max:
                left_max = height[l]
            else:
                water += left_max - height[l]
            l += 1
        else:
            if height[r] >= right_max:
                right_max = height[r]
            else:
                water += right_max - height[r]
            r -= 1
    return f"Trapped Water: {water} units"

def solve_problem_9():
    print("--- Running Problem 9 (Trapping Rain Water) ---")
    try:
        height = list(map(int, input("Enter elevation heights: ").split()))
        print(trap_rain_water(height))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_9()
````

````python
# 🧪 Problem 9 Test Suite
def test_problem_9():
    cases = [
        (([0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1],), "Trapped Water: 6 units"),
        (([4, 2, 0, 3, 2, 5],), "Trapped Water: 9 units"),
        (([3, 0, 2, 0, 4],), "Trapped Water: 7 units"),
    ]
    for args, expected in cases:
        actual = trap_rain_water(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_9()
````
