# 🟢 Problem 1: Kinetic Energy & Momentum Calculator (Easy — Sequential/IO)

### 📖 Problem Statement
Given the mass `m` in kilograms and the velocity `v` in meters per second of an object moving in a straight line, calculate both its linear momentum and kinetic energy:

Formulas:
Momentum = m × v (kg·m/s)
Kinetic Energy = 0.5 × m × v² (Joules)

Display both calculated values formatted to exactly 2 decimal places.

### 📥 Input Format
A single line containing two space-separated positive numbers:
`mass_kg` (float) `velocity_mps` (float)

### 📤 Output Format
Print a single line formatted as:
`Momentum: <p:.2f> kg*m/s | Kinetic Energy: <ke:.2f> J`

### 🔒 Constraints
- 0.01 <= mass_kg <= 100,000.00
- 0.00 <= velocity_mps <= 10,000.00

### 💡 Example 1
```text
Input:
10.0 5.0

Output:
Momentum: 50.00 kg*m/s | Kinetic Energy: 125.00 J

Explanation: Momentum = 10.0 × 5.0 = 50.00 kg*m/s. Kinetic Energy = 0.5 × 10.0 × 25 = 125.00 J.
```

### 💡 Example 2
```text
Input:
2.5 12.0

Output:
Momentum: 30.00 kg*m/s | Kinetic Energy: 180.00 J

Explanation: Momentum = 2.5 × 12.0 = 30.00 kg*m/s. Kinetic Energy = 0.5 × 2.5 × 144 = 180.00 J.
```


````python
# 💡 Problem 1 Solution: Kinetic Energy & Momentum Calculator

def kinetic_momentum_calc(mass_kg: float, velocity_mps: float) -> str:
    momentum = mass_kg * velocity_mps
    ke = 0.5 * mass_kg * (velocity_mps ** 2)
    return f"Momentum: {momentum:.2f} kg*m/s | Kinetic Energy: {ke:.2f} J"

def solve_problem_1():
    print("--- Running Problem 1 (Physics Math) ---")
    try:
        raw = input("Enter mass (kg) and velocity (m/s): ").split()
        if raw:
            m, v = float(raw[0]), float(raw[1])
            print(kinetic_momentum_calc(m, v))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((10.0, 5.0), "Momentum: 50.00 kg*m/s | Kinetic Energy: 125.00 J"),
        ((2.5, 12.0), "Momentum: 30.00 kg*m/s | Kinetic Energy: 180.00 J"),
        ((1.0, 0.0), "Momentum: 0.00 kg*m/s | Kinetic Energy: 0.00 J"),
    ]
    for args, expected in cases:
        actual = kinetic_momentum_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Circle Sector Arc Length & Area (Easy — Sequential/IO)

### 📖 Problem Statement
Given the radius `r` of a circle and a central angle `theta` in degrees on a single line, calculate the arc length and the area of the corresponding sector:

Formulas:
Arc Length = 2 × π × r × (theta / 360)
Sector Area = π × r² × (theta / 360)

Use `math.pi` for the value of π and format both results to 2 decimal places.

### 📥 Input Format
A single line containing two space-separated positive numbers:
`radius` (float) `theta_deg` (float)

### 📤 Output Format
Print a single line formatted as:
`Arc Length: <arc:.2f> | Sector Area: <area:.2f>`

### 🔒 Constraints
- 0.01 <= radius <= 10,000.00
- 0.01 <= theta_deg <= 360.00

### 💡 Example 1
```text
Input:
5.0 90.0

Output:
Arc Length: 7.85 | Sector Area: 19.63

Explanation: Arc = 2 × π × 5 × (90/360) = 7.85. Area = π × 25 × (90/360) = 19.63.
```

### 💡 Example 2
```text
Input:
10.0 60.0

Output:
Arc Length: 10.47 | Sector Area: 52.36

Explanation: Arc = 2 × π × 10 × (1/6) = 10.47. Area = π × 100 × (1/6) = 52.36.
```


````python
# 💡 Problem 2 Solution: Circle Sector Arc Length & Area
import math

def sector_metrics_calc(radius: float, theta_deg: float) -> str:
    arc = 2.0 * math.pi * radius * (theta_deg / 360.0)
    area = math.pi * (radius ** 2) * (theta_deg / 360.0)
    return f"Arc Length: {arc:.2f} | Sector Area: {area:.2f}"

def solve_problem_2():
    print("--- Running Problem 2 (Geometry) ---")
    try:
        raw = input("Enter radius and theta_deg: ").split()
        if raw:
            r, t = float(raw[0]), float(raw[1])
            print(sector_metrics_calc(r, t))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        ((5.0, 90.0), "Arc Length: 7.85 | Sector Area: 19.63"),
        ((10.0, 60.0), "Arc Length: 10.47 | Sector Area: 52.36"),
        ((7.0, 180.0), "Arc Length: 21.99 | Sector Area: 76.97"),
    ]
    for args, expected in cases:
        actual = sector_metrics_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Air Quality Index Category Classifier (Average — Selection)

### 📖 Problem Statement
Given an integer Air Quality Index (AQI) value, classify the public health concern category according to standard environmental criteria:

- 0 to 50: `Good`
- 51 to 100: `Moderate`
- 101 to 150: `Unhealthy for Sensitive Groups`
- 151 to 200: `Unhealthy`
- 201 to 300: `Very Unhealthy`
- 301 to 500: `Hazardous`

If the AQI value is negative or strictly exceeds 500, output `Invalid AQI`.

### 📥 Input Format
A single line containing an integer:
`aqi_score`

### 📤 Output Format
If invalid: `Invalid AQI`
Otherwise: `AQI: <score> | Status: <Category>`

### 🔒 Constraints
- -1,000 <= aqi_score <= 1,000

### 💡 Example 1
```text
Input:
45

Output:
AQI: 45 | Status: Good

Explanation: 45 falls into the 0-50 Good range.
```

### 💡 Example 2
```text
Input:
135

Output:
AQI: 135 | Status: Unhealthy for Sensitive Groups

Explanation: 135 falls into the 101-150 range.
```


````python
# 💡 Problem 3 Solution: Air Quality Index Category Classifier

def aqi_classifier(score: int) -> str:
    if score < 0 or score > 500:
        return "Invalid AQI"
    if score <= 50:
        status = "Good"
    elif score <= 100:
        status = "Moderate"
    elif score <= 150:
        status = "Unhealthy for Sensitive Groups"
    elif score <= 200:
        status = "Unhealthy"
    elif score <= 300:
        status = "Very Unhealthy"
    else:
        status = "Hazardous"
    return f"AQI: {score} | Status: {status}"

def solve_problem_3():
    print("--- Running Problem 3 (AQI Classifier) ---")
    try:
        raw = input("Enter AQI score: ").strip()
        if raw:
            print(aqi_classifier(int(raw)))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((45,), "AQI: 45 | Status: Good"),
        ((135,), "AQI: 135 | Status: Unhealthy for Sensitive Groups"),
        ((220,), "AQI: 220 | Status: Very Unhealthy"),
        ((400,), "AQI: 400 | Status: Hazardous"),
        ((550,), "Invalid AQI"),
        ((-10,), "Invalid AQI"),
    ]
    for args, expected in cases:
        actual = aqi_classifier(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Triangle Classification by Angles (Average — Selection)

### 📖 Problem Statement
Given three positive angles `A`, `B`, and `C` (floats in degrees) of a candidate triangle on a single line:

1. **Validity Check**: All three angles must be strictly positive (`A > 0`, `B > 0`, `C > 0`) and their sum must equal 180 degrees (`abs((A + B + C) - 180.0) <= 0.001`). If this condition is not met, output `Invalid Triangle`.
2. **Angle Classification**: If valid, identify the largest angle `max(A, B, C)`:
   - If largest angle < 90.0: `Acute`
   - If largest angle == 90.0: `Right`
   - If largest angle > 90.0: `Obtuse`

### 📥 Input Format
A single line containing three space-separated numbers:
`A` (float) `B` (float) `C` (float)

### 📤 Output Format
If invalid: `Invalid Triangle`
Otherwise: `Type: <Acute/Right/Obtuse> | Max Angle: <max_angle:.1f>°`

### 🔒 Constraints
- -100.0 <= A, B, C <= 360.0

### 💡 Example 1
```text
Input:
60.0 60.0 60.0

Output:
Type: Acute | Max Angle: 60.0°

Explanation: Sum = 180.0. Max angle is 60.0 < 90, so Acute.
```

### 💡 Example 2
```text
Input:
90.0 45.0 45.0

Output:
Type: Right | Max Angle: 90.0°

Explanation: Sum = 180.0. Largest angle is exactly 90.0, so Right triangle.
```


````python
# 💡 Problem 4 Solution: Triangle Classification by Angles

def triangle_angles_classifier(a: float, b: float, c: float) -> str:
    if a <= 0 or b <= 0 or c <= 0 or abs((a + b + c) - 180.0) > 1e-3:
        return "Invalid Triangle"
    max_ang = max(a, b, c)
    if abs(max_ang - 90.0) < 1e-3:
        kind = "Right"
    elif max_ang > 90.0:
        kind = "Obtuse"
    else:
        kind = "Acute"
    return f"Type: {kind} | Max Angle: {max_ang:.1f}°"

def solve_problem_4():
    print("--- Running Problem 4 (Triangle Angles) ---")
    try:
        raw = input("Enter three angles: ").split()
        if raw:
            a, b, c = float(raw[0]), float(raw[1]), float(raw[2])
            print(triangle_angles_classifier(a, b, c))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ((60.0, 60.0, 60.0), "Type: Acute | Max Angle: 60.0°"),
        ((90.0, 45.0, 45.0), "Type: Right | Max Angle: 90.0°"),
        ((120.0, 30.0, 30.0), "Type: Obtuse | Max Angle: 120.0°"),
        ((50.0, 50.0, 50.0), "Invalid Triangle"),
        ((-10.0, 100.0, 90.0), "Invalid Triangle"),
    ]
    for args, expected in cases:
        actual = triangle_angles_classifier(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Digital Root & Iterative Reduction Counter (Difficult — Loops)

### 📖 Problem Statement
Given a non-negative integer `n`, repeatedly sum its digits in a loop until the value is reduced to a single digit (0 through 9), known as the digital root.

Count the number of reduction rounds (iterations) required to achieve a single digit, and output the final root. If `n` is already a single digit (< 10), the number of rounds is 0.

### 📥 Input Format
A single line containing a single non-negative integer:
`n`

### 📤 Output Format
Print a single line formatted as:
`Rounds: <rounds> | Digital Root: <root>`

### 🔒 Constraints
- 0 <= n <= 10^18

### 💡 Example 1
```text
Input:
9875

Output:
Rounds: 3 | Digital Root: 2

Explanation: Round 1: 9+8+7+5 = 29. Round 2: 2+9 = 11. Round 3: 1+1 = 2. Total 3 rounds.
```

### 💡 Example 2
```text
Input:
38

Output:
Rounds: 2 | Digital Root: 2

Explanation: Round 1: 3+8 = 11. Round 2: 1+1 = 2. Total 2 rounds.
```


````python
# 💡 Problem 5 Solution: Digital Root & Iterative Reduction Counter

def digital_root_tracker(n: int) -> str:
    rounds = 0
    curr = n
    while curr >= 10:
        curr = sum(int(d) for d in str(curr))
        rounds += 1
    return f"Rounds: {rounds} | Digital Root: {curr}"

def solve_problem_5():
    print("--- Running Problem 5 (Digital Root) ---")
    try:
        raw = input("Enter non-negative integer n: ").strip()
        if raw:
            print(digital_root_tracker(int(raw)))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        ((9875,), "Rounds: 3 | Digital Root: 2"),
        ((38,), "Rounds: 2 | Digital Root: 2"),
        ((7,), "Rounds: 0 | Digital Root: 7"),
        ((0,), "Rounds: 0 | Digital Root: 0"),
        ((999,), "Rounds: 2 | Digital Root: 9"),
    ]
    for args, expected in cases:
        actual = digital_root_tracker(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_5()
````
