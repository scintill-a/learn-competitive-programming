# 🔴 Problem 8: Budget Spender with Early Break (Hard — Loops & Control Flow)

### 📖 Problem Statement
You have a starting `budget` (integer) and a shopping cart of item prices `prices` (list of integers). You iterate through `prices` in order and purchase items:

1. If `price > 0` and you have enough budget (`budget >= price`), deduct the price from `budget` and increment `items_bought` by 1.
2. If `price > 0` but you **cannot afford it** (`budget < price`), skip this item (do not deduct, do not increment `items_bought`).
3. If `price == 0` (it's a free gift!), you take it: `items_bought` increases by 1.
4. **Early Termination:** If `budget` becomes exactly `0`, **stop shopping immediately** (break the loop).

Return `(remaining_budget, items_bought)` as a tuple.

### 📥 Input Format
- Line 1: Space-separated integers representing item `prices`.
- Line 2: An integer `budget`.

### 📤 Output Format
Print `(remaining_budget, items_bought)`.

### 💡 Example 1
```text
Input:
20 15 50 10 5
50

Output:
(5, 3)
```
*Explanation:*
- Buy 20: budget = 30, items = 1
- Buy 15: budget = 15, items = 2
- Item 50: Can't afford ($15 < 50$), skip!
- Buy 10: budget = 5, items = 3
- Item 5: Can afford! Buy 5: budget = 0, items = 4. Wait, let's trace: 50 - 20 = 30, 30 - 15 = 15, 15 - 10 = 5, 5 - 5 = 0. Budget is 0, stops with `(0, 4)`.

### 💡 Example 2
```text
Input:
10 0 100 20
30

Output:
(0, 3)
```
*Explanation:*
- Buy 10: budget = 20, items = 1
- Free item (0): budget = 20, items = 2
- Skip 100 ($20 < 100$)
- Buy 20: budget = 0, items = 3. Budget hits 0, stops immediately!
