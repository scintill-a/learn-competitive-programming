# 🟢 Problem 1: Cylinder Volume & Surface Area (Easy — I/O)

### 📖 Problem Statement
Given the **radius** ($r$) and **height** ($h$) of a cylinder on a single line, calculate both its **volume** and **total surface area** using $\pi = 3.141592653589793$ (or `math.pi`).

Formulas:
- $\text{Volume} = \pi r^2 h$
- $\text{Surface Area} = 2 \pi r h + 2 \pi r^2$

Format both numbers to **exactly 2 decimal places**.

### 📥 Input Format
- A single line containing two space-separated floats: `radius` and `height`.

### 📤 Output Format
Print: `Volume: <V>, Surface Area: <A>` where both are formatted to 2 decimal places.

### 🔒 Constraints
- $0.1 \le r, h \le 1000.0$

### 💡 Example 1
```text
Input:
3.0 5.0

Output:
Volume: 141.37, Surface Area: 150.80
```

### 💡 Example 2
```text
Input:
1.5 2.0

Output:
Volume: 14.14, Surface Area: 32.99
```

````python
# 💡 Problem 1 Solution: Cylinder Volume & Surface Area
import math

def cylinder_metrics(radius: float, height: float) -> str:
    """
    Calculates cylinder volume and total surface area, formatted to 2 decimal places.
    """
    volume = math.pi * (radius ** 2) * height
    surface_area = (2 * math.pi * radius * height) + (2 * math.pi * (radius ** 2))
    return f"Volume: {volume:.2f}, Surface Area: {surface_area:.2f}"


def solve_problem_1():
    try:
        line = input("Enter radius and height (e.g. 3.0 5.0): ").strip()
        r_str, h_str = line.split()
        radius = float(r_str)
        height = float(h_str)
        print(cylinder_metrics(radius, height))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    test_cases = [
        (3.0, 5.0, "Volume: 141.37, Surface Area: 150.80"),
        (1.5, 2.0, "Volume: 14.14, Surface Area: 32.99"),
        (1.0, 1.0, "Volume: 3.14, Surface Area: 12.57"),
        (10.0, 20.0, "Volume: 6283.19, Surface Area: 1884.96"),
    ]
    
    for r, h, expected in test_cases:
        actual = cylinder_metrics(r, h)
        assert actual == expected, f"Failed for r={r}, h={h}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: cylinder_metrics({r}, {h}) -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Currency Breakdown Cashier (Easy — I/O)

### 📖 Problem Statement
Given an integer amount in whole dollars, break it down into the minimum number of banknotes using only the standard denominations: **$100, $50, $20, $10, $5, and $1**.

Compute how many of each note are needed, starting from the largest denomination ($100) down to the smallest ($1).

### 📥 Input Format
- A single line containing an integer `amount` ($1 \le \text{amount} \le 100000$).

### 📤 Output Format
Print the breakdown in the format:
`100s: <a>, 50s: <b>, 20s: <c>, 10s: <d>, 5s: <e>, 1s: <f>`

### 💡 Example 1
```text
Input:
288

Output:
100s: 2, 50s: 1, 20s: 1, 10s: 1, 5s: 1, 1s: 3
```
*Explanation:* $2\times100 + 1\times50 + 1\times20 + 1\times10 + 1\times5 + 3\times1 = 200 + 50 + 20 + 10 + 5 + 3 = 288$.

### 💡 Example 2
```text
Input:
100

Output:
100s: 1, 50s: 0, 20s: 0, 10s: 0, 5s: 0, 1s: 0
```

````python
# 💡 Problem 2 Solution: Currency Breakdown Cashier

def currency_breakdown(amount: int) -> str:
    """
    Breaks down an amount in whole dollars into 100s, 50s, 20s, 10s, 5s, and 1s.
    """
    c100 = amount // 100
    rem = amount % 100
    
    c50 = rem // 50
    rem = rem % 50
    
    c20 = rem // 20
    rem = rem % 20
    
    c10 = rem // 10
    rem = rem % 10
    
    c5 = rem // 5
    c1 = rem % 5
    
    return f"100s: {c100}, 50s: {c50}, 20s: {c20}, 10s: {c10}, 5s: {c5}, 1s: {c1}"


def solve_problem_2():
    try:
        amt = int(input("Enter amount in whole dollars: ").strip())
        print(currency_breakdown(amt))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    test_cases = [
        (288, "100s: 2, 50s: 1, 20s: 1, 10s: 1, 5s: 1, 1s: 3"),
        (100, "100s: 1, 50s: 0, 20s: 0, 10s: 0, 5s: 0, 1s: 0"),
        (49, "100s: 0, 50s: 0, 20s: 2, 10s: 0, 5s: 1, 1s: 4"),
        (1, "100s: 0, 50s: 0, 20s: 0, 10s: 0, 5s: 0, 1s: 1"),
        (386, "100s: 3, 50s: 1, 20s: 1, 10s: 1, 5s: 1, 1s: 1"),
    ]
    
    for amt, expected in test_cases:
        actual = currency_breakdown(amt)
        assert actual == expected, f"Failed for amount={amt}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: currency_breakdown({amt}) -> {actual}")

test_problem_2()
````

# 🟢 Problem 3: Speed & Travel Time Estimator (Easy — I/O)

### 📖 Problem Statement
A road trip planner needs to estimate travel duration. Given a **distance in kilometers** (float) and an average **speed in kilometers per hour** (float), compute the total travel time.

Output the result converted into full **hours and minutes** (e.g. `2 hours and 15 minutes`). Round the minutes to the nearest whole minute.

### 📥 Input Format
- A single line containing two space-separated floats: `distance` and `speed`.

### 📤 Output Format
Print: `<H> hours and <M> minutes`

### 🔒 Constraints
- $0.5 \le \text{distance} \le 10000.0$
- $1.0 \le \text{speed} \le 300.0$

### 💡 Example 1
```text
Input:
150.0 60.0

Output:
2 hours and 30 minutes
```
*Explanation:* $150 / 60 = 2.5$ hours = 2 hours and 30 minutes.

### 💡 Example 2
```text
Input:
85.0 80.0

Output:
1 hours and 4 minutes
```
*Explanation:* $85 / 80 = 1.0625$ hours. $0.0625 \times 60 = 3.75 \approx 4$ minutes.

````python
# 💡 Problem 3 Solution: Speed & Travel Time Estimator

def estimate_travel_time(distance: float, speed: float) -> str:
    """
    Calculates travel duration in hours and minutes (rounded to the nearest minute).
    """
    total_minutes = round((distance / speed) * 60)
    hours = total_minutes // 60
    minutes = total_minutes % 60
    return f"{hours} hours and {minutes} minutes"


def solve_problem_3():
    try:
        line = input("Enter distance (km) and speed (km/h): ").strip()
        dist_str, spd_str = line.split()
        dist = float(dist_str)
        spd = float(spd_str)
        print(estimate_travel_time(dist, spd))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    test_cases = [
        (150.0, 60.0, "2 hours and 30 minutes"),
        (85.0, 80.0, "1 hours and 4 minutes"),
        (100.0, 50.0, "2 hours and 0 minutes"),
        (45.0, 90.0, "0 hours and 30 minutes"),
        (119.9, 60.0, "2 hours and 0 minutes"),
    ]
    
    for dist, spd, expected in test_cases:
        actual = estimate_travel_time(dist, spd)
        assert actual == expected, f"Failed for distance={dist}, speed={spd}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: estimate_travel_time({dist}, {spd}) -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: 2D Cartesian Quadrant & Axis Identifier (Average — Conditionals)

### 📖 Problem Statement
Given the coordinates of a 2D point $(x, y)$, classify where the point lies:
- `Origin` if $x = 0$ and $y = 0$.
- `X-Axis` if $y = 0$ and $x \ne 0$.
- `Y-Axis` if $x = 0$ and $y \ne 0$.
- `Quadrant 1` if $x > 0$ and $y > 0$.
- `Quadrant 2` if $x < 0$ and $y > 0$.
- `Quadrant 3` if $x < 0$ and $y < 0$.
- `Quadrant 4` if $x > 0$ and $y < 0$.

### 📥 Input Format
- A single line containing two space-separated floats: `x` and `y`.

### 📤 Output Format
Print the exact location name string as defined above.

### 💡 Example 1
```text
Input:
4.5 -2.0

Output:
Quadrant 4
```

### 💡 Example 2
```text
Input:
0 7.2

Output:
Y-Axis
```

### 💡 Example 3
```text
Input:
0 0

Output:
Origin
```

````python
# 💡 Problem 4 Solution: 2D Cartesian Quadrant & Axis Identifier

def classify_point(x: float, y: float) -> str:
    """
    Classifies a 2D Cartesian point into Origin, X-Axis, Y-Axis, or Quadrant 1-4.
    """
    if x == 0 and y == 0:
        return "Origin"
    elif x == 0:
        return "Y-Axis"
    elif y == 0:
        return "X-Axis"
    elif x > 0 and y > 0:
        return "Quadrant 1"
    elif x < 0 and y > 0:
        return "Quadrant 2"
    elif x < 0 and y < 0:
        return "Quadrant 3"
    else:
        return "Quadrant 4"


def solve_problem_4():
    try:
        line = input("Enter coordinates x and y (e.g. 4.5 -2.0): ").strip()
        x_str, y_str = line.split()
        x = float(x_str)
        y = float(y_str)
        print(classify_point(x, y))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    test_cases = [
        (4.5, -2.0, "Quadrant 4"),
        (0.0, 7.2, "Y-Axis"),
        (0.0, 0.0, "Origin"),
        (-3.0, 0.0, "X-Axis"),
        (1.0, 1.0, "Quadrant 1"),
        (-2.0, 5.0, "Quadrant 2"),
        (-4.0, -4.0, "Quadrant 3"),
        (5.0, -0.1, "Quadrant 4"),
    ]
    
    for x, y, expected in test_cases:
        actual = classify_point(x, y)
        assert actual == expected, f"Failed for x={x}, y={y}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: classify_point({x}, {y}) -> {actual}")

test_problem_4()
````

# 🟡 Problem 5: Triangle Validity & Type Classifier (Average — Conditionals)

### 📖 Problem Statement
Given three positive numbers representing the side lengths of a triangle $a, b, c$:

1. First, check if they form a **valid triangle** according to the Triangle Inequality Theorem:
   - $a + b > c$
   - $a + c > b$
   - $b + c > a$
   - If any condition fails, print `Invalid Triangle`.

2. If valid, classify the triangle into one of three types:
   - `Equilateral`: All three sides are equal ($a = b = c$).
   - `Isosceles`: Exactly two sides are equal.
   - `Scalene`: All three sides have different lengths.

### 📥 Input Format
- A single line containing three space-separated numbers (float or int): `a b c`.

### 📤 Output Format
Print `Equilateral`, `Isosceles`, `Scalene`, or `Invalid Triangle`.

### 💡 Example 1
```text
Input:
5 5 5

Output:
Equilateral
```

### 💡 Example 2
```text
Input:
5 5 8

Output:
Isosceles
```

### 💡 Example 3
```text
Input:
1 2 10

Output:
Invalid Triangle
```

````python
# 💡 Problem 5 Solution: Triangle Validity & Type Classifier

def classify_triangle(a: float, b: float, c: float) -> str:
    """
    Checks triangle inequality theorem and classifies triangle as Equilateral, Isosceles, Scalene, or Invalid Triangle.
    """
    if a <= 0 or b <= 0 or c <= 0 or (a + b <= c) or (a + c <= b) or (b + c <= a):
        return "Invalid Triangle"
    
    if a == b == c:
        return "Equilateral"
    elif a == b or b == c or a == c:
        return "Isosceles"
    else:
        return "Scalene"


def solve_problem_5():
    try:
        line = input("Enter sides a, b, c (e.g. 5 5 8): ").strip()
        a_str, b_str, c_str = line.split()
        a = float(a_str)
        b = float(b_str)
        c = float(c_str)
        print(classify_triangle(a, b, c))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    test_cases = [
        (5, 5, 5, "Equilateral"),
        (5, 5, 8, "Isosceles"),
        (1, 2, 10, "Invalid Triangle"),
        (3, 4, 5, "Scalene"),
        (7, 10, 7, "Isosceles"),
        (2, 2, 4, "Invalid Triangle"),
        (0, 5, 5, "Invalid Triangle"),
        (-3, 4, 5, "Invalid Triangle"),
    ]
    
    for a, b, c, expected in test_cases:
        actual = classify_triangle(a, b, c)
        assert actual == expected, f"Failed for a={a}, b={b}, c={c}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: classify_triangle({a}, {b}, {c}) -> {actual}")

test_problem_5()
````

# 🟡 Problem 6: Recursive Substring Search Without Loops (Average — Recursion)

### 📖 Problem Statement
Given a string `s` and a character `target_char`, determine if `target_char` exists in `s`, and return the **first index** where it appears. If it does not appear, return `-1`.

⚠️ **STRICT CONSTRAINT: NO 'for' or 'while' loops are permitted!** ⚠️
⚠️ **NO `s.find()`, `s.index()`, or `target_char in s` are permitted!** ⚠️
You must solve this recursively using indexing and conditional base cases.

### 📥 Input Format
- Line 1: A string `s`.
- Line 2: A single character `target_char`.

### 📤 Output Format
An integer representing the 0-based index of the first occurrence, or `-1` if not found.

### 💡 Example 1
```text
Input:
competitive
e

Output:
4
```
*Explanation:* The first 'e' appears at index 4 (`c-0, o-1, m-2, p-3, e-4`).

### 💡 Example 2
```text
Input:
python
z

Output:
-1
```

````python
# 💡 Problem 6 Solution: Recursive Substring Search Without Loops

def recursive_find_char(s: str, target_char: str, index: int = 0) -> int:
    """
    Finds first index of target_char in string s recursively without loops or built-in find/index/in.
    """
    # Base Case 1: Reached end of string without match
    if index >= len(s):
        return -1
    
    # Base Case 2: Target character matched at current index
    if s[index] == target_char:
        return index
    
    # Recursive Step: Inspect next position
    return recursive_find_char(s, target_char, index + 1)


def solve_problem_6():
    try:
        s = input("Enter string: ").strip()
        target = input("Enter target character: ").strip()
        print(recursive_find_char(s, target))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_6()
````

````python
# 🧪 Problem 6 Test Suite
def test_problem_6():
    test_cases = [
        ("competitive", "e", 4),
        ("python", "z", -1),
        ("hello", "h", 0),
        ("hello", "o", 4),
        ("", "a", -1),
        ("banana", "a", 1),
    ]
    
    for s, target, expected in test_cases:
        actual = recursive_find_char(s, target)
        assert actual == expected, f"Failed for s='{s}', target='{target}'. Expected {expected}, got {actual}"
        print(f"✅ PASS: recursive_find_char('{s}', '{target}') -> {actual}")

test_problem_6()
````

# 🔴 Problem 7: Alternating Sign Transformation Loop (Hard — Loops & Comprehensions)

### 📖 Problem Statement
Given a list of integers `nums`, build a new list `transformed` by applying the following rule based on the element's **index**:

1. If the index $i$ is **even** ($0, 2, 4, \dots$):
   - Double the value: $nums[i] \times 2$.
2. If the index $i$ is **odd** ($1, 3, 5, \dots$):
   - Negate the value: $-nums[i]$.

Try solving this with both a standard loop and a **list comprehension**!

### 📥 Input Format
- A single line containing space-separated integers `nums`.

### 📤 Output Format
Print the transformed list.

### 💡 Example 1
```text
Input:
1 2 3 4 5

Output:
[2, -2, 6, -4, 10]
```
*Explanation:*
- Index 0 (even): $1 \times 2 = 2$
- Index 1 (odd): $-(2) = -2$
- Index 2 (even): $3 \times 2 = 6$
- Index 3 (odd): $-(4) = -4$
- Index 4 (even): $5 \times 2 = 10$

### 💡 Example 2
```text
Input:
10 -5 20

Output:
[20, 5, 40]
```

````python
# 💡 Problem 7 Solution: Alternating Sign Transformation Loop

def alternating_transform(nums: list[int]) -> list[int]:
    """
    Transforms list by doubling elements at even indices and negating elements at odd indices.
    """
    result = []
    for i in range(len(nums)):
        if i % 2 == 0:
            result.append(nums[i] * 2)
        else:
            result.append(-nums[i])
    return result

# Alternative list comprehension:
# def alternating_transform_comp(nums: list[int]) -> list[int]:
#     return [nums[i] * 2 if i % 2 == 0 else -nums[i] for i in range(len(nums))]


def solve_problem_7():
    try:
        raw = input("Enter space-separated integers: ").strip()
        nums = list(map(int, raw.split()))
        print(alternating_transform(nums))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_7()
````

````python
# 🧪 Problem 7 Test Suite
def test_problem_7():
    test_cases = [
        ([1, 2, 3, 4, 5], [2, -2, 6, -4, 10]),
        ([10, -5, 20], [20, 5, 40]),
        ([], []),
        ([7], [14]),
        ([0, 0, 0], [0, 0, 0]),
        ([-2, -4, -6], [-4, 4, -12]),
    ]
    
    for nums, expected in test_cases:
        actual = alternating_transform(nums)
        assert actual == expected, f"Failed for nums={nums}. Expected {expected}, got {actual}"
        print(f"✅ PASS: alternating_transform({nums}) -> {actual}")

test_problem_7()
````

# 🔴 Problem 8: Budget Spender with Early Break (Hard — Loops & Control Flow)

### 📖 Problem Statement
You have a starting `budget` (integer) and a shopping cart of item prices `prices` (list of integers). You iterate through `prices` in order and purchase items:

1. If `price > 0` and you have enough budget (`budget >= price`), deduct the price from `budget` and increment `items_bought` by 1.
2. If `price > 0` but you **cannot afford it** (`budget < price`), skip this item (do not deduct, do not increment `items_bought`).
3. If `price == 0` (it's a free gift!), you take it: `items_bought` increases by 1.
4. **Early Termination:** If `budget` becomes exactly `0`, **stop shopping immediately** (break the loop).

Return `(remaining_budget, items_bought)` as a tuple.

### 📥 Input Format
- Line 1: Space-separated integers representing item `prices`.
- Line 2: An integer `budget`.

### 📤 Output Format
Print `(remaining_budget, items_bought)`.

### 💡 Example 1
```text
Input:
20 15 50 10 5
50

Output:
(0, 4)
```
*Explanation:*
- Buy 20: budget = 30, items = 1
- Buy 15: budget = 15, items = 2
- Item 50: Can't afford ($15 < 50$), skip!
- Buy 10: budget = 5, items = 3
- Item 5: Can afford! Buy 5: budget = 0, items = 4 (budget hits 0, stops immediately).

### 💡 Example 2
```text
Input:
10 0 100 20
30

Output:
(0, 3)
```
*Explanation:*
- Buy 10: budget = 20, items = 1
- Free item (0): budget = 20, items = 2
- Skip 100 ($20 < 100$)
- Buy 20: budget = 0, items = 3. Budget hits 0, stops immediately!

````python
# 💡 Problem 8 Solution: Budget Spender with Early Break

def spend_budget(prices: list[int], budget: int) -> tuple[int, int]:
    """
    Simulates sequential item purchasing with budget deduction, free items, and early exit when budget hits 0.
    """
    remaining_budget = budget
    items_bought = 0
    
    for price in prices:
        if remaining_budget == 0:
            break
            
        if price > 0 and remaining_budget >= price:
            remaining_budget -= price
            items_bought += 1
        elif price == 0:
            items_bought += 1
        # price > remaining_budget is skipped (cannot afford)
        
        if remaining_budget == 0:
            break
            
    return remaining_budget, items_bought


def solve_problem_8():
    try:
        raw_prices = input("Enter space-separated prices: ").strip()
        prices = list(map(int, raw_prices.split()))
        budget = int(input("Enter starting budget: ").strip())
        print(spend_budget(prices, budget))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_8()
````

````python
# 🧪 Problem 8 Test Suite
def test_problem_8():
    test_cases = [
        ([20, 15, 50, 10, 5], 50, (0, 4)),
        ([10, 0, 100, 20], 30, (0, 3)),
        ([100, 200], 50, (50, 0)),
        ([0, 0, 0], 10, (10, 3)),
        ([5, 5], 10, (0, 2)),
        ([10, 20, 30], 0, (0, 0)),
    ]
    
    for prices, budget, expected in test_cases:
        actual = spend_budget(prices, budget)
        assert actual == expected, f"Failed for prices={prices}, budget={budget}. Expected {expected}, got {actual}"
        print(f"✅ PASS: spend_budget({prices}, {budget}) -> {actual}")

test_problem_8()
````

# 🔴 Problem 9: First Non-Repeating Element via Frequency Map (Hard — Dicts & Frequency)

### 📖 Problem Statement
Given a list of integers `nums`, find the **first element** in the list that occurs **only once** (frequency == 1).

- If every element repeats or the list is empty, return `None` (or print `None`).
- *Hint:* Count the frequencies of all numbers first into a dictionary (using `.get()` or `Counter`), then iterate through `nums` a second time to see which element was the first to have a count of 1.

### 📥 Input Format
- A single line containing space-separated integers `nums`.

### 📤 Output Format
The first non-repeating integer, or `None`.

### 💡 Example 1
```text
Input:
4 5 1 2 0 4 1 2

Output:
5
```
*Explanation:*
- 4 appears twice.
- 5 appears once (this is the first non-repeating element!).
- 1 appears twice, 2 appears twice, 0 appears once (but appears after 5).

### 💡 Example 2
```text
Input:
7 7 8 8 9 9

Output:
None
```

*Explanation:* All numbers repeat.

````python
# 💡 Problem 9 Solution: First Non-Repeating Element via Frequency Map

def first_non_repeating(nums: list[int]) -> int | None:
    """
    Finds first element in nums that occurs exactly once using a hash map frequency counter.
    """
    freq = {}
    for x in nums:
        freq[x] = freq.get(x, 0) + 1
        
    for x in nums:
        if freq[x] == 1:
            return x
            
    return None


def solve_problem_9():
    try:
        raw = input("Enter space-separated integers: ").strip()
        nums = list(map(int, raw.split()))
        print(first_non_repeating(nums))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_9()
````
