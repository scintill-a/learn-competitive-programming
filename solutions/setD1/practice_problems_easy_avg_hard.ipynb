# 🟢 Problem 1: Simple Interest & Maturity Value (Easy — Sequential/IO)

### 📖 Problem Statement
Given the initial principal amount, an annual interest rate in percent, and the investment duration in years on a single line, calculate both the simple interest earned and the final maturity value.

Formulas:
Interest = principal × (rate_pct / 100) × years
Maturity Value = principal + Interest

Display both monetary values with a dollar sign prefix and rounded to 2 decimal places.

### 📥 Input Format
A single line containing three space-separated numbers:
`principal` (float) `rate_pct` (float) `years` (float)

### 📤 Output Format
Print a single line formatted as:
`Interest: $XX.XX | Maturity: $XX.XX`

### 🔒 Constraints
- 1.00 <= principal <= 1,000,000.00
- 0.0 <= rate_pct <= 100.0
- 0.1 <= years <= 100.0

### 💡 Example 1
```text
Input:
1000.00 5.0 3.0

Output:
Interest: $150.00 | Maturity: $1150.00

Explanation: Interest = 1000.00 × 0.05 × 3 = 150.00. Maturity = 1000.00 + 150.00 = 1150.00.
```

### 💡 Example 2
```text
Input:
2500.00 4.5 2.0

Output:
Interest: $225.00 | Maturity: $2725.00

Explanation: Interest = 2500.00 × 0.045 × 2 = 225.00. Maturity = 2500.00 + 225.00 = 2725.00.
```


````python
# 💡 Problem 1 Solution: Simple Interest & Maturity Value

def simple_interest_calc(principal: float, rate_pct: float, years: float) -> str:
    interest = principal * (rate_pct / 100.0) * years
    maturity = principal + interest
    return f"Interest: ${interest:.2f} | Maturity: ${maturity:.2f}"

def solve_problem_1():
    print("--- Running Problem 1 (Input/Output) ---")
    try:
        raw = input("Enter principal, rate_pct, years: ").split()
        if raw:
            p, r, y = float(raw[0]), float(raw[1]), float(raw[2])
            print(simple_interest_calc(p, r, y))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((1000.00, 5.0, 3.0), "Interest: $150.00 | Maturity: $1150.00"),
        ((2500.00, 4.5, 2.0), "Interest: $225.00 | Maturity: $2725.00"),
        ((500.00, 10.0, 1.5), "Interest: $75.00 | Maturity: $575.00"),
    ]
    for args, expected in cases:
        actual = simple_interest_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Elapsed Time & Duration in Minutes (Easy — Sequential/Time Math)

### 📖 Problem Statement
Given a starting time and an ending time on the same 24-hour day in `HH:MM` format on a single line (where the ending time is greater than or equal to the starting time), calculate the total elapsed time in minutes and format the duration as hours and minutes.

### 📥 Input Format
A single line containing two space-separated time strings:
`start_time` and `end_time` in `HH:MM` format (00:00 to 23:59).

### 📤 Output Format
Print a single line formatted as:
`Total Minutes: <minutes> | Duration: <H>h <M>m`

### 🔒 Constraints
- 00:00 <= start_time <= end_time <= 23:59

### 💡 Example 1
```text
Input:
09:15 11:45

Output:
Total Minutes: 150 | Duration: 2h 30m

Explanation: 09:15 is 555 mins from midnight; 11:45 is 705 mins. Elapsed = 150 mins = 2h 30m.
```

### 💡 Example 2
```text
Input:
14:00 14:45

Output:
Total Minutes: 45 | Duration: 0h 45m

Explanation: Same hour, difference is exactly 45 minutes = 0h 45m.
```


````python
# 💡 Problem 2 Solution: Elapsed Time & Duration in Minutes

def elapsed_time_calc(start_time: str, end_time: str) -> str:
    sh, sm = map(int, start_time.split(":"))
    eh, em = map(int, end_time.split(":"))
    total_min = (eh * 60 + em) - (sh * 60 + sm)
    h, m = divmod(total_min, 60)
    return f"Total Minutes: {total_min} | Duration: {h}h {m}m"

def solve_problem_2():
    print("--- Running Problem 2 (Time Math) ---")
    try:
        raw = input("Enter start_time (HH:MM) and end_time (HH:MM): ").split()
        if raw:
            print(elapsed_time_calc(raw[0], raw[1]))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        (("09:15", "11:45"), "Total Minutes: 150 | Duration: 2h 30m"),
        (("14:00", "14:45"), "Total Minutes: 45 | Duration: 0h 45m"),
        (("08:30", "18:00"), "Total Minutes: 570 | Duration: 9h 30m"),
        (("10:00", "10:00"), "Total Minutes: 0 | Duration: 0h 0m"),
    ]
    for args, expected in cases:
        actual = elapsed_time_calc(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Body Mass Index & Category Classifier (Average — Selection)

### 📖 Problem Statement
Given an individual's weight in kilograms and height in meters on a single line, calculate their Body Mass Index (BMI) and classify their weight category:

Formula:
BMI = weight / (height²)

Categories based on calculated BMI (rounded to 1 decimal place):
- BMI < 18.5: `Underweight`
- 18.5 <= BMI < 25.0: `Normal Weight`
- 25.0 <= BMI < 30.0: `Overweight`
- BMI >= 30.0: `Obese`

### 📥 Input Format
A single line containing two space-separated numbers:
`weight_kg` (float) `height_m` (float)

### 📤 Output Format
Print a single line formatted as:
`BMI: <bmi:.1f> | Category: <Category>`

### 🔒 Constraints
- 10.0 <= weight_kg <= 300.0
- 0.50 <= height_m <= 2.50

### 💡 Example 1
```text
Input:
70.0 1.75

Output:
BMI: 22.9 | Category: Normal Weight

Explanation: BMI = 70.0 / (1.75²) = 22.86 -> 22.9. Belongs to Normal Weight (18.5 to 24.9).
```

### 💡 Example 2
```text
Input:
95.0 1.80

Output:
BMI: 29.3 | Category: Overweight

Explanation: BMI = 95.0 / (1.80²) = 29.32 -> 29.3. Belongs to Overweight (25.0 to 29.9).
```


````python
# 💡 Problem 3 Solution: Body Mass Index & Category Classifier

def bmi_classifier(weight_kg: float, height_m: float) -> str:
    bmi = weight_kg / (height_m ** 2)
    bmi_val = round(bmi, 1)
    if bmi_val < 18.5:
        category = "Underweight"
    elif bmi_val < 25.0:
        category = "Normal Weight"
    elif bmi_val < 30.0:
        category = "Overweight"
    else:
        category = "Obese"
    return f"BMI: {bmi_val:.1f} | Category: {category}"

def solve_problem_3():
    print("--- Running Problem 3 (Selection) ---")
    try:
        raw = input("Enter weight (kg) and height (m): ").split()
        if raw:
            w, h = float(raw[0]), float(raw[1])
            print(bmi_classifier(w, h))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((70.0, 1.75), "BMI: 22.9 | Category: Normal Weight"),
        ((95.0, 1.80), "BMI: 29.3 | Category: Overweight"),
        ((50.0, 1.70), "BMI: 17.3 | Category: Underweight"),
        ((110.0, 1.75), "BMI: 35.9 | Category: Obese"),
    ]
    for args, expected in cases:
        actual = bmi_classifier(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Quadratic Roots & Discriminant Classifier (Average — Selection)

### 📖 Problem Statement
Given three real coefficients `a`, `b`, and `c` of a quadratic equation `a × x² + b × x + c = 0` (where `a ≠ 0`) on a single line, calculate the discriminant `D = b² - 4 × a × c` and determine the nature of its roots:

- If `D > 0`: Two distinct real roots.
  Roots: `(-b + √D) / (2a)` and `(-b - √D) / (2a)` in descending order (larger root first).
- If `D == 0`: One repeated real root.
  Root: `-b / (2a)`
- If `D < 0`: No real roots.

All numeric root outputs must be rounded and formatted to 2 decimal places.

### 📥 Input Format
A single line containing three space-separated numbers:
`a` (float) `b` (float) `c` (float), where a != 0.

### 📤 Output Format
- Two Real Roots: `Type: Two Real Roots | Roots: <larger:.2f>, <smaller:.2f>`
- One Real Root: `Type: One Real Root | Root: <root:.2f>`
- No Real Roots: `Type: No Real Roots`

### 🔒 Constraints
- -1000.0 <= a, b, c <= 1000.0, with a != 0

### 💡 Example 1
```text
Input:
1.0 -5.0 6.0

Output:
Type: Two Real Roots | Roots: 3.00, 2.00

Explanation: D = (-5)² - 4(1)(6) = 1. Roots are (5 + 1)/2 = 3.00 and (5 - 1)/2 = 2.00.
```

### 💡 Example 2
```text
Input:
1.0 -4.0 4.0

Output:
Type: One Real Root | Root: 2.00

Explanation: D = (-4)² - 4(1)(4) = 0. Single repeated root = 4 / 2 = 2.00.
```


````python
# 💡 Problem 4 Solution: Quadratic Roots & Discriminant Classifier
import math

def quadratic_roots_classifier(a: float, b: float, c: float) -> str:
    disc = b**2 - 4.0 * a * c
    if disc > 1e-9:
        r1 = (-b + math.sqrt(disc)) / (2.0 * a)
        r2 = (-b - math.sqrt(disc)) / (2.0 * a)
        larger = max(r1, r2)
        smaller = min(r1, r2)
        return f"Type: Two Real Roots | Roots: {larger:.2f}, {smaller:.2f}"
    elif abs(disc) <= 1e-9:
        root = -b / (2.0 * a)
        return f"Type: One Real Root | Root: {root:.2f}"
    else:
        return "Type: No Real Roots"

def solve_problem_4():
    print("--- Running Problem 4 (Quadratic Solver) ---")
    try:
        raw = input("Enter coefficients a, b, c: ").split()
        if raw:
            a, b, c = float(raw[0]), float(raw[1]), float(raw[2])
            print(quadratic_roots_classifier(a, b, c))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ((1.0, -5.0, 6.0), "Type: Two Real Roots | Roots: 3.00, 2.00"),
        ((1.0, -4.0, 4.0), "Type: One Real Root | Root: 2.00"),
        ((1.0, 2.0, 5.0), "Type: No Real Roots"),
        ((2.0, -7.0, 3.0), "Type: Two Real Roots | Roots: 3.00, 0.50"),
    ]
    for args, expected in cases:
        actual = quadratic_roots_classifier(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Collatz 3n+1 Sequence Peak & Steps (Difficult — Loops)

### 📖 Problem Statement
Given a starting positive integer `n`, generate the 3n + 1 (Collatz) sequence using a loop until the value reaches 1:
- If `n` is even: divide it by 2 (`n = n // 2`).
- If `n` is odd: multiply by 3 and add 1 (`n = 3 × n + 1`).

Count the total number of transition steps taken to reach 1, and determine the peak (maximum) value attained at any point during the trajectory (including the starting number). If `n` is already 1, the number of steps is 0 and peak is 1.

### 📥 Input Format
A single line containing a single positive integer:
`n`

### 📤 Output Format
Print a single line formatted as:
`Steps: <step_count> | Peak: <peak_value>`

### 🔒 Constraints
- 1 <= n <= 1,000,000

### 💡 Example 1
```text
Input:
6

Output:
Steps: 8 | Peak: 16

Explanation: 6 → 3 → 10 → 5 → 16 → 8 → 4 → 2 → 1. Steps = 8, Peak = 16.
```

### 💡 Example 2
```text
Input:
11

Output:
Steps: 14 | Peak: 52

Explanation: Sequence: 11 → 34 → 17 → 52 → 26 → 13 → 40 → 20 → 10 → 5 → 16 → 8 → 4 → 2 → 1. Steps = 14, Peak = 52.
```


````python
# 💡 Problem 5 Solution: Collatz 3n+1 Sequence Peak & Steps

def collatz_tracker(n: int) -> str:
    steps = 0
    peak = n
    curr = n
    while curr > 1:
        if curr % 2 == 0:
            curr = curr // 2
        else:
            curr = 3 * curr + 1
        steps += 1
        if curr > peak:
            peak = curr
    return f"Steps: {steps} | Peak: {peak}"

def solve_problem_5():
    print("--- Running Problem 5 (Loop Simulation) ---")
    try:
        raw = input("Enter starting integer n: ").strip()
        if raw:
            n = int(raw)
            print(collatz_tracker(n))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        ((6,), "Steps: 8 | Peak: 16"),
        ((11,), "Steps: 14 | Peak: 52"),
        ((1,), "Steps: 0 | Peak: 1"),
        ((2,), "Steps: 1 | Peak: 2"),
        ((7,), "Steps: 16 | Peak: 52"),
    ]
    for args, expected in cases:
        actual = collatz_tracker(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_5()
````
