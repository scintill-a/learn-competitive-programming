# 🟢 Problem 1: Cart Checkout & Remaining Balance Calculator (Easy — Sequential/IO)

### 📖 Problem Statement
A customer arrives at a retail checkout counter with an initial cash balance. They purchase `item_count` identical items at `unit_price` each. 
A promotional discount of `discount_pct` percent is applied directly to the subtotal of the items. Following the discount, a local sales tax of `tax_pct` percent is applied to the discounted subtotal.

Calculate:
1. The **final total cost** (rounded to 2 decimal places).
2. The customer's **remaining balance** after payment (initial balance - total cost, can be negative if overspent).
3. The **average effective cost per item** (total cost / item count).

### 📥 Input Format
A single line containing five space-separated values:
`initial_balance` (float) `item_count` (int) `unit_price` (float) `discount_pct` (float) `tax_pct` (float)

### 📤 Output Format
Print a single line formatted exactly as:
`Total Cost: $XX.XX | Remaining: $XX.XX | Avg Per Item: $XX.XX`

### 🔒 Constraints
- 0.00 <= initial_balance <= 100,000.00
- 1 <= item_count <= 10,000
- 0.01 <= unit_price <= 5,000.00
- 0.0 <= discount_pct <= 100.0
- 0.0 <= tax_pct <= 50.0

### 💡 Example 1
```text
Input:
100.00 4 20.00 10 5

Output:
Total Cost: $75.60 | Remaining: $24.40 | Avg Per Item: $18.90
```
*Explanation:*
- Subtotal = 4 × 20.00 = 80.00
- After 10% discount: 80.00 × 0.90 = 72.00
- After 5% tax: 72.00 × 1.05 = 75.60
- Remaining = 100.00 - 75.60 = 24.40
- Avg per item = 75.60 / 4 = 18.90

### 💡 Example 2
```text
Input:
50.00 2 25.00 0 8

Output:
Total Cost: $54.00 | Remaining: $-4.00 | Avg Per Item: $27.00
```


````python
# 💡 Problem 1 Solution: Cart Checkout & Remaining Balance

def checkout_calculator(initial_balance: float, item_count: int, unit_price: float, discount_pct: float, tax_pct: float) -> str:
    subtotal = item_count * unit_price
    discounted = subtotal * (1.0 - discount_pct / 100.0)
    total_cost = discounted * (1.0 + tax_pct / 100.0)
    remaining_balance = initial_balance - total_cost
    avg_per_item = total_cost / item_count if item_count > 0 else 0.0
    
    return f"Total Cost: ${total_cost:.2f} | Remaining: ${remaining_balance:.2f} | Avg Per Item: ${avg_per_item:.2f}"

def solve_problem_1():
    print("--- Running Problem 1 (Input/Output) ---")
    try:
        raw = input("Enter initial_balance, item_count, unit_price, discount_pct, tax_pct: ").split()
        if raw:
            bal, count, price, disc, tax = float(raw[0]), int(raw[1]), float(raw[2]), float(raw[3]), float(raw[4])
            print(checkout_calculator(bal, count, price, disc, tax))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    cases = [
        ((100.00, 4, 20.00, 10.0, 5.0), "Total Cost: $75.60 | Remaining: $24.40 | Avg Per Item: $18.90"),
        ((50.00, 2, 25.00, 0.0, 8.0), "Total Cost: $54.00 | Remaining: $-4.00 | Avg Per Item: $27.00"),
        ((200.00, 5, 30.00, 20.0, 0.0), "Total Cost: $120.00 | Remaining: $80.00 | Avg Per Item: $24.00"),
    ]
    for args, expected in cases:
        actual = checkout_calculator(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Race Split & Pace Formatter (Easy — Sequential/Time Math)

### 📖 Problem Statement
A long-distance runner completes a timed trial. Given the distance in kilometers and the total finish time in `HH:MM:SS` format on a single line:
1. Compute the **total elapsed seconds**.
2. Compute the **average speed** in km/h, rounded to 2 decimal places.
3. Compute the **running pace** (minutes and seconds per kilometer, rounded to nearest whole second), formatted strictly as `MM:SS/km` with two zero-padded digits each.

### 📥 Input Format
A single line containing:
`distance_km` (float) `finish_time` (string in `HH:MM:SS` format)

### 📤 Output Format
Print a single line formatted as:
`Total Seconds: <total_seconds> | Speed: <speed:.2f> km/h | Pace: <MM>:<SS>/km`

### 🔒 Constraints
- 0.1 <= distance_km <= 500.0
- 00:00:01 <= finish_time <= 99:59:59

### 💡 Example 1
```text
Input:
10.0 00:50:00

Output:
Total Seconds: 3000 | Speed: 12.00 km/h | Pace: 05:00/km
```

### 💡 Example 2
```text
Input:
5.0 00:27:30

Output:
Total Seconds: 1650 | Speed: 10.91 km/h | Pace: 05:30/km
```


````python
# 💡 Problem 2 Solution: Race Split & Pace Formatter

def calculate_race_pace(distance_km: float, finish_time: str) -> str:
    h, m, s = map(int, finish_time.split(":"))
    total_seconds = h * 3600 + m * 60 + s
    
    hours = total_seconds / 3600.0
    speed_kmh = distance_km / hours if hours > 0 else 0.0
    
    seconds_per_km = round(total_seconds / distance_km) if distance_km > 0 else 0
    pace_min = int(seconds_per_km // 60)
    pace_sec = int(seconds_per_km % 60)
    
    return f"Total Seconds: {total_seconds} | Speed: {speed_kmh:.2f} km/h | Pace: {pace_min:02d}:{pace_sec:02d}/km"

def solve_problem_2():
    print("--- Running Problem 2 (Input/Output) ---")
    try:
        raw = input("Enter distance and finish_time (e.g. 10.0 00:50:00): ").split()
        if raw:
            dist = float(raw[0])
            t_str = raw[1]
            print(calculate_race_pace(dist, t_str))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    cases = [
        ((10.0, "00:50:00"), "Total Seconds: 3000 | Speed: 12.00 km/h | Pace: 05:00/km"),
        ((5.0, "00:27:30"), "Total Seconds: 1650 | Speed: 10.91 km/h | Pace: 05:30/km"),
        ((42.195, "03:00:00"), "Total Seconds: 10800 | Speed: 14.06 km/h | Pace: 04:16/km"),
    ]
    for args, expected in cases:
        actual = calculate_race_pace(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_2()
````

# 🟡 Problem 3: Tiered Utility Billing & Surcharge Classifier (Average — Selection)

### 📖 Problem Statement
An electric cooperative bills residential consumers based on a progressive three-tier tariff schedule plus a mandatory base service fee:
- **Base Service Fee**: $10.00 (charged to all accounts).
- **Progressive Consumption Brackets**:
  - **Tier 1 (First 100 kWh)**: $0.12 per kWh (for usage between 0.0 and 100.0 kWh).
  - **Tier 2 (Next 200 kWh)**: $0.18 per kWh (for usage between 100.1 and 300.0 kWh).
  - **Tier 3 (Above 300 kWh)**: $0.25 per kWh for every kWh exceeding 300.0 kWh.
- **Low-Usage Subsidy**: If total consumption is <= 50.0 kWh, a flat $5.00 discount is deducted from the total bill.
- **Peak Hour Surcharge**: If the meter flag is `PEAK`, a 15% surcharge is added to the energy consumption charge (the surcharge does not apply to base fee or subsidy). If `NORMAL`, no surcharge is added.

Identify the **Tier Category**:
- `Tier 1 (Low)` if usage <= 100.0
- `Tier 2 (Moderate)` if 100.0 < usage <= 300.0
- `Tier 3 (High)` if usage > 300.0

### 📥 Input Format
A single line containing:
`kwh_usage` (float) `peak_flag` (string: `PEAK` or `NORMAL`)

### 📤 Output Format
`Tier Category: <category> | Energy Charge: $XX.XX | Total Bill: $XX.XX`

### 🔒 Constraints
- 0.0 <= kwh_usage <= 10,000.0
- peak_flag in {"PEAK", "NORMAL"}

### 💡 Example 1
```text
Input:
80.0 NORMAL

Output:
Tier Category: Tier 1 (Low) | Energy Charge: $9.60 | Total Bill: $19.60
```

### 💡 Example 2
```text
Input:
250.0 PEAK

Output:
Tier Category: Tier 2 (Moderate) | Energy Charge: $44.85 | Total Bill: $54.85
```


````python
# 💡 Problem 3 Solution: Tiered Utility Billing & Surcharge Classifier

def calculate_electric_bill(kwh: float, peak_flag: str) -> str:
    base_fee = 10.00
    
    if kwh <= 100.0:
        tier_category = "Tier 1 (Low)"
        raw_energy = kwh * 0.12
    elif kwh <= 300.0:
        tier_category = "Tier 2 (Moderate)"
        raw_energy = (100.0 * 0.12) + ((kwh - 100.0) * 0.18)
    else:
        tier_category = "Tier 3 (High)"
        raw_energy = (100.0 * 0.12) + (200.0 * 0.18) + ((kwh - 300.0) * 0.25)
        
    if peak_flag.upper() == "PEAK":
        energy_charge = raw_energy * 1.15
    else:
        energy_charge = raw_energy
        
    subsidy = 5.00 if kwh <= 50.0 else 0.00
    total_bill = base_fee + energy_charge - subsidy
    
    return f"Tier Category: {tier_category} | Energy Charge: ${energy_charge:.2f} | Total Bill: ${total_bill:.2f}"

def solve_problem_3():
    print("--- Running Problem 3 (Decision / Selection) ---")
    try:
        raw = input("Enter kwh and peak_flag (e.g. 250.0 PEAK): ").split()
        if raw:
            kwh = float(raw[0])
            flag = raw[1]
            print(calculate_electric_bill(kwh, flag))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    cases = [
        ((80.0, "NORMAL"), "Tier Category: Tier 1 (Low) | Energy Charge: $9.60 | Total Bill: $19.60"),
        ((250.0, "PEAK"), "Tier Category: Tier 2 (Moderate) | Energy Charge: $44.85 | Total Bill: $54.85"),
        ((40.0, "NORMAL"), "Tier Category: Tier 1 (Low) | Energy Charge: $4.80 | Total Bill: $9.80"),
        ((350.0, "NORMAL"), "Tier Category: Tier 3 (High) | Energy Charge: $60.50 | Total Bill: $70.50"),
    ]
    for args, expected in cases:
        actual = calculate_electric_bill(*args)
        assert actual == expected, f"Failed for {args}: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {args} -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Calendar Date Validator & Day-of-Year Offset (Average — Selection)

### 📖 Problem Statement
Given a date string formatted as `YYYY-MM-DD`, determine whether it represents a valid calendar date according to the standard Gregorian calendar:
- **Year** must be an integer >= 1.
- **Month** must be between 1 and 12.
- **Day** must be between 1 and the maximum number of days in that specific month.
- **Leap Year Rules**:
  A year is a leap year if and only if:
  `year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)`
  - February has 29 days in a leap year and 28 days in a standard year.
  - April, June, September, November have 30 days.
  - January, March, May, July, August, October, December have 31 days.

If the date is invalid, output `Invalid Date`.
If valid, calculate the **Day of Year** (1 for Jan 1st, up to 365 or 366 for Dec 31st).

*Rule: Do NOT use the `datetime` or `calendar` modules. Implement the logic using selection structures.*

### 📥 Input Format
A single line containing a date string: `YYYY-MM-DD`

### 📤 Output Format
If invalid: `Invalid Date`
If valid: `Status: Valid | Leap Year: <Yes/No> | Day of Year: <day_number>`

### 🔒 Constraints
- Input string has form `Y-M-D` or `YYYY-MM-DD`

### 💡 Example 1
```text
Input:
2024-02-29

Output:
Status: Valid | Leap Year: Yes | Day of Year: 60
```

### 💡 Example 2
```text
Input:
2023-02-29

Output:
Invalid Date
```


````python
# 💡 Problem 4 Solution: Calendar Date Validator & Day-of-Year Offset

def is_leap_year(year: int) -> bool:
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def validate_date(date_str: str) -> str:
    parts = date_str.strip().split("-")
    if len(parts) != 3:
        return "Invalid Date"
        
    try:
        year, month, day = int(parts[0]), int(parts[1]), int(parts[2])
    except ValueError:
        return "Invalid Date"
        
    if year < 1 or month < 1 or month > 12:
        return "Invalid Date"
        
    leap = is_leap_year(year)
    days_in_months = [0, 31, 29 if leap else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    
    if day < 1 or day > days_in_months[month]:
        return "Invalid Date"
        
    day_of_year = sum(days_in_months[1:month]) + day
    leap_str = "Yes" if leap else "No"
    
    return f"Status: Valid | Leap Year: {leap_str} | Day of Year: {day_of_year}"

def solve_problem_4():
    print("--- Running Problem 4 (Decision / Selection) ---")
    try:
        raw = input("Enter date (YYYY-MM-DD): ")
        if raw:
            print(validate_date(raw))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    cases = [
        ("2024-02-29", "Status: Valid | Leap Year: Yes | Day of Year: 60"),
        ("2023-02-29", "Invalid Date"),
        ("2000-12-31", "Status: Valid | Leap Year: Yes | Day of Year: 366"),
        ("1900-02-29", "Invalid Date"),
        ("2021-04-31", "Invalid Date"),
        ("2021-01-15", "Status: Valid | Leap Year: No | Day of Year: 15"),
    ]
    for date_str, expected in cases:
        actual = validate_date(date_str)
        assert actual == expected, f"Failed for '{date_str}': expected '{expected}', got '{actual}'"
        print(f"✅ PASS: {date_str} -> {actual}")

test_problem_4()
````

# 🔴 Problem 5: Inventory Transaction Stream & Alert Monitor (Difficult — Loop)

### 📖 Problem Statement
A logistics warehouse maintains an inventory management monitor. Given an `initial_stock` and a `threshold` level, process an incoming stream of transaction tokens:
- `+N`: Restock by `N` units. Current stock increases by `N`.
- `-N`: Fulfill an order of `N` units. 
  - If `current_stock >= N`: order is fulfilled, current stock decreases by `N`, and `units_sold` increases by `N`. If the new stock is <= `threshold`, trigger a low-stock alert (+1 alert count).
  - If `current_stock < N`: order is **rejected** (insufficient inventory). Stock remains unchanged, and `rejected_count` increases by 1.
- `END`: Marks the end of transaction sequence.

**Streak Tracking**:
- Track consecutive fulfilled restocks (`+N`).
- Track consecutive fulfilled orders (`-N`).
- If an order is rejected, active order and restock streaks both reset to 0.
- When an order occurs, the restock streak resets to 0. When a restock occurs, the order streak resets to 0.

### 📥 Input Format
- Line 1: `initial_stock` (int) `threshold` (int)
- Line 2: Sequence of tokens ending with `END` (e.g. `+50 -20 -40 END`)

### 📤 Output Format
`Final Stock: <stock> | Units Sold: <sold> | Rejected: <rejected> | Alerts: <alerts> | Max Restock Streak: <max_r> | Max Order Streak: <max_o>`

### 🔒 Constraints
- 0 <= initial_stock <= 1,000,000
- 0 <= threshold <= 1,000,000
- 1 <= N <= 100,000

### 💡 Example 1
```text
Input:
100 25
+50 -60 -40 -30 +20 -15 END

Output:
Final Stock: 25 | Units Sold: 145 | Rejected: 0 | Alerts: 2 | Max Restock Streak: 1 | Max Order Streak: 3
```

### 💡 Example 2
```text
Input:
10 10
-15 -5 -8 END

Output:
Final Stock: 5 | Units Sold: 5 | Rejected: 2 | Alerts: 1 | Max Restock Streak: 0 | Max Order Streak: 1
```


````python
# 💡 Problem 5 Solution: Inventory Transaction Stream & Alert Monitor

def simulate_inventory(initial_stock: int, threshold: int, tokens: list) -> str:
    current_stock = initial_stock
    units_sold = 0
    rejected_count = 0
    alert_count = 0
    
    current_restock_streak = 0
    max_restock_streak = 0
    current_order_streak = 0
    max_order_streak = 0
    
    for token in tokens:
        if token == "END":
            break
            
        if token.startswith("+"):
            qty = int(token[1:])
            current_stock += qty
            current_restock_streak += 1
            if current_restock_streak > max_restock_streak:
                max_restock_streak = current_restock_streak
            current_order_streak = 0
            
        elif token.startswith("-"):
            qty = int(token[1:])
            if current_stock < qty:
                rejected_count += 1
                current_order_streak = 0
                current_restock_streak = 0
            else:
                current_stock -= qty
                units_sold += qty
                if current_stock <= threshold:
                    alert_count += 1
                current_order_streak += 1
                if current_order_streak > max_order_streak:
                    max_order_streak = current_order_streak
                current_restock_streak = 0
                
    return (f"Final Stock: {current_stock} | Units Sold: {units_sold} | Rejected: {rejected_count} | "
            f"Alerts: {alert_count} | Max Restock Streak: {max_restock_streak} | Max Order Streak: {max_order_streak}")

def solve_problem_5():
    print("--- Running Problem 5 (Loop / Stream Simulation) ---")
    try:
        raw_header = input("Enter initial_stock and threshold: ").split()
        if raw_header:
            stock = int(raw_header[0])
            thresh = int(raw_header[1])
            tokens = input("Enter transactions ending with END (e.g. +50 -20 END): ").split()
            print(simulate_inventory(stock, thresh, tokens))
    except Exception as e:
        print(f"Error: {e}")

# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    cases = [
        ((100, 25, ["+50", "-60", "-40", "-30", "+20", "-15", "END"]),
         "Final Stock: 25 | Units Sold: 145 | Rejected: 0 | Alerts: 2 | Max Restock Streak: 1 | Max Order Streak: 3"),
        ((10, 10, ["-15", "-5", "-8", "END"]),
         "Final Stock: 5 | Units Sold: 5 | Rejected: 2 | Alerts: 1 | Max Restock Streak: 0 | Max Order Streak: 1"),
        ((50, 20, ["+10", "+15", "+25", "-30", "-20", "-10", "END"]),
         "Final Stock: 40 | Units Sold: 60 | Rejected: 0 | Alerts: 0 | Max Restock Streak: 3 | Max Order Streak: 3"),
    ]
    for (stock, thresh, tokens), expected in cases:
        actual = simulate_inventory(stock, thresh, tokens)
        assert actual == expected, f"Failed: expected '{expected}', got '{actual}'"
        print(f"✅ PASS: initial={stock}, threshold={thresh} -> {actual}")

test_problem_5()
````
