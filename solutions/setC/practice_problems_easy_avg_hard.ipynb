```python
def parse_server_log(log_line: str) -> str:
    tokens = log_line.strip().split()
    timestamp = tokens[0][1:-1]  # Strip leading '[' and trailing ']'
    
    latency_sec = 0.0
    for tok in tokens:
        if tok.startswith("LATENCY="):
            ms_str = tok[len("LATENCY="):].rstrip("ms")
            latency_sec = int(ms_str) / 1000.0
            break
            
    return f"Timestamp: {timestamp}, Latency: {latency_sec:.3f}s"

def solve():
    line = input().strip()
    print(parse_server_log(line))
```

**Why it works:**
- Splitting the raw line on spaces breaks it into predictable tokens: `['[14:30:25]', 'ENDPOINT=/api/v1/checkout', 'STATUS=200', 'LATENCY=142ms']`.
- Using `[1:-1]` on the timestamp token safely strips the brackets without altering any colons inside.
- Searching for `LATENCY=` using `.startswith()` guarantees robustness even if attributes appear in varying positions.

**Trap to avoid:**
Do not use `tok.replace("ms", "")` blindly because an endpoint or path might contain the letters `"ms"` (e.g. `ENDPOINT=/items/summary`). Always extract from the designated `LATENCY=` token specifically.
</details>

```python
import math

def calculate_distances(x1: float, y1: float, x2: float, y2: float) -> str:
    dx = x2 - x1
    dy = y2 - y1
    euclidean = math.hypot(dx, dy)
    manhattan = abs(dx) + abs(dy)
    return f"Euclidean: {euclidean:.2f}, Manhattan: {manhattan:.2f}"

def solve():
    x1, y1, x2, y2 = map(float, input().split())
    print(calculate_distances(x1, y1, x2, y2))
```

**Why it works:**
- `math.hypot(dx, dy)` calculates `√(dx² + dy²)` directly, avoiding intermediate numerical overflow when coordinates are large.
- `abs()` returns the absolute value for Manhattan grid distance.
- Unpacking `x1, y1, x2, y2 = map(float, input().split())` is the standard, fast competitive programming I/O pattern.

**Trap to avoid:**
Remember that Manhattan distance is the sum of *absolute* coordinate differences (`abs(dx) + abs(dy)`), not `abs(dx + dy)`.
</details>

```python
def hex_to_rgb_luminance(hex_code: str) -> str:
    s = hex_code.strip()
    if s.startswith("#"):
        s = s[1:]
    r = int(s[0:2], 16)
```

- The standard ITU-R BT.709 photometric coefficients sum to exactly 1.0 (`0.2126 + 0.7152 + 0.0722 = 1.0`), ensuring pure white (`#FFFFFF`) produces `255.0`.

# 🟡 Problem 4: Network Packet Firewall Filter (Average — Conditionals) ⭐⭐

### 📖 Problem Statement
A network security firewall inspects incoming packets. Each packet header is described by four fields: `PROTOCOL` (TCP or UDP), `PORT` (integer), `PACKET_SIZE_KB` (integer), and `IS_ENCRYPTED` (YES or NO).

Classify each packet by applying the following strict priority rules:
1. **Port 80 or 8080 (Plain HTTP):**
   - If `IS_ENCRYPTED` is `'YES'`: return `REJECT` (encrypted payload on plaintext port).
   - Otherwise: return `ALLOW_INSECURE`.
2. **Port 443 or 8443 (HTTPS/TLS):**
   - If `IS_ENCRYPTED` is `'NO'`: return `REJECT` (unencrypted payload on secure port).
   - If `PACKET_SIZE_KB > 500`: return `QUARANTINE_LARGE` (suspiciously oversized payload).
   - Otherwise: return `ALLOW_SECURE`.
3. **Port 53 (DNS) or 123 (NTP):**
   - If `PROTOCOL == 'UDP'` and `PACKET_SIZE_KB <= 64`: return `ALLOW_SYSTEM`.
   - Otherwise: return `BLOCK_SUSPICIOUS`.
4. **Any other port:**
   - If `PORT < 1024`: return `BLOCK_PRIVILEGED`.
   - Otherwise: return `ALLOW_CUSTOM`.

### 📥 Input Format
- A single line containing: `PROTOCOL PORT PACKET_SIZE_KB IS_ENCRYPTED`.

### 📤 Output Format
- Print the exact uppercase classification string.

### 🔒 Constraints
- `PROTOCOL` is `'TCP'` or `'UDP'`.
- `1 ≤ PORT ≤ 65535`.
- `1 ≤ PACKET_SIZE_KB ≤ 100000`.
- `IS_ENCRYPTED` is `'YES'` or `'NO'`.

### 💡 Example 1
```text
Input:
TCP 443 120 YES

Output:
ALLOW_SECURE
```

### 💡 Example 2
```text
Input:
UDP 53 128 NO

Output:
BLOCK_SUSPICIOUS
```
*Explanation:* Port 53 packet size 128 KB exceeds the 64 KB limit.

Handle well-known port groups in order: `port in (80, 8080)`, `port in (443, 8443)`, `port in (53, 123)`. Make sure to return immediately upon matching a branch to avoid accidental fall-through to the default port rule.

```python
def classify_packet(protocol: str, port: int, size_kb: int, is_encrypted: str) -> str:
    encrypted = (is_encrypted.upper() == "YES")
    proto = protocol.upper()
    
    # Rule 1: HTTP Plaintext Ports
    if port in (80, 8080):
        return "REJECT" if encrypted else "ALLOW_INSECURE"
        
    # Rule 2: HTTPS / TLS Ports
    if port in (443, 8443):
        if not encrypted:
            return "REJECT"
        if size_kb > 500:
            return "QUARANTINE_LARGE"
        return "ALLOW_SECURE"
        
    # Rule 3: Core System Ports (DNS / NTP)
    if port in (53, 123):
        if proto == "UDP" and size_kb <= 64:
            return "ALLOW_SYSTEM"
        return "BLOCK_SUSPICIOUS"
        
    # Rule 4: General Fallback Ports
    if port < 1024:
        return "BLOCK_PRIVILEGED"
        
    return "ALLOW_CUSTOM"

def solve():
    proto, port, size_kb, enc = input().split()
    print(classify_packet(proto, int(port), int(size_kb), enc))
```

**Why it works:**
- Organizing rules by port category prevents nested condition tangles.
- Checking `if not encrypted:` first for secure ports provides early rejection before checking payload size.
- The default branch cleanly divides low privileged ports (`< 1024`) from user ports (`≥ 1024`).

**Trap to avoid:**
Don't check `port < 1024` *before* checking ports 80, 443, 53, or 123! Those are also `< 1024`, so placing the general check first would accidentally intercept and misclassify standard web/DNS traffic.
</details>

# 🟡 Problem 5: 1D Interval Overlap & Collision Classifier (Average — Conditionals) ⭐⭐

### 📖 Problem Statement
Given two closed integer intervals `[s1, e1]` and `[s2, e2]` on a single line (where `s1 ≤ e1` and `s2 ≤ e2`), determine their spatial relationship and the exact number of shared integer points (intersection length).

Classify into one of four mutually exclusive categories:
- `DISJOINT`: No common points (intersection length = 0).
- `TOUCHING`: Share exactly one boundary point (intersection length = 1).
- `SUBSET`: One interval lies entirely inside the other (i.e. `s1 ≥ s2 and e1 ≤ e2`, OR `s2 ≥ s1 and e2 ≤ e1`).
- `OVERLAPPING`: Share two or more points (intersection length > 1), but neither interval is a subset of the other.

### 📥 Input Format
- A single line containing four space-separated integers: `s1 e1 s2 e2`.

### 📤 Output Format
- Print: `<RELATION> (Intersection: <length>)`

### 🔒 Constraints
- `-10000 ≤ s1 ≤ e1 ≤ 10000`
- `-10000 ≤ s2 ≤ e2 ≤ 10000`

### 💡 Example 1
```text
Input:
1 5 3 8

Output:
OVERLAPPING (Intersection: 3)
```
*Explanation:* Shared points are `{3, 4, 5}` (length 3). Neither interval is a subset of the other.

### 💡 Example 2
```text
Input:
2 4 1 6

Output:
SUBSET (Intersection: 3)
```
*Explanation:* Interval `[2, 4]` is completely contained within `[1, 6]`.

### 💡 Example 3
```text
Input:
1 3 3 7

Output:
TOUCHING (Intersection: 1)
```
*Explanation:* Only boundary point `3` is shared.

### 🔑 Regional Hint
The intersection interval has `start = max(s1, s2)` and `end = min(e1, e2)`. If `start > end`, intersection length is 0 (`DISJOINT`). If `start == end`, length is 1 (`TOUCHING`). Otherwise length is `end - start + 1`. Check `SUBSET` condition before `OVERLAPPING`!


````python
# 💡 Problem 5 Starter Code & Interactive Runner

def classify_interval_collision(s1: int, e1: int, s2: int, e2: int) -> str:
    """
    Determines the relationship between two closed intervals and their intersection length.
    """
    pass


def solve():
    try:
        parts = input().split()
        if parts:
            s1, e1, s2, e2 = map(int, parts)
            print(classify_interval_collision(s1, e1, s2, e2))
    except (EOFError, ValueError):
        pass

if __name__ == '__main__':
    # solve()
    pass
````

<summary>💡 Solution: Problem 5 — 1D Interval Overlap & Collision Classifier</summary>

```python
def classify_interval_collision(s1: int, e1: int, s2: int, e2: int) -> str:
    inter_start = max(s1, s2)
    inter_end = min(e1, e2)
    
    if inter_start > inter_end:
        return "DISJOINT (Intersection: 0)"
        
    length = inter_end - inter_start + 1
    
    if length == 1:
        # If one is a single point that lies inside the other, check subset:
        if (s1 >= s2 and e1 <= e2) or (s2 >= s1 and e2 <= e1):
            if s1 == e1 and s2 == e2:  # Both are the identical single point
                return "SUBSET (Intersection: 1)"
            # If one is single point and the other is length > 1, single point is inside -> SUBSET
            # Otherwise, if they only touch at boundaries:
            if (e1 == s2 and s1 < e1 and s2 < e2) or (e2 == s1 and s2 < e2 and s1 < e1):
                return "TOUCHING (Intersection: 1)"
            return "SUBSET (Intersection: 1)"
        return "TOUCHING (Intersection: 1)"
        
    # length > 1: check if one interval is completely inside the other
    if (s1 >= s2 and e1 <= e2) or (s2 >= s1 and e2 <= e1):
        return f"SUBSET (Intersection: {length})"
        
    return f"OVERLAPPING (Intersection: {length})"

def solve():
    s1, e1, s2, e2 = map(int, input().split())
    print(classify_interval_collision(s1, e1, s2, e2))
```

**Why it works:**
- Finding the intersection of any two intervals `[s1, e1]` and `[s2, e2]` in 1D is always `[max(s1, s2), min(e1, e2)]`.
- The number of discrete points in a closed integer range `[a, b]` is `b - a + 1`.
- A range is a subset of another if and only if `s1 >= s2 and e1 <= e2` (interval 1 inside 2) or `s2 >= s1 and e2 <= e1` (interval 2 inside 1).

**Trap to avoid:**

```python
def analyze_peaks_and_valleys(nums: list[int]) -> str:
    if len(nums) < 3:
        return "Peaks: 0, Valleys: 0, Max Amplitude: 0"
        
    peaks = 0
    valleys = 0
    extrema = []
    
    for i in range(1, len(nums) - 1):
        if nums[i] > nums[i - 1] and nums[i] > nums[i + 1]:
            peaks += 1
            extrema.append(nums[i])
        elif nums[i] < nums[i - 1] and nums[i] < nums[i + 1]:
            valleys += 1
            extrema.append(nums[i])
            
    max_amp = 0
    if peaks > 0 and valleys > 0 and len(extrema) >= 2:
        for k in range(len(extrema) - 1):
            diff = abs(extrema[k] - extrema[k + 1])
            if diff > max_amp:
                max_amp = diff
                
    return f"Peaks: {peaks}, Valleys: {valleys}, Max Amplitude: {max_amp}"

def solve():
```

```python
from collections import defaultdict

def find_largest_anagram_group(words: list[str]) -> str:
    if not words:
        return "Largest Group (0):"
        
    groups = defaultdict(list)
    first_seen = {}
    
    for idx, word in enumerate(words):
        key = "".join(sorted(word))
        groups[key].append(word)
        if key not in first_seen:
            first_seen[key] = idx
            
    # Sort keys by: size descending, then first appearance index ascending
    best_key = max(groups.keys(), key=lambda k: (len(groups[k]), -first_seen[k]))
    best_words = sorted(groups[best_key])
    
    return f"Largest Group ({len(best_words)}): {' '.join(best_words)}"

def solve():
    words = input().split()
    print(find_largest_anagram_group(words))
```

**Why it works:**
- Sorting characters of each word `"".join(sorted(word))` is the canonical signature for anagram detection (O(L log L) per word of length L).
- Python dictionaries preserve insertion order, but explicitly recording `first_seen[key]` and using `(len(groups[k]), -first_seen[k])` makes tie-breaking deterministic and bug-proof.
- Sorting the winning group alphabetically before printing satisfies exact judge output comparison.

**Trap to avoid:**
Be sure to sort the output words *alphabetically*, not in the order they appeared in the input string!
</details>
