# 🧩 Dynamic Programming — Build Up From Subproblems

DP is the technique that separates intermediate programmers from advanced ones.
It's not about memorizing solutions — it's about a **thinking framework**.

**The DP Mindset:**
1. Can I break this into smaller subproblems?
2. Do subproblems overlap (same subproblem computed multiple times)?
3. Does the optimal solution contain optimal solutions to subproblems?

If yes to all three → DP.

**Recognition Triggers:**
- "Count the number of ways"
- "Minimum/maximum cost/value"
- "Can you reach / is it possible"
- "Longest common / longest increasing"
- Problem has overlapping subproblems

---
## 📋 The DP Template

### Step 1: Define State
What does `dp[i]` (or `dp[i][j]`) represent?

### Step 2: Write Recurrence
How does `dp[i]` relate to smaller states?

### Step 3: Base Cases
What are the smallest states with known answers?

### Step 4: Computation Order
Fill smaller states before larger ones (bottom-up) or use recursion with memoization (top-down).

### Top-Down vs Bottom-Up

```python
# Example: Fibonacci — the simplest DP problem
from functools import lru_cache

# Top-Down (Memoization)
@lru_cache(maxsize=None)
def fib_top_down(n):
    if n <= 1: return n
    return fib_top_down(n - 1) + fib_top_down(n - 2)

# Bottom-Up (Tabulation)
def fib_bottom_up(n):
    if n <= 1: return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

# Space-Optimized (only need last 2 values)
def fib_optimized(n):
    if n <= 1: return n
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

print(f"fib(10) = {fib_top_down(10)} = {fib_bottom_up(10)} = {fib_optimized(10)}")  # 55
```

## 🔥 Problem: Climbing Stairs

```python
def climb_stairs(n):
    """How many ways to climb n stairs (1 or 2 steps at a time)?
    State: dp[i] = number of ways to reach step i
    Recurrence: dp[i] = dp[i-1] + dp[i-2] (come from 1 step back or 2 steps back)
    Base: dp[0] = 1, dp[1] = 1
    This is just Fibonacci! Time: O(n), Space: O(1)"""
    if n <= 2: return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b

print("Climbing Stairs:")
for n in [2, 3, 5, 10]:
    print(f"  n={n}: {climb_stairs(n)} ways")
```

## 🔥 Problem: House Robber

```python
def rob(nums):
    """Max money robbing houses (can't rob adjacent).
    State: dp[i] = max money robbing from houses 0..i
    Recurrence: dp[i] = max(dp[i-1], dp[i-2] + nums[i])
      Either skip house i (take dp[i-1]) or rob it (dp[i-2] + nums[i])
    Time: O(n), Space: O(1)"""
    if not nums: return 0
    if len(nums) == 1: return nums[0]
    prev2, prev1 = 0, 0
    for num in nums:
        prev2, prev1 = prev1, max(prev1, prev2 + num)
    return prev1

print("House Robber:")
print("  [1,2,3,1]:", rob([1,2,3,1]))        # 4 (rob house 0 and 2)
print("  [2,7,9,3,1]:", rob([2,7,9,3,1]))    # 12 (rob house 0, 2, 4)
```

## 🔥 Problem: Coin Change

```python
def coin_change(coins, amount):
    """Minimum coins to make amount.
    State: dp[i] = min coins to make amount i
    Recurrence: dp[i] = min(dp[i - coin] + 1) for each coin
    Base: dp[0] = 0
    Time: O(amount * len(coins)), Space: O(amount)"""
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    for i in range(1, amount + 1):
        for coin in coins:
            if coin <= i:
                dp[i] = min(dp[i], dp[i - coin] + 1)
    return dp[amount] if dp[amount] != float('inf') else -1

print("Coin Change:")
print("  coins=[1,5,11], amount=15:", coin_change([1,5,11], 15))  # 3 (5+5+5)
print("  coins=[2], amount=3:", coin_change([2], 3))               # -1
```

## 🔥 Problem: Longest Common Subsequence

```python
def lcs(text1, text2):
    """Length of longest common subsequence.
    State: dp[i][j] = LCS of text1[:i] and text2[:j]
    Recurrence:
      If text1[i-1] == text2[j-1]: dp[i][j] = dp[i-1][j-1] + 1
      Else: dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    Time: O(m*n), Space: O(m*n) (can be optimized to O(min(m,n)))"""
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    
    return dp[m][n]

print("Longest Common Subsequence:")
print("  'abcde', 'ace':", lcs("abcde", "ace"))       # 3 ("ace")
print("  'abc', 'abc':", lcs("abc", "abc"))            # 3
print("  'abc', 'def':", lcs("abc", "def"))            # 0
```

## 🔥 Problem: 0/1 Knapsack

```python
def knapsack(weights, values, capacity):
    """Maximum value with given capacity.
    State: dp[i][w] = max value using items 0..i-1 with capacity w
    Recurrence:
      Skip item i: dp[i][w] = dp[i-1][w]
      Take item i: dp[i][w] = dp[i-1][w - weights[i-1]] + values[i-1]
    Time: O(n*W), Space: O(n*W) → can optimize to O(W)"""
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        for w in range(capacity + 1):
            dp[i][w] = dp[i-1][w]  # Skip
            if weights[i-1] <= w:
                dp[i][w] = max(dp[i][w], dp[i-1][w - weights[i-1]] + values[i-1])
    
    return dp[n][capacity]

# Space-optimized version (1D array)
def knapsack_optimized(weights, values, capacity):
    dp = [0] * (capacity + 1)
    for i in range(len(weights)):
        for w in range(capacity, weights[i] - 1, -1):  # Reverse to avoid using same item twice!
            dp[w] = max(dp[w], dp[w - weights[i]] + values[i])
    return dp[capacity]

weights = [1, 3, 4, 5]
values = [1, 4, 5, 7]
capacity = 7
print(f"0/1 Knapsack: {knapsack(weights, values, capacity)}")  # 9
print(f"Optimized:    {knapsack_optimized(weights, values, capacity)}")  # 9
```

## 🔥 Problem: Longest Increasing Subsequence

```python
def lis(nums):
    """Length of longest increasing subsequence.
    DP: O(n²). Binary search: O(n log n)."""
    if not nums: return 0
    n = len(nums)
    dp = [1] * n
    for i in range(1, n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)
    return max(dp)

# O(n log n) version using patience sorting / bisect
from bisect import bisect_left

def lis_fast(nums):
    """Maintain a list of smallest tail elements. O(n log n)."""
    tails = []
    for num in nums:
        pos = bisect_left(tails, num)
        if pos == len(tails):
            tails.append(num)
        else:
            tails[pos] = num
    return len(tails)

print("LIS:")
print("  [10,9,2,5,3,7,101,18]:", lis([10,9,2,5,3,7,101,18]))      # 4
print("  Fast:", lis_fast([10,9,2,5,3,7,101,18]))                    # 4
```

## 🔥 Problem: Edit Distance

```python
def edit_distance(word1, word2):
    """Minimum operations (insert, delete, replace) to convert word1 to word2.
    State: dp[i][j] = edit distance of word1[:i] and word2[:j]
    Time: O(m*n), Space: O(m*n)"""
    m, n = len(word1), len(word2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(m + 1): dp[i][0] = i
    for j in range(n + 1): dp[0][j] = j
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i-1] == word2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = 1 + min(
                    dp[i-1][j],    # Delete
                    dp[i][j-1],    # Insert
                    dp[i-1][j-1]   # Replace
                )
    return dp[m][n]

print("Edit Distance:")
print("  'horse' → 'ros':", edit_distance("horse", "ros"))          # 3
print("  'intention' → 'execution':", edit_distance("intention", "execution"))  # 5
```

## 🔥 Problem: Partition Equal Subset Sum

```python
def can_partition(nums):
    """Can we partition into two subsets with equal sum?
    Reduces to: can we find a subset with sum = total/2?
    This is a 0/1 Knapsack problem!
    Time: O(n * sum/2), Space: O(sum/2)"""
    total = sum(nums)
    if total % 2 != 0:
        return False
    
    target = total // 2
    dp = [False] * (target + 1)
    dp[0] = True
    
    for num in nums:
        for j in range(target, num - 1, -1):  # Reverse!
            dp[j] = dp[j] or dp[j - num]
    
    return dp[target]

print("Partition Equal Subset Sum:")
print("  [1,5,11,5]:", can_partition([1,5,11,5]))  # True (1+5+5 = 11)
print("  [1,2,3,5]:", can_partition([1,2,3,5]))    # False
```

## 🔥 Problem: Burst Balloons

```python
def max_coins(nums):
    """Interval DP. State: dp[i][j] = max coins from bursting balloons i..j.
    Key trick: think about which balloon to burst LAST in each interval.
    Time: O(n³), Space: O(n²)"""
    nums = [1] + nums + [1]  # Add boundary balloons
    n = len(nums)
    dp = [[0] * n for _ in range(n)]
    
    for length in range(2, n):  # Window length
        for left in range(0, n - length):
            right = left + length
            for k in range(left + 1, right):  # k = last balloon to burst
                coins = nums[left] * nums[k] * nums[right]
                dp[left][right] = max(dp[left][right],
                                      dp[left][k] + coins + dp[k][right])
    
    return dp[0][n - 1]

print("Burst Balloons:")
print("  [3,1,5,8]:", max_coins([3,1,5,8]))  # 167
```

## 📋 DP Pattern Families

| Family | State | Examples |
|--------|-------|----------|
| **Linear** | `dp[i]` | Climbing Stairs, House Robber, LIS |
| **Grid** | `dp[i][j]` | Unique Paths, Min Path Sum |
| **String** | `dp[i][j]` on two strings | LCS, Edit Distance |
| **Interval** | `dp[i][j]` = best on range [i,j] | Burst Balloons, Matrix Chain |
| **Knapsack** | `dp[i][w]` | 0/1 Knapsack, Partition Subset |
| **Bitmask** | `dp[mask]` | TSP, Set Cover |

## 🏆 Summary

| Problem | Family | Time | Space |
|---------|--------|------|-------|
| Climbing Stairs | Linear | O(n) | O(1) |
| House Robber | Linear | O(n) | O(1) |
| Coin Change | Knapsack | O(n×W) | O(W) |
| LCS | String | O(m×n) | O(m×n) |
| 0/1 Knapsack | Knapsack | O(n×W) | O(W) |
| LIS | Linear | O(n log n) | O(n) |
| Edit Distance | String | O(m×n) | O(m×n) |
| Partition Subset | Knapsack | O(n×S) | O(S) |
| Burst Balloons | Interval | O(n³) | O(n²) |
