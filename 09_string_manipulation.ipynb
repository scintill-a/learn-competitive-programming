## 🔥 Problem: Valid Anagram

```python
from collections import Counter

def is_anagram(s, t):
    return Counter(s) == Counter(t)

# Sorting approach
def is_anagram_sort(s, t):
    return sorted(s) == sorted(t)

print("Anagram 'anagram', 'nagaram':", is_anagram("anagram", "nagaram"))  # True
print("Anagram 'rat', 'car':", is_anagram("rat", "car"))  # False
```

## 🔥 Problem: Longest Palindromic Substring

```python
def longest_palindrome(s):
    """Expand around center. Check both odd and even length palindromes.
    Time: O(n²), Space: O(1)"""
    if not s: return ""
    start, max_len = 0, 1
    
    def expand(left, right):
        nonlocal start, max_len
        while left >= 0 and right < len(s) and s[left] == s[right]:
            if right - left + 1 > max_len:
                start = left
                max_len = right - left + 1
            left -= 1
            right += 1
    
    for i in range(len(s)):
        expand(i, i)      # Odd length
        expand(i, i + 1)  # Even length
    
    return s[start:start + max_len]

print("Longest Palindrome:")
print("  'babad':", longest_palindrome("babad"))  # "bab" or "aba"
print("  'cbbd':", longest_palindrome("cbbd"))    # "bb"
```

## 🔥 Problem: Count Palindromic Substrings

```python
def count_palindromes(s):
    """Same expand-around-center technique. Time: O(n²)"""
    count = 0
    def expand(left, right):
        nonlocal count
        while left >= 0 and right < len(s) and s[left] == s[right]:
            count += 1
            left -= 1
            right += 1
    for i in range(len(s)):
        expand(i, i)
        expand(i, i + 1)
    return count

print("Count Palindromes in 'abc':", count_palindromes("abc"))    # 3
print("Count Palindromes in 'aaa':", count_palindromes("aaa"))    # 6
```

## 🔥 Problem: Encode and Decode Strings

```python
def encode(strs):
    """Encode: length + '#' + string for each word."""
    return ''.join(f"{len(s)}#{s}" for s in strs)

def decode(s):
    result = []
    i = 0
    while i < len(s):
        j = s.index('#', i)
        length = int(s[i:j])
        result.append(s[j+1:j+1+length])
        i = j + 1 + length
    return result

original = ["hello", "world", "foo#bar"]
encoded = encode(original)
decoded = decode(encoded)
print(f"Encode: {original} → '{encoded}'")
print(f"Decode: '{encoded}' → {decoded}")
assert original == decoded
```

## 🔥 String + DP: Longest Palindromic Subsequence

```python
def longest_palindrome_subseq(s):
    """DP: LPS(s) = LCS(s, reverse(s)). Time: O(n²)"""
    n = len(s)
    rev = s[::-1]
    dp = [[0] * (n + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        for j in range(1, n + 1):
            if s[i-1] == rev[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    return dp[n][n]

print("Longest Palindromic Subsequence:")
print("  'bbbab':", longest_palindrome_subseq("bbbab"))  # 4 ("bbbb")
print("  'cbbd':", longest_palindrome_subseq("cbbd"))    # 2 ("bb")
```

## 🔥 Problem: Zigzag Conversion

```python
def zigzag_convert(s, num_rows):
    """Simulate the zigzag pattern. Time: O(n)"""
    if num_rows == 1 or num_rows >= len(s): return s
    rows = [''] * num_rows
    current_row = 0
    going_down = False
    for char in s:
        rows[current_row] += char
        if current_row == 0 or current_row == num_rows - 1:
            going_down = not going_down
        current_row += 1 if going_down else -1
    return ''.join(rows)

print("Zigzag 'PAYPALISHIRING', 3:", zigzag_convert("PAYPALISHIRING", 3))  # PAHNAPLSIIGYIR
```
