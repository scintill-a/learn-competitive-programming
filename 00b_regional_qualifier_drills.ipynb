## 1. 💀 The 4 Deadly Sins of Regionals

1. **Presentation Error:** You printed `"Case 1:"` instead of `"Case #1:"`. You lose. Always copy-paste the exact output format.
2. **Type Error:** You tried to add `"5" + 3`. Python crashes. Always convert input strings to `int()` immediately.
3. **Off-By-One Error:** You looped `range(1, N)` instead of `range(1, N+1)`. You missed the last item.
4. **Infinite Loop / EOF Crash:** The platform didn't tell you how many lines of input there were, and your code crashed trying to read a line that didn't exist.

## 2. 📥 Deep Dive: Input/Output & Parsing

In regionals, and especially when building standalone `.exe` tools, data comes as text from the user's keyboard. You must be an absolute master at pulling that text in, handling weird spacing, and knowing when to stop.

If you are running an `.exe` and typing input manually, **you must know how to send an EOF (End Of File) signal** to tell Python you are done typing:
- **Windows:** `Ctrl + Z` then `Enter`
- **Mac/Linux:** `Ctrl + D`


````python
def io_masterclass():
    """
    DO NOT run this cell directly as it requires manual input, 
    but study these patterns. This is how you start EVERY script.
    """
    import sys
    
    # === PATTERN 1: The Standard N Test Cases ===
    # Input:
    # 2
    # apple
    # banana
    def pattern_1():
        T = int(input().strip())  # Read 2, convert to integer
        for _ in range(T):
            word = input().strip()
            print(word.upper())

    # === PATTERN 2: Space-Separated Numbers ===
    # Input: 10 20 30 40 50
    def pattern_2():
        # input() gets "10 20 30 40 50"
        # .split() makes it ["10", "20", "30", "40", "50"]
        # map(int, ...) applies int() to every string
        # list(...) turns the map object back into a real list
        nums = list(map(int, input().split()))
        print(sum(nums))

    # === PATTERN 3: The Infinite Interactive Loop (Interactive .exe) ===
    # If you want the program to keep asking for input until the user types "exit"
    def pattern_3():
        while True:
            cmd = input("Enter command (or 'exit' to quit): ").strip()
            if cmd.lower() == 'exit':
                break
            print(f"Executing: {cmd}")

    # === PATTERN 4: Unknown Number of Lines (EOF Signal Required) ===
    # The platform (or user) just feeds you lines until they press Ctrl+D / Ctrl+Z
    def pattern_4():
        while True:
            try:
                line = input().strip()
                if not line: continue  # Skip empty lines
                print(f"Processed: {line}")
            except EOFError:
                break  # User pressed Ctrl+D (Mac/Linux) or Ctrl+Z (Windows)

    # === PATTERN 5: The "Fast IO" Bulk Reader (For Massive Data) ===
    # If you have 100,000 numbers mixed across multiple lines and spaces,
    # calling input() 100,000 times is too slow. Read EVERYTHING at once.
    def pattern_5():
        # sys.stdin.read() reads the ENTIRE file/input until EOF
        # .split() automatically breaks it apart by ALL whitespace (spaces AND newlines)
        tokens = sys.stdin.read().split()
        
        # Now you just pop tokens off the list as you need them
        if not tokens: return
        n = int(tokens[0])
        nums = [int(x) for x in tokens[1:n+1]]
        print(f"Read {n} numbers: {sum(nums)}")

    # === PATTERN 6: Reading a 2D Grid ===
    # Input:
    # 3 3
    # 1 2 3
    # 4 5 6
    # 7 8 9
    def pattern_6():
        r, c = map(int, input().split())
        grid = []
        for _ in range(r):
            row = list(map(int, input().split()))
            grid.append(row)
        print(f"Center element is: {grid[1][1]}")

    # === PATTERN 7: Parsing Comma-Separated Key-Value Pairs ===
    # Input: name=alice, age=25, role=admin
    def pattern_7():
        line = input().strip()
        data = {}
        for pair in line.split(','):
            key, val = pair.split('=')
            data[key.strip()] = val.strip()
        print(data['name'])  # 'alice'
````

## 4. 🔀 Deep Dive: Conditionals (The Tricky Ones)

Regionals test if you know the difference between `if/if/if` and `if/elif/else`. 

*   `if / if / if`: ALL conditions are checked. Multiple blocks can run.
*   `if / elif / else`: Only the FIRST true block runs. They are mutually exclusive.

````python
def analyze_number(n):
    print(f"\nAnalyzing {n}:")
    
    # Example of if/if/if (Multiple things can be true)
    if n % 2 == 0:
        print("- It is even")
    if n > 0:
        print("- It is positive")
    if n % 10 == 0:
        print("- It ends in zero")
        
    # Example of if/elif/else (Categorization, only one runs)
    if n > 100:
        print("Category: Huge")
    elif n > 10:
        print("Category: Big")
    elif n == 0:
        print("Category: Zero")
    else:
        print("Category: Small/Negative")

analyze_number(20)
analyze_number(150)

print("\n--- Compound Conditions (AND / OR) ---")
# Problem: Is a person eligible for a discount?
# Rules: Must be (under 18 OR over 65) AND have a membership card.
def discount_eligibility(age, has_card):
    # Use parentheses to group logic explicitly!
    if (age < 18 or age > 65) and has_card:
        return "Eligible"
    return "Not Eligible"

print(f"16 year old, no card: {discount_eligibility(16, False)}")
print(f"16 year old, has card: {discount_eligibility(16, True)}")
print(f"30 year old, has card: {discount_eligibility(30, True)}")
print(f"70 year old, has card: {discount_eligibility(70, True)}")
````

## 5. 🔁 Deep Dive: Loops, Aggregations & Neighbors

You need to be able to loop over arrays, sum them up, find extremes, and compare an element to the one next to it.

````python
# === THE MASTER LOOP TEMPLATE ===
grades = [85, 92, 78, 90, 88]

# We want to find: Total, Average, Highest, Lowest, and Count of grades > 85
total = 0
highest = float('-inf')  # Start with negative infinity
lowest = float('inf')    # Start with positive infinity
count_high = 0

for grade in grades:
    total += grade
    
    if grade > highest:
        highest = grade
        
    if grade < lowest:
        lowest = grade
        
    if grade > 85:
        count_high += 1

average = total / len(grades)

print(f"Grades: {grades}")
print(f"Total: {total}, Average: {average:.1f}")
print(f"Highest: {highest}, Lowest: {lowest}")
print(f"Count > 85: {count_high}")

# === LOOPING OVER NEIGHBORS (CRITICAL SKILL) ===
print("\n--- Comparing Neighbors ---")
# Problem: Count how many times the array INCREASES from one number to the next.
nums = [10, 15, 12, 18, 20, 19]
increases = 0

# Notice the loop goes up to len(nums) - 1. Why? Because we check i and i+1.
# If we went to the end, nums[i+1] would cause an IndexError.
for i in range(len(nums) - 1):
    if nums[i+1] > nums[i]:
        increases += 1
        print(f"Increase found: {nums[i]} -> {nums[i+1]}")

print(f"Total increases: {increases}")

# === NESTED LOOPS (For grids or pairs) ===
print("\n--- Nested Loops ---")
# Print a 3x3 coordinate grid
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print() # Newline at the end of each row
````

## 6. 🪢 Deep Dive: Strings & Arrays (Regional Style)

Splitting, slicing, joining, and building strings dynamically.

````python
text = "The Quick Brown Fox"

print("Original:", text)
print("Reversed:", text[::-1])
print("Every second char:", text[::2])

words = text.split()  # Splits by space into a list
print("\nWords list:", words)

# Join them back with a hyphen
hyphenated = "-".join(words)
print("Joined with hyphens:", hyphenated)

# === BUILDING A STRING CHARACTER BY CHARACTER ===
# Problem: Keep only the uppercase letters, but convert them to lowercase.
result = []
for char in text:
    if char.isupper():
        result.append(char.lower())

final_string = "".join(result)
print("\nExtracted upper->lower:", final_string)

# === PREFIX SUMS (Advanced Regional Trick) ===
# If a problem asks "sum of a subarray from index L to R" multiple times,
# a basic loop inside a loop will TIME OUT. You must use Prefix Sums.
print("\n--- Prefix Sums ---")
arr = [2, 4, 6, 8, 10]
# Create an array where prefix[i] is the sum of everything up to i
prefix = [0] * (len(arr) + 1)
for i in range(len(arr)):
    prefix[i+1] = prefix[i] + arr[i]
    
print(f"Array: {arr}")
print(f"Prefix Sum Array: {prefix}")
# Now, sum of arr[1 to 3] (which is 4+6+8 = 18)
# We simply do prefix[4] - prefix[1] = 20 - 2 = 18. Instant O(1) time!
L = 1
R = 3
subarray_sum = prefix[R+1] - prefix[L]
print(f"Sum from index {L} to {R}: {subarray_sum}")
````

    s = input()
