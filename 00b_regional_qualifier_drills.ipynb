## 1. 💀 The 4 Deadly Sins of Regionals

These four mistakes account for the vast majority of Wrong Answer and Runtime Error verdicts in regionals. Memorize them. Tattoo them on your brain.

### Sin #1: Presentation Error 🖨️
You printed `"Case 1:"` instead of `"Case #1:"`. You printed an extra space at the end. You forgot the newline. **The judge is a heartless robot** — it compares your output character-by-character.

### Sin #2: Type Error 💥
You tried `"5" + 3`. Python crashes. Or worse: `"5" + "3"` gives `"53"`, not `8`. **Always convert input strings to `int()` or `float()` IMMEDIATELY** after reading them.

### Sin #3: Off-By-One Error 🔢
You looped `range(1, N)` instead of `range(1, N+1)`. You accessed `arr[N]` on an array of length `N` (valid indices are `0` to `N-1`). These are silent killers — your code runs fine but gives wrong answers.

### Sin #4: Infinite Loop / EOF Crash 💀
The platform didn't tell you how many lines of input there were, and your code crashed trying to read a line that didn't exist. Or your `while` loop never terminates because you forgot to update the condition variable.

````python
# === LIVE DEMONSTRATION OF EACH SIN ===

# SIN 1: Presentation Error — spot the difference!
print("Wrong:", "Case 1: answer is 42")
print("Right:", "Case #1: answer is 42")
# One character difference. One is ACCEPTED, the other is WRONG ANSWER.
# RULE: Always copy-paste the exact output format from the problem statement.

# SIN 2: Type Error — the silent killer
user_input = "5"       # input() ALWAYS returns a string
# bad  = user_input + 3   # ❌ TypeError: can't add str + int
# sneaky = user_input + "3"  # ❌ Gives "53" not 8
good = int(user_input) + 3  # ✅ Gives 8
print(f"\nType demo: '{user_input}' + 3 = {good} (must convert first!)")

# SIN 3: Off-by-one — the most common bug in programming
print(f"\nOff-by-one demo:")
print(f"  range(1, 5)   = {list(range(1, 5))}     ← Misses 5!")
print(f"  range(1, 5+1) = {list(range(1, 5+1))}  ← Includes 5!")
print(f"  range(5)      = {list(range(5))}     ← Starts at 0, not 1!")

arr = [10, 20, 30, 40, 50]
print(f"\n  arr = {arr}")
print(f"  len(arr) = {len(arr)}")
print(f"  arr[0] = {arr[0]}, arr[4] = {arr[4]}")
# arr[5] would crash with IndexError! Valid indices: 0 to len(arr)-1

# SIN 4: EOF handling
print("\nEOF pattern: use try/except EOFError for unknown number of lines")
print("Or read all at once with sys.stdin.read().split()") 
````

## 2. 📥 Deep Dive: Input/Output & Parsing

In regionals, and especially when building standalone `.exe` tools, data comes as text from the user's keyboard. You must be an absolute master at pulling that text in, handling weird spacing, and knowing when to stop.

If you are running an `.exe` and typing input manually, **you must know how to send an EOF (End Of File) signal** to tell Python you are done typing:
- **Windows:** `Ctrl + Z` then `Enter`
- **Mac/Linux:** `Ctrl + D`

### ⚡ The Golden Rule of Input
**`input()` always returns a string.** Every. Single. Time. Even if the user types `42`, you get the string `"42"`, not the integer `42`. This is why Sin #2 exists.

### 📝 The `input()` → Process → `print()` Pipeline
Every competitive programming solution follows this exact pattern:
1. **Read** raw text with `input()` or `sys.stdin`
2. **Parse** it into the right types (`int`, `float`, `list`, etc.)
3. **Compute** the answer
4. **Print** the result in the exact required format

````python
def io_masterclass():
    # DO NOT run this cell directly as it requires manual input,
    # but study these patterns. This is how you start EVERY script.
    import sys
    
    # === PATTERN 1: The Standard N Test Cases ===
    # This is THE most common input format in competitive programming.
    # First line tells you how many test cases follow.
    #
    # Input:
    # 2
    # apple
    # banana
    #
    # Output:
    # APPLE
    # BANANA
    def pattern_1():
        T = int(input().strip())  # Read "2", convert to integer
        for _ in range(T):        # _ means "I don't need this variable"
            word = input().strip()
            print(word.upper())

    # === PATTERN 2: Space-Separated Numbers on One Line ===
    # Very common: all data on a single line, separated by spaces.
    #
    # Input: 10 20 30 40 50
    # Output: 150
    def pattern_2():
        # Step by step:
        # input()           → "10 20 30 40 50"     (one big string)
        # .split()          → ["10", "20", "30", "40", "50"]  (list of strings)
        # map(int, ...)     → map object that converts each to int on demand
        # list(...)         → [10, 20, 30, 40, 50]  (actual list of integers)
        nums = list(map(int, input().split()))
        print(sum(nums))

    # === PATTERN 3: Two Values on One Line ===
    # When a problem says "the first line contains N and M"
    #
    # Input: 5 3
    def pattern_3():
        # Unpacking directly into variables (no list needed)
        n, m = map(int, input().split())
        # n = 5, m = 3 — both are already integers!
        print(f"n={n}, m={m}")

    # === PATTERN 4: The Infinite Interactive Loop (Interactive .exe) ===
    # If you want the program to keep asking for input until the user types "exit"
    def pattern_4():
        while True:
            cmd = input("Enter command (or 'exit' to quit): ").strip()
            if cmd.lower() == 'exit':
                break
            print(f"Executing: {cmd}")

    # === PATTERN 5: Unknown Number of Lines (EOF Signal Required) ===
    # The platform (or user) just feeds you lines until EOF.
    # This is common in UVa Online Judge problems.
    def pattern_5():
        while True:
            try:
                line = input().strip()
                if not line: continue  # Skip empty lines
                print(f"Processed: {line}")
            except EOFError:
                break  # No more input

    # === PATTERN 6: The "Fast IO" Bulk Reader (For Massive Data) ===
    # If you have 100,000 numbers mixed across multiple lines and spaces,
    # calling input() 100,000 times is too slow. Read EVERYTHING at once.
    #
    # This is the COMPETITIVE PROGRAMMER'S SECRET WEAPON.
    def pattern_6():
        # sys.stdin.read() reads the ENTIRE input until EOF — one huge string
        # .split() breaks it by ALL whitespace (spaces AND newlines)
        tokens = sys.stdin.read().split()
        
        # Now use an index pointer to walk through tokens
        idx = 0
        n = int(tokens[idx]); idx += 1
        nums = [int(tokens[idx + i]) for i in range(n)]; idx += n
        print(f"Read {n} numbers: {sum(nums)}")

    # === PATTERN 7: Reading a 2D Grid ===
    # Input:
    # 3 3
    # 1 2 3
    # 4 5 6
    # 7 8 9
    def pattern_7():
        r, c = map(int, input().split())
        grid = []
        for _ in range(r):
            row = list(map(int, input().split()))
            grid.append(row)
        # One-liner version: grid = [list(map(int, input().split())) for _ in range(r)]
        print(f"Center element is: {grid[r//2][c//2]}")

    # === PATTERN 8: Parsing Comma-Separated Key-Value Pairs ===
    # Input: name=alice, age=25, role=admin
    def pattern_8():
        line = input().strip()
        data = {}
        for pair in line.split(','):
            key, val = pair.split('=')
            data[key.strip()] = val.strip()
        print(data['name'])  # 'alice'

    # === PATTERN 9: Reading Until a Sentinel Value ===
    # Some problems say "read lines until you see 0" or "until you see END"
    #
    # Input:
    # 5
    # 3
    # 8
    # 0       ← stop here, don't process this
    def pattern_9():
        while True:
            n = int(input())
            if n == 0:
                break
            print(f"Processing: {n}")
````

### 🧠 Input Parsing Cheatsheet

| You receive | You need | Code |
|------------|----------|------|
| `"42"` | `int` | `int(input())` |
| `"3.14"` | `float` | `float(input())` |
| `"10 20 30"` | `list[int]` | `list(map(int, input().split()))` |
| `"5 3"` | two ints | `a, b = map(int, input().split())` |
| `"hello world"` | `str` | `input().strip()` |
| `"a,b,c"` | `list[str]` | `input().split(',')` |
| `"1 2\n3 4"` (2 lines) | 2D grid | `[list(map(int, input().split())) for _ in range(2)]` |
| Multiple lines until EOF | all lines | `sys.stdin.read().split()` |

### ⚠️ Common Input Traps

1. **Trailing whitespace**: Always use `.strip()` on string inputs
2. **Mixed delimiters**: `"a, b, c"` — note the space after comma! Use `.split(', ')` not `.split(',')`  
3. **Empty lines**: `input()` on an empty line returns `""`, which `int("")` will crash on
4. **`input()` vs `sys.stdin`**: `input()` strips the trailing newline automatically; `sys.stdin.readline()` does NOT

## 3. 🧮 Deep Dive: Type Conversions & Math

Regionals test your ability to juggle strings, integers, floats, and ASCII codes. This section covers every conversion and math trick you'll ever need at the regional level.

### The Type Hierarchy
```
str ←→ int ←→ float
 ↕         ↕
chr ←→ ord (ASCII)
```
Everything in competitive programming flows through these conversions.

````python
# === TYPE CONVERSION CHEATSHEET ===
print("=== String ↔ Number ===")
s = "123"
print(f"  str→int:   int('123')    = {int(s)}")
print(f"  str→float: float('3.14') = {float('3.14')}")
print(f"  int→str:   str(456)      = '{str(456)}'  (length {len(str(456))})")
print(f"  float→int: int(3.999)    = {int(3.999)}   ← TRUNCATES, does NOT round!")
print(f"  float→int: int(-3.7)     = {int(-3.7)}   ← Truncates toward zero!")

# CRITICAL: int() truncates toward zero, NOT toward negative infinity!
# For floor division toward negative infinity, use math.floor() or //
import math
print(f"\n  int(-3.7)        = {int(-3.7)}    ← toward zero")
print(f"  math.floor(-3.7) = {math.floor(-3.7)}   ← toward -infinity")

# === DIGIT EXTRACTION (Super common in regionals) ===
print("\n=== Digit Extraction ===")
n = 4827
print(f"  Number: {n}")
print(f"  Last digit:        {n} % 10   = {n % 10}")
print(f"  Remove last digit: {n} // 10  = {n // 10}")
print(f"  Number of digits:  len(str({n})) = {len(str(n))}")
print(f"  Sum of digits:     {sum(int(d) for d in str(n))}")
print(f"  Digits as list:    {[int(d) for d in str(n)]}")
print(f"  Reverse number:    {int(str(n)[::-1])}")

# === DIVISION VARIETIES ===
print("\n=== Division ===")
a, b = 17, 5
print(f"  {a} / {b}  = {a / b}    (true division → always float)")
print(f"  {a} // {b} = {a // b}      (floor division → rounds DOWN)")
print(f"  {a} % {b}  = {a % b}      (modulo → remainder)")
print(f"  divmod({a}, {b}) = {divmod(a, b)}  (quotient AND remainder in one call!)")

# IMPORTANT: Floor division with negatives
print(f"\n  -17 // 5 = {-17 // 5}   ← Floors toward -infinity, NOT toward zero!")
print(f"  -17 % 5  = {-17 % 5}    ← In Python, result has sign of divisor")

# === MODULO APPLICATIONS (Extremely common) ===
print("\n=== Modulo Applications ===")
# 1. Cyclic patterns (days of week, circular arrays)
today = 2  # Tuesday (0=Sun, 1=Mon, 2=Tue...)
future_day = (today + 100) % 7
print(f"  Day of week in 100 days: ({today} + 100) % 7 = {future_day} (Thursday)")

# 2. Check divisibility
print(f"  Is 15 divisible by 3? {15 % 3 == 0}")
print(f"  Is 15 divisible by 4? {15 % 4 == 0}")

# 3. Even/odd check
print(f"  Is 7 even? {7 % 2 == 0}, Is 8 even? {8 % 2 == 0}")

# 4. Wrap around an array index
arr = [10, 20, 30, 40, 50]
idx = 7
print(f"  arr[{idx} % {len(arr)}] = arr[{idx % len(arr)}] = {arr[idx % len(arr)]}")

# === ROUNDING & ABSOLUTE VALUE ===
print("\n=== Rounding ===")
print(f"  round(3.5)   = {round(3.5)}   ← Python uses BANKER'S ROUNDING (to even)!")
print(f"  round(4.5)   = {round(4.5)}   ← rounds to 4, not 5!")
print(f"  round(3.6)   = {round(3.6)}")
print(f"  round(3.456, 2) = {round(3.456, 2)}  ← round to 2 decimal places")
print(f"  abs(-7)      = {abs(-7)}")
print(f"  abs(5 - 10)  = {abs(5 - 10)}  ← Distance between numbers")

# === ASCII CODES (Crucial for string problems) ===
print("\n=== ASCII Codes ===")
# chr() converts number → letter. ord() converts letter → number
print(f"  ord('A') = {ord('A')}, ord('Z') = {ord('Z')}")
print(f"  ord('a') = {ord('a')}, ord('z') = {ord('z')}")
print(f"  ord('0') = {ord('0')}, ord('9') = {ord('9')}")

# The 0-25 alphabet index trick:
letter = 'e'
index = ord(letter) - ord('a')
print(f"\n  '{letter}' is at alphabet index {index}")

# Caesar cipher (shift by 3):
shifted = chr(ord('a') + ((index + 3) % 26))
print(f"  Shifting '{letter}' by 3 → '{shifted}'")

# Check if character is a digit without .isdigit():
c = '7'
print(f"\n  Is '{c}' a digit? {ord('0') <= ord(c) <= ord('9')}")

# === BASE CONVERSIONS ===
print("\n=== Base Conversions ===")
n = 255
print(f"  Decimal {n}:")
print(f"    Binary:  bin({n})  = {bin(n)}")
print(f"    Octal:   oct({n})  = {oct(n)}")
print(f"    Hex:     hex({n})  = {hex(n)}")
print(f"    Binary (clean): {bin(n)[2:]}")
print(f"    Hex (clean):    {hex(n)[2:].upper()}")
print(f"  Back: int('FF', 16) = {int('FF', 16)}")
print(f"  Back: int('11111111', 2) = {int('11111111', 2)}")

# === POWER & EXPONENTIATION ===
print("\n=== Powers ===")
print(f"  2**10 = {2**10}")
print(f"  pow(2, 10, 1000) = {pow(2, 10, 1000)}  ← modular exponentiation (fast!)")
````

## 4. 🔀 Deep Dive: Conditionals (The Tricky Ones)

Regionals test if you know the difference between `if/if/if` and `if/elif/else`, whether you can handle compound conditions, and whether you fall for the classic logical traps.

### The Two Types of Conditional Chains

| Pattern | Behavior | Use When |
|---------|----------|----------|
| `if / if / if` | ALL conditions checked independently. Multiple blocks can run. | Checking multiple independent properties |
| `if / elif / else` | Only the FIRST true block runs. Mutually exclusive. | Categorizing into exactly one group |

### ⚠️ Common Conditional Traps
1. **Using `==` on floats**: `0.1 + 0.2 == 0.3` is `False`! Use `abs(a - b) < 1e-9`
2. **Forgetting `elif`**: Using `if/if` when you meant `if/elif` — multiple blocks run
3. **Short-circuit gotcha**: `x != 0 and 10/x > 5` — the order matters! If x is 0 and you swap the conditions, you get ZeroDivisionError

````python
# === if/if/if vs if/elif/else — THE CRITICAL DIFFERENCE ===
def analyze_number(n):
    print(f"\nAnalyzing {n}:")
    
    # if/if/if: Multiple things can be true simultaneously
    if n % 2 == 0:
        print("  ✓ It is even")
    if n > 0:
        print("  ✓ It is positive")
    if n % 10 == 0:
        print("  ✓ It ends in zero")
        
    # if/elif/else: Categorization — only ONE runs
    if n > 100:
        print("  Category: Huge")
    elif n > 10:
        print("  Category: Big")
    elif n == 0:
        print("  Category: Zero")
    else:
        print("  Category: Small/Negative")

analyze_number(20)   # even + positive + ends in zero + Big
analyze_number(150)  # even + positive + Huge
analyze_number(-5)   # Small/Negative (nothing from the if/if/if block)

# === COMPOUND CONDITIONS (AND / OR) ===
print("\n--- Compound Conditions ---")
# Problem: Is a person eligible for a discount?
# Rules: Must be (under 18 OR over 65) AND have a membership card.
def discount_eligibility(age, has_card):
    # ALWAYS use parentheses to make grouping explicit!
    if (age < 18 or age > 65) and has_card:
        return "Eligible"
    return "Not Eligible"

print(f"16, no card:  {discount_eligibility(16, False)}")
print(f"16, has card: {discount_eligibility(16, True)}")
print(f"30, has card: {discount_eligibility(30, True)}")
print(f"70, has card: {discount_eligibility(70, True)}")

# === CHAINED COMPARISONS (Python superpower!) ===
print("\n--- Chained Comparisons ---")
x = 15
# Instead of: if x >= 10 and x <= 20:
# Python lets you write:
print(f"Is {x} between 10 and 20? {10 <= x <= 20}")  # True!
print(f"Is {x} between 1 and 10?  {1 <= x <= 10}")   # False!
# This works with any comparison: a < b < c, a <= b < c, etc.

# === TERNARY EXPRESSIONS (One-line if/else) ===
print("\n--- Ternary Expressions ---")
score = 85
# Long form:
# if score >= 60:
#     result = "PASS"
# else:
#     result = "FAIL"
# Short form:
result = "PASS" if score >= 60 else "FAIL"
print(f"Score {score}: {result}")

# Nested ternary (use sparingly — can get unreadable):
grade = "A" if score >= 90 else "B" if score >= 80 else "C" if score >= 70 else "F"
print(f"Score {score}: Grade {grade}")

# === TRUTHINESS & FALSINESS ===
print("\n--- Truthiness ---")
# These are all FALSY in Python (evaluate to False in an if statement):
falsy_values = [False, 0, 0.0, "", [], {}, set(), None]
print("Falsy values:", [repr(v) for v in falsy_values])
# EVERYTHING ELSE is truthy.
# This means you can write:
my_list = [1, 2, 3]
if my_list:          # Instead of: if len(my_list) > 0:
    print("List is not empty!")
name = ""
if not name:         # Instead of: if name == "":
    print("Name is empty!")

# === SHORT-CIRCUIT EVALUATION ===
print("\n--- Short-Circuit Evaluation ---")
# Python stops evaluating as soon as the result is determined.
# This is USEFUL for guard conditions:
x = 0
# Safe: checks x != 0 FIRST. If False, Python never evaluates 10/x.
if x != 0 and 10/x > 5:
    print("Big quotient!")
else:
    print("x is zero or quotient is small")
# If you wrote: if 10/x > 5 and x != 0: ← CRASH! ZeroDivisionError
````

## 5. 🔁 Deep Dive: Loops, Aggregations & Neighbors

You need to be able to loop over arrays, sum them up, find extremes, compare neighbors, and handle 2D grids. Master these patterns and you can solve 80% of regional problems.

### The Loop Pattern Catalog

| Pattern | When to Use |
|---------|-------------|
| `for x in arr` | Process each element (don't need index) |
| `for i in range(len(arr))` | Need the index |
| `for i, x in enumerate(arr)` | Need BOTH index and value |
| `for i in range(len(arr)-1)` | Compare neighboring elements |
| `for a, b in zip(arr1, arr2)` | Process two arrays in parallel |
| `while condition:` | Loop until a condition changes |

````python
# === THE MASTER AGGREGATION TEMPLATE ===
# This pattern solves DOZENS of regional problems.
grades = [85, 92, 78, 90, 88]

# We want: Total, Average, Highest, Lowest, Count > 85
total = 0
highest = float('-inf')  # Start with negative infinity (any number beats this)
lowest = float('inf')    # Start with positive infinity (any number beats this)
count_high = 0

for grade in grades:
    total += grade
    if grade > highest:
        highest = grade
    if grade < lowest:
        lowest = grade
    if grade > 85:
        count_high += 1

average = total / len(grades)

print(f"Grades: {grades}")
print(f"Total: {total}, Average: {average:.1f}")
print(f"Highest: {highest}, Lowest: {lowest}")
print(f"Count > 85: {count_high}")

# === enumerate() — GET INDEX AND VALUE TOGETHER ===
print("\n--- enumerate() ---")
fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits):
    print(f"  Index {i}: {fruit}")
# This replaces the ugly: for i in range(len(fruits)): fruit = fruits[i]

# === zip() — PARALLEL ITERATION ===
print("\n--- zip() ---")
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f"  {name}: {score}")

# === LOOPING OVER NEIGHBORS (CRITICAL SKILL) ===
print("\n--- Comparing Neighbors ---")
# Problem: Count how many times the array INCREASES from one element to the next.
nums = [10, 15, 12, 18, 20, 19]
increases = 0

# The loop goes to len(nums) - 1. Why? Because we check i and i+1.
# If we went to the end, nums[i+1] would cause an IndexError.
for i in range(len(nums) - 1):
    if nums[i+1] > nums[i]:
        increases += 1
        print(f"  Increase: {nums[i]} → {nums[i+1]}")

print(f"Total increases: {increases}")

# === WHILE LOOPS WITH BREAK/CONTINUE ===
print("\n--- while with break/continue ---")
# break: exit the loop immediately
# continue: skip to the next iteration
n = 100
while n > 1:
    if n % 2 == 0:
        n //= 2
    else:
        n = 3 * n + 1
    # This is the Collatz Conjecture!
print(f"Collatz sequence reached {n}")

# === NESTED LOOPS (For grids and pairs) ===
print("\n--- Nested Loops: Coordinate Grid ---")
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print()  # Newline at the end of each row

# === FINDING PAIRS (Nested loop pattern) ===
print("\n--- All Pairs ---")
arr = [1, 2, 3, 4]
pair_count = 0
for i in range(len(arr)):
    for j in range(i + 1, len(arr)):  # j starts AFTER i to avoid duplicates
        if arr[i] + arr[j] == 5:
            print(f"  Pair found: ({arr[i]}, {arr[j]})")
            pair_count += 1
print(f"Pairs summing to 5: {pair_count}")

# === LIST COMPREHENSIONS (Pythonic speed) ===
print("\n--- List Comprehensions ---")
# Long form:
squares = []
for x in range(1, 6):
    squares.append(x ** 2)
print(f"Squares (loop): {squares}")

# Short form (same result, often faster):
squares = [x ** 2 for x in range(1, 6)]
print(f"Squares (comp): {squares}")

# With filtering:
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(f"Even squares: {even_squares}")
````

## 6. 🪢 Deep Dive: Strings & Arrays (Regional Style)

Splitting, slicing, joining, building strings, counting frequencies, and prefix sums. These are the bread and butter of regional problems.

````python
# === STRING OPERATIONS MASTER CLASS ===
text = "The Quick Brown Fox"

print("=== Slicing ===")
print(f"  Original:        '{text}'")
print(f"  Reversed:        '{text[::-1]}'")
print(f"  First 3:         '{text[:3]}'")
print(f"  Last 3:          '{text[-3:]}'")
print(f"  Every 2nd char:  '{text[::2]}'")
print(f"  Chars 4 to 9:    '{text[4:9]}'")

print("\n=== Split & Join ===")
words = text.split()  # Splits by whitespace into a list
print(f"  Words: {words}")
print(f"  Hyphenated: '{'-'.join(words)}'")
print(f"  Split 'a,b,c': {'a,b,c'.split(',')}")
print(f"  Split 'a, b, c': {'a, b, c'.split(', ')}")  # Note the space!

print("\n=== Essential String Methods ===")
s = "  Hello, World!  "
print(f"  strip():    '{s.strip()}'")
print(f"  lower():    '{s.strip().lower()}'")
print(f"  upper():    '{s.strip().upper()}'")
print(f"  replace():  '{'abcabc'.replace('a', 'X')}'")
print(f"  count():    '{'abcabc'.count('a')}' occurrences of 'a'")
print(f"  find():     '{'hello'.find('ll')}' (index of 'll', or -1 if not found)")
print(f"  startswith(): '{'hello'.startswith('he')}'")
print(f"  endswith():   '{'hello'.endswith('lo')}'")

# === CHARACTER CHECKS ===
print("\n=== Character Classification ===")
for c in ['A', 'a', '5', '!', ' ']:
    print(f"  '{c}': upper={c.isupper()}, lower={c.islower()}, "
          f"digit={c.isdigit()}, alpha={c.isalpha()}, alnum={c.isalnum()}")

# === BUILDING A STRING CHARACTER BY CHARACTER ===
print("\n=== Building Strings ===")
# Problem: Keep only the uppercase letters, convert to lowercase.
# ALWAYS build with a list, then join. Never do string += char (it's O(n²)!)
result = []
for char in text:
    if char.isupper():
        result.append(char.lower())
final_string = "".join(result)
print(f"  Extracted upper→lower: '{final_string}'")

# === FREQUENCY COUNTING (The #1 regional trick) ===
print("\n=== Frequency Counting ===")
word = "mississippi"
freq = {}
for c in word:
    freq[c] = freq.get(c, 0) + 1  # .get(key, default) avoids KeyError
print(f"  Frequencies of '{word}': {freq}")
# Most common letter:
most_common = max(freq, key=freq.get)
print(f"  Most common: '{most_common}' ({freq[most_common]} times)")

# collections.Counter does this in one line (allowed in most contests):
from collections import Counter
freq2 = Counter(word)
print(f"  Counter: {dict(freq2)}")
print(f"  Most common 2: {freq2.most_common(2)}")

# === PREFIX SUMS (Advanced Regional Trick) ===
print("\n=== Prefix Sums ===")
# If a problem asks "sum of subarray from L to R" MULTIPLE times,
# a basic loop inside a loop is O(N*Q) and will TIME OUT.
# Prefix sums make each query O(1).
arr = [2, 4, 6, 8, 10]
prefix = [0] * (len(arr) + 1)
for i in range(len(arr)):
    prefix[i+1] = prefix[i] + arr[i]
    
print(f"  Array:      {arr}")
print(f"  Prefix Sum: {prefix}")
# Sum of arr[L..R] (inclusive) = prefix[R+1] - prefix[L]
L, R = 1, 3  # Sum of arr[1..3] = 4 + 6 + 8 = 18
subarray_sum = prefix[R+1] - prefix[L]
print(f"  Sum from index {L} to {R}: {subarray_sum}  (should be 18)")

# === 2D ARRAY TRAVERSAL PATTERNS ===
print("\n=== 2D Grid Directions ===")
# The 4 cardinal directions (up, down, left, right)
# Used in flood fill, BFS, path finding, etc.
dr = [-1, 1, 0, 0]
dc = [0, 0, -1, 1]
dirs = ['Up', 'Down', 'Left', 'Right']
r, c = 2, 3  # Current position
for i in range(4):
    nr, nc = r + dr[i], c + dc[i]
    print(f"  {dirs[i]}: ({r},{c}) → ({nr},{nc})")
````

````python
def solve():
````

```python
        print('')
```

```python
    arr = input().split()
    k = int(input())
    n = len(arr)
    if n == 0:
        print('')
```


**Two-pass pattern:** First pass counts frequencies, second pass finds the first character with frequency 1. We iterate through the *original string* in the second pass (not the dictionary) to preserve the original order.

**Why not one pass?** You can't know if a character is unique until you've seen ALL characters. So two passes is necessary.

```python
def solve():
    arr = list(map(int, input().split()))
    max_sum = current = arr[0]
    for x in arr[1:]:
        current = max(x, current + x)
        max_sum = max(max_sum, current)
    print(max_sum)
```

**Kadane's Algorithm in 3 lines.** At each element, you make a simple decision:
- `current + x`: extend the current subarray to include x
- `x`: start a fresh subarray beginning at x

You pick whichever is larger. If `current + x < x`, the accumulated sum is negative, so it's better to start fresh.

**Why this is O(n):** One pass through the array, constant work per element. The brute force (checking all subarrays) is O(n²). Kadane's is one of the most elegant algorithms ever — memorize it.
