# 🟢 Problem 1: Restaurant Bill Splitter & Formatter (Easy — I/O)

### 📖 Problem Statement
You and your friends went out for dinner. Given the **total bill amount** (float) and the **number of people** (integer) sharing the bill equally on a single line, calculate how much each person must pay. 

The final per-person amount must be displayed with **exactly 2 decimal places** and a dollar sign prefix (e.g., `$25.00`).

### 📥 Input Format
- A single line containing two space-separated values: `total_bill` (float) and `num_people` (integer).

### 📤 Output Format
Print the string: `Each person pays: $<amount>` where `<amount>` is formatted to 2 decimal places.

### 🔒 Constraints
- $0.01 \le \text{total\_bill} \le 100000.00$
- $1 \le \text{num\_people} \le 1000$

### 💡 Example 1
```text
Input:
100.00 4

Output:
Each person pays: $25.00
```
*Explanation:* $100.00 / 4 = 25.00$.

### 💡 Example 2
```text
Input:
15.50 2

Output:
Each person pays: $7.75
```
*Explanation:* $15.50 / 2 = 7.75$.


````python
# 💡 Problem 1 Solution: Restaurant Bill Splitter

def split_bill(total_bill: float, num_people: int) -> str:
    """
    Calculates the per-person share of a bill, formatted to 2 decimal places.
    """
    share = total_bill / num_people
    return f"Each person pays: ${share:.2f}"


def solve_problem_1():
    # Interactive / Standard Input Runner (reads space-separated input in one line)
    print("--- Running Problem 1 (Input/Output) ---")
    try:
        line = input("Enter total bill and number of people (e.g. 100.00 4): ")
        raw_bill, raw_people = line.split()
        
        total_bill = float(raw_bill)
        num_people = int(raw_people)
        
        result = split_bill(total_bill, num_people)
        print(result)
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_1()
````

````python
# 🧪 Problem 1 Test Suite
def test_problem_1():
    test_cases = [
        (100.00, 4, "Each person pays: $25.00"),
        (15.50, 2, "Each person pays: $7.75"),
        (120.00, 3, "Each person pays: $40.00"),
        (99.99, 3, "Each person pays: $33.33"),
    ]
    
    for total, people, expected in test_cases:
        actual = split_bill(total, people)
        assert actual == expected, f"Failed for total={total}, people={people}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: split_bill({total}, {people}) -> {actual}")

test_problem_1()
````

# 🟢 Problem 2: Digital Clock Time Converter (Easy — I/O)

### 📖 Problem Statement
Given a non-negative integer representing the total number of **seconds** elapsed, convert it into the standard 24-hour digital clock format: **`HH:MM:SS`** (Hours, Minutes, Seconds).

Hours, minutes, and seconds must always be displayed as two digits with leading zeros when necessary (e.g., 5 seconds is displayed as `05`).

### 📥 Input Format
- A single line containing an integer `total_seconds` ($0 \le \text{total\_seconds} < 360000$).

### 📤 Output Format
Print the formatted time string: `HH:MM:SS`.

### 🔒 Constraints
- $0 \le \text{total\_seconds} \le 359999$ (up to 99 hours, 59 minutes, 59 seconds)

### 💡 Example 1
```text
Input:
3665

Output:
01:01:05
```
*Explanation:* 3665 seconds is equal to 1 hour, 1 minute, and 5 seconds.

### 💡 Example 2
```text
Input:
42

Output:
00:00:42
```


````python
# 💡 Problem 2 Solution: Digital Clock Time Converter

def seconds_to_time(total_seconds: int) -> str:
    """
    Converts total seconds into formatted HH:MM:SS string using integer math.
    """
    hours = total_seconds // 3600
    remaining_seconds = total_seconds % 3600
    minutes = remaining_seconds // 60
    seconds = remaining_seconds % 60
    
    return f"{hours:02d}:{minutes:02d}:{seconds:02d}"


def solve_problem_2():
    try:
        raw_sec = input("Enter total seconds: ")
        total_seconds = int(raw_sec.strip())
        print(seconds_to_time(total_seconds))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_2()
````

````python
# 🧪 Problem 2 Test Suite
def test_problem_2():
    test_cases = [
        (3665, "01:01:05"),
        (42, "00:00:42"),
        (0, "00:00:00"),
        (3600, "01:00:00"),
        (86399, "23:59:59"),
        (359999, "99:59:59"),
    ]
    
    for sec, expected in test_cases:
        actual = seconds_to_time(sec)
        assert actual == expected, f"Failed for {sec}s. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: seconds_to_time({sec}) -> {actual}")

test_problem_2()
````

# 🟢 Problem 3: Temperature Converter (Easy — I/O)

### 📖 Problem Statement
Given a temperature reading in **Celsius** ($C$), write a program to convert it into **Fahrenheit** ($F$) using the standard formula:

$$F = C \* \(9 / 5) + 32$$

The final temperature must be displayed rounded to **2 decimal places**.

### 📥 Input Format
- A single line containing a floating point number representing temperature in Celsius (`celsius`).

### 📤 Output Format
Print the string: `Temperature: <F> F` where `<F>` is formatted to 2 decimal places.

### 🔒 Constraints
- $-273.15 \le \text{celsius} \le 10000.00$

### 💡 Example 1
```text
Input:
25.0

Output:
Temperature: 77.00 F
```
*Explanation:* $25.0 \* \(9 / 5) + 32 = 45.0 + 32 = 77.00$.

### 💡 Example 2
```text
Input:
0.0

Output:
Temperature: 32.00 F
```

### 💡 Example 3
```text
Input:
-40.0

Output:
Temperature: -40.00 F
```


````python
# 💡 Problem 3 Solution: Temperature Converter

def celsius_to_fahrenheit(celsius: float) -> str:
    """
    Converts Celsius to Fahrenheit, formatted to 2 decimal places.
    """
    fahrenheit = (celsius * 9 / 5) + 32
    return f"Temperature: {fahrenheit:.2f} F"


def solve_problem_3():
    try:
        line = input("Enter temperature in Celsius: ")
        c = float(line.strip())
        print(celsius_to_fahrenheit(c))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_3()
````

````python
# 🧪 Problem 3 Test Suite
def test_problem_3():
    test_cases = [
        (25.0, "Temperature: 77.00 F"),
        (0.0, "Temperature: 32.00 F"),
        (-40.0, "Temperature: -40.00 F"),
        (100.0, "Temperature: 212.00 F"),
        (37.0, "Temperature: 98.60 F"),
    ]
    
    for c, expected in test_cases:
        actual = celsius_to_fahrenheit(c)
        assert actual == expected, f"Failed for C={c}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: celsius_to_fahrenheit({c}) -> {actual}")

test_problem_3()
````

# 🟡 Problem 4: Academic Grade Classifier (Average — Conditionals)

### 📖 Problem Statement
Given a student's numeric exam score (from 0 to 100), classify their letter grade and status according to the following scale:

| Score Range | Grade Letter | Status |
| :--- | :--- | :--- |
| **$90 \le \text{score} \le 100$** | `A` | `Excellent` |
| **$80 \le \text{score} < 90$** | `B` | `Good` |
| **$70 \le \text{score} < 80$** | `C` | `Satisfactory` |
| **$60 \le \text{score} < 70$** | `D` | `Needs Improvement` |
| **$0 \le \text{score} < 60$** | `F` | `Fail` |

If the score is less than 0 or greater than 100, return `Invalid Score`.

### 📥 Input Format
- A single line containing a float or integer `score`.

### 📤 Output Format
Print the formatted string: `Grade: <Letter> (<Status>)` or `Invalid Score`.

### 💡 Example 1
```text
Input:
85

Output:
Grade: B (Good)
```

### 💡 Example 2
```text
Input:
55.5

Output:
Grade: F (Fail)
```


````python
# 💡 Problem 4 Solution: Academic Grade Classifier

def classify_grade(score: float) -> str:
    """
    Classifies a score into a letter grade and descriptive status using chained conditionals.
    """
    if score < 0 or score > 100:
        return "Invalid Score"
    elif score >= 90:
        return "Grade: A (Excellent)"
    elif score >= 80:
        return "Grade: B (Good)"
    elif score >= 70:
        return "Grade: C (Satisfactory)"
    elif score >= 60:
        return "Grade: D (Needs Improvement)"
    else:
        return "Grade: F (Fail)"


def solve_problem_4():
    try:
        score = float(input("Enter exam score (0-100): ").strip())
        print(classify_grade(score))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_4()
````

````python
# 🧪 Problem 4 Test Suite
def test_problem_4():
    test_cases = [
        (95, "Grade: A (Excellent)"),
        (90, "Grade: A (Excellent)"),
        (89.9, "Grade: B (Good)"),
        (80, "Grade: B (Good)"),
        (75, "Grade: C (Satisfactory)"),
        (60, "Grade: D (Needs Improvement)"),
        (59.9, "Grade: F (Fail)"),
        (0, "Grade: F (Fail)"),
        (-5, "Invalid Score"),
        (105, "Invalid Score"),
    ]
    
    for score, expected in test_cases:
        actual = classify_grade(score)
        assert actual == expected, f"Failed for score={score}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: classify_grade({score}) -> {actual}")

test_problem_4()
````

# 🟡 Problem 5: Gregorian Leap Year Checker (Average — Conditionals)

### 📖 Problem Statement
Write a program that takes an integer `year` and determines whether it is a **Leap Year** on the Gregorian calendar.

### 📜 Gregorian Leap Year Rules:
1. A year is a leap year if it is **divisible by 4**...
2. ...**EXCEPT** if it is divisible by **100**, in which case it is **NOT** a leap year...
3. ...**UNLESS** it is also divisible by **400**, in which case it **IS** a leap year!

### 📥 Input Format
- A single line containing an integer `year` ($1 \le \text{year} \le 9999$).

### 📤 Output Format
Print `Leap Year` if the year satisfies the rules, otherwise print `Not a Leap Year`.

### 💡 Example 1
```text
Input:
2024

Output:
Leap Year
```

### 💡 Example 2
```text
Input:
1900

Output:
Not a Leap Year
```

### 💡 Example 3
```text
Input:
2000

Output:
Leap Year
```


````python
# 💡 Problem 5 Solution: Leap Year Checker

def is_leap_year(year: int) -> str:
    """
    Determines if a year is a leap year using standard Gregorian rules.
    """
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return "Leap Year"
    else:
        return "Not a Leap Year"


def solve_problem_5():
    try:
        year = int(input("Enter year: ").strip())
        print(is_leap_year(year))
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_5()
````

````python
# 🧪 Problem 5 Test Suite
def test_problem_5():
    test_cases = [
        (2024, "Leap Year"),
        (2023, "Not a Leap Year"),
        (1900, "Not a Leap Year"),
        (2000, "Leap Year"),
        (2400, "Leap Year"),
        (1800, "Not a Leap Year"),
        (1600, "Leap Year"),
    ]
    
    for yr, expected in test_cases:
        actual = is_leap_year(yr)
        assert actual == expected, f"Failed for year={yr}. Expected '{expected}', got '{actual}'"
        print(f"✅ PASS: is_leap_year({yr}) -> {actual}")

test_problem_5()
````

# 🟡 Problem 6: Two Sum Without Loops — Recursive Conditionals (Average)

### 📖 Problem Statement
Given a list of integers `nums` and an integer `target`, find the **0-based indices** of the two numbers such that they add up to `target`.

⚠️ **STRICT CONSTRAINT: NO 'for' or 'while' loops are permitted!** ⚠️
You must solve this problem recursively using conditional logic.

### 📥 Input Format
- Line 1: Space-separated integers representing `nums`.
- Line 2: An integer representing `target`.

### 📤 Output Format
Returns a list `[index1, index2]` with the zero-based indices of the two numbers, or `None` if no pair exists.

### 💡 Example 1
```text
Input:
2 7 11 15
9

Output:
[0, 1]
```
*Explanation:* `nums[0] + nums[1] = 2 + 7 = 9`.

### 💡 Example 2
```text
Input:
3 2 4
6

Output:
[1, 2]
```
*Explanation:* `nums[1] + nums[2] = 2 + 4 = 6`.


````python
# 💡 Problem 6 Solution: Two Sum Without Loops (Recursive Conditionals)

def recursive_two_sum(nums: list[int], target: int, index: int = 0, seen: dict = None) -> list[int] | None:
    """
    Finds indices of two numbers that sum to target without using any loops.
    Uses recursion and dictionary lookups with conditional base cases.
    """
    if seen is None:
        seen = {}
        
    # Base Case: Reached the end of the list with no solution
    if index >= len(nums):
        return None

    current_val = nums[index]
    comp = target - current_val

    # Conditional Branch: Complement already seen?
    if comp in seen:
        return [seen[comp], index]

    # Store current value's index and recurse to the next element
    seen[current_val] = index
    return recursive_two_sum(nums, target, index + 1, seen)


def solve_problem_6():
    try:
        raw_nums = input("Enter space-separated numbers: ")
        nums = list(map(int, raw_nums.split()))
        target = int(input("Enter target sum: ").strip())
        ans = recursive_two_sum(nums, target)
        print(f"Indices: {ans}")
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_6()
````

````python
# 🧪 Problem 6 Test Suite
def test_problem_6():
    test_cases = [
        ([2, 7, 11, 15], 9, [0, 1]),
        ([3, 2, 4], 6, [1, 2]),
        ([3, 3], 6, [0, 1]),
        ([1, 5, 8, 12], 20, [2, 3]),
        ([1, 2, 3], 10, None),
    ]
    
    for nums, target, expected in test_cases:
        actual = recursive_two_sum(nums, target)
        assert actual == expected, f"Failed for nums={nums}, target={target}. Expected {expected}, got {actual}"
        print(f"✅ PASS: recursive_two_sum({nums}, target={target}) -> {actual}")

test_problem_6()
````

# 🔴 Problem 7: Conditional List Transformation Loop (Hard — Loops)

### 📖 Problem Statement
Given a list of integers `nums`, iterate through the list using a loop and transform each number according to the following rules:
- If the number is **EVEN**: multiply it by 8.
- If the number is **ODD**: add 2 to it.

Return a new list containing all the transformed values in their original order.

### 📥 Input Format
- A single line containing space-separated integers `nums`.

### 📤 Output Format
Return / print the transformed list of integers.

### 🔒 Constraints
- $1 \le \text{len}(nums) \le 10^5$
- $-10^4 \le nums[i] \le 10^4$

### 💡 Example 1
```text
Input:
1 2 3 4 5

Output:
[3, 16, 5, 32, 7]
```
*Explanation:*
- `1` is odd $\rightarrow 1 + 2 = 3$
- `2` is even $\rightarrow 2 \* 8 = 16$
- `3` is odd $\rightarrow 3 + 2 = 5$
- `4` is even $\rightarrow 4 \* 8 = 32$
- `5` is odd $\rightarrow 5 + 2 = 7$

### 💡 Example 2
```text
Input:
0 -2 3

Output:
[0, -16, 5]
```


````python
# 💡 Problem 7 Solution: Conditional List Transformation Loop

def transform_list(nums: list[int]) -> list[int]:
    """
    Iterates through nums and applies conditional arithmetic:
    - Even -> multiply by 8
    - Odd  -> add 2
    """
    result = []
    for x in nums:
        if x % 2 == 0:
            result.append(x * 8)
        else:
            result.append(x + 2)
    return result


def solve_problem_7():
    try:
        raw = input("Enter space-separated integers: ")
        nums = list(map(int, raw.split()))
        transformed = transform_list(nums)
        print(f"Transformed: {transformed}")
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_7()
````

````python
# 🧪 Problem 7 Test Suite
def test_problem_7():
    test_cases = [
        ([1, 2, 3, 4, 5], [3, 16, 5, 32, 7]),
        ([0, -2, 3], [0, -16, 5]),
        ([2, 4, 6], [16, 32, 48]),
        ([1, 3, 5], [3, 5, 7]),
        ([], []),
    ]
    
    for nums, expected in test_cases:
        actual = transform_list(nums)
        assert actual == expected, f"Failed for nums={nums}. Expected {expected}, got {actual}"
        print(f"✅ PASS: transform_list({nums}) -> {actual}")

test_problem_7()
````

# 🔴 Problem 8: Accumulator with Threshold Break Loop (Hard — Loops)

### 📖 Problem Statement
Given a list of integers `nums` and a threshold integer `threshold`, process elements in order while maintaining a `running_sum` and a count of `elements_processed`:

1. If the number is **positive** ($> 0$): add 3 times the number to `running_sum`.
2. If the number is **negative** ($< 0$): add the square of the number to `running_sum`.
3. If the number is **zero** ($== 0$): add nothing ($0$).
4. Increment `elements_processed` by 1.
5. **Early Termination:** If `running_sum` reaches or exceeds `threshold`, **stop processing immediately**.

Return `(final_sum, elements_processed)`.

### 📥 Input Format
- Line 1: Space-separated integers `nums`.
- Line 2: An integer `threshold`.

### 📤 Output Format
Return `(running_sum, elements_processed)`.

### 💡 Example 1
```text
Input:
2 -3 4 10 5
20

Output:
(27, 3)
```
*Explanation:*
- Index 0 ($x = 2$): $2 \* 3 = 6$. Sum = 6, Processed = 1.
- Index 1 ($x = -3$): $(-3)^2 = 9$. Sum = 6 + 9 = 15, Processed = 2.
- Index 2 ($x = 4$): $4 \* 3 = 12$. Sum = 15 + 12 = 27, Processed = 3.
- Sum is now $27 \ge 20$, so processing stops immediately.


````python
# 💡 Problem 8 Solution: Accumulator with Threshold Break Loop

def accumulate_until_threshold(nums: list[int], threshold: int) -> tuple[int, int]:
    """
    Iterates through nums, conditionally modifying and accumulating values
    until running_sum reaches or exceeds threshold.
    """
    running_sum = 0
    elements_processed = 0
    
    for x in nums:
        if x > 0:
            running_sum += x * 3
        elif x < 0:
            running_sum += x ** 2
        # x == 0 adds nothing
        
        elements_processed += 1
        
        # Check early termination condition
        if running_sum >= threshold:
            break
            
    return running_sum, elements_processed


def solve_problem_8():
    try:
        raw_nums = input("Enter space-separated numbers: ")
        nums = list(map(int, raw_nums.split()))
        threshold = int(input("Enter threshold: ").strip())
        total, count = accumulate_until_threshold(nums, threshold)
        print(f"Final Sum: {total}, Elements Processed: {count}")
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_8()
````

````python
# 🧪 Problem 8 Test Suite
def test_problem_8():
    test_cases = [
        ([2, -3, 4, 10, 5], 20, (27, 3)),
        ([1, 2, 3], 100, (18, 3)),
        ([-5, 10], 25, (25, 1)),
        ([0, 0, 2], 5, (6, 3)),
    ]
    
    for nums, thresh, expected in test_cases:
        actual = accumulate_until_threshold(nums, thresh)
        assert actual == expected, f"Failed for nums={nums}, threshold={thresh}. Expected {expected}, got {actual}"
        print(f"✅ PASS: accumulate_until_threshold({nums}, {thresh}) -> {actual}")

test_problem_8()
````

# 🔴 Problem 9: Find All Duplicates via Frequency Map (Hard — Loops & Dicts)

### 📖 Problem Statement
Given a list of integers `nums`, iterate through the numbers using a loop to count occurrences, and identify all values that appear **more than once** (duplicates) in the list.

Return a list containing all duplicate values.

### 📥 Input Format
- A single line containing space-separated integers `nums`.

### 📤 Output Format
A list of all numbers that appear more than once.

### 💡 Example 1
```text
Input:
4 3 2 7 8 2 3 1

Output:
[3, 2]
```
*Explanation:* `3` appears twice and `2` appears twice. `4, 7, 8, 1` appear only once.

### 💡 Example 2
```text
Input:
1 1 2

Output:
[1]
```

### 💡 Example 3
```text
Input:
1 2 3 4 5

Output:
[]
```


````python
# 💡 Problem 9 Solution: Find All Duplicates via Frequency Loop

def find_duplicates(nums: list[int]) -> list[int]:
    """
    Counts frequencies using a for loop and seen.get(),
    then returns all elements appearing more than once.
    """
    seen = {}
    for i in nums:
        seen[i] = seen.get(i, 0) + 1

    new_seen = [i for i, x in seen.items() if x > 1]
    return new_seen


def solve_problem_9():
    try:
        raw = input("Enter space-separated integers: ")
        nums = list(map(int, raw.split()))
        duplicates = find_duplicates(nums)
        print(f"Duplicates: {duplicates}")
    except (ValueError, EOFError) as e:
        print(f"Invalid input: {e}")

# Uncomment below to test with manual input:
# solve_problem_9()
````

````python
# 🧪 Problem 9 Test Suite
def test_problem_9():
    test_cases = [
        ([4, 3, 2, 7, 8, 2, 3, 1], [3, 2]),
        ([1, 1, 2], [1]),
        ([1, 2, 3, 4, 5], []),
        ([5, 5, 5, 5], [5]),
        ([10, 20, 10, 30, 20, 40], [10, 20]),
    ]
    
    for nums, expected in test_cases:
        actual = find_duplicates(nums)
        assert actual == expected, f"Failed for nums={nums}. Expected {expected}, got {actual}"
        print(f"✅ PASS: find_duplicates({nums}) -> {actual}")

test_problem_9()
````
