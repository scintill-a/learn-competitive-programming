# 🔍 Binary Search — Cut the Problem in Half

Binary search is the **single most powerful technique** when the answer space is ordered.
It turns O(n) searches into O(log n) — a 1,000,000-element search becomes 20 steps.

**Recognition Triggers:**
- "Find position in sorted array" → Binary Search
- "Minimum/maximum that satisfies condition" → Binary Search on Answer
- "Sorted array" → probably Binary Search
- "Find the boundary" → Binary Search

---
## 🧠 The Binary Search Template — Never Get Confused Again

There are TWO variants. Know which one you need:

### Variant 1: Find exact target (inclusive bounds)

```python
def binary_search(arr, target):
    """Find exact target in sorted array. Returns index or -1."""
    lo, hi = 0, len(arr) - 1
    while lo <= hi:           # Note: <= not <
        mid = (lo + hi) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

arr = [1, 3, 5, 7, 9, 11, 13]
print("Binary Search:")
print(f"  Find 7 in {arr}: index {binary_search(arr, 7)}")   # 3
print(f"  Find 4 in {arr}: index {binary_search(arr, 4)}")   # -1
```

### Variant 2: Find boundary (bisect left / bisect right)

```python
def bisect_left(arr, target):
    """Find leftmost position to insert target (first position >= target)."""
    lo, hi = 0, len(arr)
    while lo < hi:            # Note: < not <=
        mid = (lo + hi) // 2
        if arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid          # Note: mid, not mid-1
    return lo

def bisect_right(arr, target):
    """Find rightmost position to insert target (first position > target)."""
    lo, hi = 0, len(arr)
    while lo < hi:
        mid = (lo + hi) // 2
        if arr[mid] <= target:  # Note: <= not <
            lo = mid + 1
        else:
            hi = mid
    return lo

arr = [1, 3, 3, 3, 5, 7]
print(f"bisect_left({arr}, 3) = {bisect_left(arr, 3)}")    # 1 (first 3)
print(f"bisect_right({arr}, 3) = {bisect_right(arr, 3)}")  # 4 (after last 3)
print(f"Count of 3s = {bisect_right(arr, 3) - bisect_left(arr, 3)}")  # 3
```

```python
# Python's bisect module — use this in contests!
from bisect import bisect_left, bisect_right, insort

arr = [1, 3, 3, 3, 5, 7]
print("Python bisect module:")
print(f"  bisect_left({arr}, 3) = {bisect_left(arr, 3)}")    # 1
print(f"  bisect_right({arr}, 3) = {bisect_right(arr, 3)}")  # 4

# insort — insert while keeping sorted
sorted_list = [1, 3, 5, 7]
insort(sorted_list, 4)
print(f"  insort 4 into [1,3,5,7] → {sorted_list}")  # [1, 3, 4, 5, 7]
```

## 🔥 Search on Answer (Binary Search on the Solution Space)

This is the **most powerful** binary search pattern. Instead of searching IN an array,
you search the ANSWER SPACE.

### The Monotonic Predicate Mental Model

If you can write a function `can_do(x)` that returns:
- `False, False, ..., False, True, True, ..., True`

as x increases, then you can binary search for the transition point!

```python
def search_on_answer(lo, hi, predicate):
    """Find smallest x in [lo, hi] where predicate(x) is True.
    Predicate must be monotonic: F,F,...,F,T,T,...,T
    """
    while lo < hi:
        mid = (lo + hi) // 2
        if predicate(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

# Example: Find smallest number whose square >= 100
result = search_on_answer(1, 100, lambda x: x * x >= 100)
print(f"Smallest x where x² >= 100: {result}")  # 10
```

## 🔥 Problem: Binary Search (LeetCode 704)

```python
def search(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

print("Binary Search:")
print("  [-1,0,3,5,9,12], target=9:", search([-1,0,3,5,9,12], 9))  # 4
print("  [-1,0,3,5,9,12], target=2:", search([-1,0,3,5,9,12], 2))  # -1
```

## 🔥 Problem: Search a 2D Matrix

```python
def search_matrix(matrix, target):
    """Treat 2D matrix as sorted 1D array. O(log(m*n))."""
    if not matrix:
        return False
    m, n = len(matrix), len(matrix[0])
    lo, hi = 0, m * n - 1
    
    while lo <= hi:
        mid = (lo + hi) // 2
        val = matrix[mid // n][mid % n]  # Convert 1D index to 2D
        if val == target:
            return True
        elif val < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return False

matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
print("Search 2D Matrix:")
print(f"  Find 3: {search_matrix(matrix, 3)}")   # True
print(f"  Find 13: {search_matrix(matrix, 13)}")  # False
```

## 🔥 Problem: Koko Eating Bananas (Search on Answer!)

```python
import math

def min_eating_speed(piles, h):
    """Binary search on the answer: what's the minimum eating speed?
    
    Predicate: can_finish(speed) — can Koko eat all bananas at this speed in h hours?
    This is monotonic: if speed k works, any speed > k also works.
    
    Time: O(n * log(max(piles)))
    """
    def can_finish(speed):
        hours = sum(math.ceil(pile / speed) for pile in piles)
        return hours <= h
    
    lo, hi = 1, max(piles)
    while lo < hi:
        mid = (lo + hi) // 2
        if can_finish(mid):
            hi = mid
        else:
            lo = mid + 1
    return lo

print("Koko Eating Bananas:")
print("  piles=[3,6,7,11], h=8:", min_eating_speed([3,6,7,11], 8))     # 4
print("  piles=[30,11,23,4,20], h=5:", min_eating_speed([30,11,23,4,20], 5))  # 30
print("  piles=[30,11,23,4,20], h=6:", min_eating_speed([30,11,23,4,20], 6))  # 23
```

## 🔥 Problem: Find Minimum in Rotated Sorted Array

```python
def find_min(nums):
    """Binary search: compare mid with right to determine which half to search.
    If nums[mid] > nums[right], min is in right half. Otherwise, left half.
    Time: O(log n)"""
    lo, hi = 0, len(nums) - 1
    while lo < hi:
        mid = (lo + hi) // 2
        if nums[mid] > nums[hi]:
            lo = mid + 1  # Min is in right half
        else:
            hi = mid      # Min is in left half (including mid)
    return nums[lo]

print("Find Min in Rotated Sorted Array:")
print("  [3,4,5,1,2]:", find_min([3,4,5,1,2]))     # 1
print("  [4,5,6,7,0,1,2]:", find_min([4,5,6,7,0,1,2]))  # 0
print("  [11,13,15,17]:", find_min([11,13,15,17]))  # 11 (not rotated)
```

## 🔥 Problem: Search in Rotated Sorted Array

```python
def search_rotated(nums, target):
    """Binary search with extra logic to determine which half is sorted.
    Time: O(log n)"""
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        
        # Left half is sorted
        if nums[lo] <= nums[mid]:
            if nums[lo] <= target < nums[mid]:
                hi = mid - 1
            else:
                lo = mid + 1
        # Right half is sorted
        else:
            if nums[mid] < target <= nums[hi]:
                lo = mid + 1
            else:
                hi = mid - 1
    return -1

print("Search in Rotated Sorted Array:")
print("  [4,5,6,7,0,1,2], target=0:", search_rotated([4,5,6,7,0,1,2], 0))  # 4
print("  [4,5,6,7,0,1,2], target=3:", search_rotated([4,5,6,7,0,1,2], 3))  # -1
```

## 🔥 Problem: Median of Two Sorted Arrays

```python
def find_median_sorted_arrays(nums1, nums2):
    """Binary search on partition position. O(log(min(m,n))).
    
    Key idea: partition both arrays such that:
    - Left half has exactly (m+n+1)//2 elements
    - All elements in left half <= all elements in right half
    """
    # Ensure nums1 is shorter
    if len(nums1) > len(nums2):
        nums1, nums2 = nums2, nums1
    
    m, n = len(nums1), len(nums2)
    lo, hi = 0, m
    
    while lo <= hi:
        i = (lo + hi) // 2       # Partition point in nums1
        j = (m + n + 1) // 2 - i  # Partition point in nums2
        
        left1 = nums1[i-1] if i > 0 else float('-inf')
        right1 = nums1[i] if i < m else float('inf')
        left2 = nums2[j-1] if j > 0 else float('-inf')
        right2 = nums2[j] if j < n else float('inf')
        
        if left1 <= right2 and left2 <= right1:
            # Found valid partition
            if (m + n) % 2 == 1:
                return max(left1, left2)
            else:
                return (max(left1, left2) + min(right1, right2)) / 2
        elif left1 > right2:
            hi = i - 1
        else:
            lo = i + 1
    
    return 0.0

print("Median of Two Sorted Arrays:")
print("  [1,3] & [2]:", find_median_sorted_arrays([1,3], [2]))         # 2.0
print("  [1,2] & [3,4]:", find_median_sorted_arrays([1,2], [3,4]))     # 2.5
```

## ⚡ Contest Trick: Always Check if Answer Space is Sortable

If you see a problem asking for "minimum X such that condition" or "maximum Y such that condition",
check if the condition is **monotonic** in X or Y. If so → binary search on answer!

### Examples of Search on Answer:
| Problem | Answer Space | Predicate |
|---------|-------------|----------|
| Koko Eating Bananas | speed ∈ [1, max(piles)] | can_finish(speed) |
| Capacity to Ship | capacity ∈ [max(weight), sum(weights)] | can_ship(capacity, days) |
| Split Array Largest Sum | max_sum ∈ [max(arr), sum(arr)] | can_split(max_sum, m) |
| Aggressive Cows | min_distance ∈ [1, max_pos - min_pos] | can_place(min_dist, k) |

## 🏆 Summary

| Problem | Technique | Time |
|---------|-----------|------|
| Basic Binary Search | Standard template | O(log n) |
| Search a 2D Matrix | Flatten to 1D, binary search | O(log(m*n)) |
| Koko Eating Bananas | Search on answer | O(n log M) |
| Find Min in Rotated | Compare mid with right | O(log n) |
| Search in Rotated | Check which half is sorted | O(log n) |
| Median of Two Sorted | Partition binary search | O(log(min(m,n))) |
