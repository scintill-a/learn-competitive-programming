Math and bit tricks can turn complex problems into one-liners.
These are especially useful in speed contests where every second counts.

## 🧮 Number Theory Basics

```python
import math

# GCD and LCM
print("GCD(12, 8) =", math.gcd(12, 8))  # 4
print("LCM(12, 8) =", math.lcm(12, 8))  # 24 (Python 3.9+)
# Manual LCM: lcm(a, b) = a * b // gcd(a, b)

# Primality check
def is_prime(n):
    if n < 2: return False
    if n < 4: return True
    if n % 2 == 0 or n % 3 == 0: return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0: return False
        i += 6
    return True

print("Is 17 prime?", is_prime(17))  # True
print("Is 15 prime?", is_prime(15))  # False

# Sieve of Eratosthenes — all primes up to n
def sieve(n):
    """O(n log log n) time, O(n) space."""
    is_p = [True] * (n + 1)
    is_p[0] = is_p[1] = False
    for i in range(2, int(n**0.5) + 1):
        if is_p[i]:
            for j in range(i*i, n + 1, i):
                is_p[j] = False
    return [i for i in range(n + 1) if is_p[i]]

print("Primes up to 30:", sieve(30))
```

```python
# Modular Arithmetic (essential for large number problems)
MOD = 10**9 + 7  # Common mod in contests

# Fast modular exponentiation
def mod_pow(base, exp, mod):
    """O(log exp). Python: pow(base, exp, mod) does this built-in!"""
    return pow(base, exp, mod)

print("2^100 mod (10^9+7) =", pow(2, 100, MOD))

# Modular inverse: a^(-1) mod p = a^(p-2) mod p (when p is prime)
def mod_inverse(a, mod):
    return pow(a, mod - 2, mod)

print("Inverse of 3 mod 7:", mod_inverse(3, 7))  # 5 (because 3*5 = 15 ≡ 1 mod 7)
```

## ⚡ Bit Manipulation

```python
# Bit Operations Cheat Sheet
a, b = 0b1010, 0b1100  # 10 and 12

print(f"AND: {a} & {b} = {a & b}  ({bin(a & b)})")
print(f"OR:  {a} | {b} = {a | b}  ({bin(a | b)})")
print(f"XOR: {a} ^ {b} = {a ^ b}  ({bin(a ^ b)})")
print(f"NOT: ~{a} = {~a}")
print(f"Left shift:  {a} << 1 = {a << 1}")
print(f"Right shift: {a} >> 1 = {a >> 1}")

# Useful bit tricks
n = 12  # 0b1100
print(f"\nBit Tricks for n={n} ({bin(n)}):")
print(f"  Check if even: {n & 1 == 0}")          # True
print(f"  Check if power of 2: {n & (n-1) == 0}")  # False (12 is not)
print(f"  Get lowest set bit: {n & (-n)}")         # 4
print(f"  Clear lowest set bit: {n & (n-1)}")      # 8
print(f"  Count set bits: {bin(n).count('1')}")    # 2
```

## 🔥 Bit Problems

```python
# Single Number — find the element that appears once (all others appear twice)
def single_number(nums):
    """XOR trick: a ^ a = 0, a ^ 0 = a. Time: O(n), Space: O(1)"""
    result = 0
    for num in nums:
        result ^= num
    return result

print("Single Number:", single_number([4,1,2,1,2]))  # 4
```

```python
# Number of 1 Bits (Hamming Weight)
def hamming_weight(n):
    count = 0
    while n:
        count += 1
        n &= (n - 1)  # Clear lowest set bit
    return count

# Python one-liner: bin(n).count('1')
print("Hamming Weight of 11:", hamming_weight(11))  # 3 (1011)
```

```python
# Counting Bits — return array where result[i] = number of 1s in binary of i
def count_bits(n):
    """DP trick: bits[i] = bits[i >> 1] + (i & 1). Time: O(n)"""
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        dp[i] = dp[i >> 1] + (i & 1)
    return dp

print("Counting Bits 0-5:", count_bits(5))  # [0, 1, 1, 2, 1, 2]
```

```python
# Missing Number — find missing number in [0, n]
def missing_number(nums):
    """XOR all indices and values. Pairs cancel out, missing remains."""
    result = len(nums)
    for i, num in enumerate(nums):
        result ^= i ^ num
    return result

# Alternative: math formula
def missing_number_math(nums):
    n = len(nums)
    return n * (n + 1) // 2 - sum(nums)

print("Missing Number in [3,0,1]:", missing_number([3,0,1]))  # 2
```

```python
# Sum of Two Integers (without + operator)
def get_sum(a, b):
    """Use XOR for addition without carry, AND+shift for carry. 
    Python handles negative numbers differently, so we need a mask."""
    MASK = 0xFFFFFFFF  # 32-bit mask
    MAX = 0x7FFFFFFF   # Max positive 32-bit int
    while b & MASK:
        carry = (a & b) << 1
        a = a ^ b
        b = carry
    return (a & MASK) if a > MAX else a

print("Sum without +: 3 + 5 =", get_sum(3, 5))  # 8
```

```python
# Reverse Bits
def reverse_bits(n):
    result = 0
    for _ in range(32):
        result = (result << 1) | (n & 1)
        n >>= 1
    return result

print(f"Reverse bits of 43261596: {reverse_bits(43261596)}")
```

```python
# Power of Two
def is_power_of_two(n):
    return n > 0 and (n & (n - 1)) == 0

for n in [1, 2, 3, 4, 16, 18]:
    print(f"  {n}: {'✅' if is_power_of_two(n) else '❌'}")
```

## 🔥 Math Problems

```python
# Palindrome Number
def is_palindrome_num(x):
    if x < 0: return False
    return str(x) == str(x)[::-1]

# Without string conversion
def is_palindrome_num_v2(x):
    if x < 0 or (x % 10 == 0 and x != 0): return False
    rev = 0
    while x > rev:
        rev = rev * 10 + x % 10
        x //= 10
    return x == rev or x == rev // 10

print("Palindrome 121:", is_palindrome_num(121))   # True
print("Palindrome -121:", is_palindrome_num(-121)) # False
```

```python
# Pow(x, n) — O(log n)
def my_pow(x, n):
    if n < 0:
        x = 1 / x
        n = -n
    result = 1
    while n > 0:
        if n & 1:
            result *= x
        x *= x
        n >>= 1
    return result

print("2^10 =", my_pow(2, 10))      # 1024
print("2^-2 =", my_pow(2, -2))      # 0.25
```

```python
# Sqrt(x) — Binary Search
def my_sqrt(x):
    if x < 2: return x
    lo, hi = 1, x // 2
    while lo <= hi:
        mid = (lo + hi) // 2
        if mid * mid == x: return mid
        elif mid * mid < x: lo = mid + 1
        else: hi = mid - 1
    return hi

print("sqrt(8) =", my_sqrt(8))    # 2
print("sqrt(16) =", my_sqrt(16))  # 4
```

```python
# Plus One
def plus_one(digits):
    for i in range(len(digits) - 1, -1, -1):
        if digits[i] < 9:
            digits[i] += 1
            return digits
        digits[i] = 0
    return [1] + digits

print("Plus One [1,2,3]:", plus_one([1,2,3]))  # [1,2,4]
print("Plus One [9,9,9]:", plus_one([9,9,9]))  # [1,0,0,0]
```

## 🏆 Summary

### Bit Tricks Quick Reference
| Trick | Code | Use Case |
|-------|------|----------|
| Check even | `n & 1 == 0` | Fast parity check |
| Power of 2 | `n & (n-1) == 0` | Validation |
| Get lowest bit | `n & (-n)` | Isolate rightmost 1 |
| Clear lowest bit | `n & (n-1)` | Count bits |
| XOR cancel | `a ^ a = 0` | Find unique element |
| Multiply by 2 | `n << 1` | Fast multiplication |
| Divide by 2 | `n >> 1` | Fast division |
