## 1. Variables & Math Operations
Python is dynamically typed. You don't need to declare variable types.

```python
x = 10
y = 3

# True division (returns float)
print("10 / 3 =", x / y)   # 3.3333333333333335

# Integer division (Floor division) - CRITICAL FOR CONTESTS
print("10 // 3 =", x // y) # 3
print("-10 // 3 =", -x // y) # -4 (Notice it rounds DOWN, not towards zero!)

# Modulo (Remainder)
print("10 % 3 =", x % y)   # 1

# Exponentiation
print("10 ** 3 =", x ** y) # 1000

# Multiple assignment (Very useful for swapping)
a, b = 5, 10
a, b = b, a  # Swaps values instantly without a temp variable!
print("Swapped a and b:", a, b)

# Infinity
max_val = float('inf')
min_val = float('-inf')
print("Is 1 million less than infinity?", 10**6 < max_val)
```

## 2. Lists & Slicing
Python lists are dynamic arrays. They are your go-to data structure.

```python
nums = [10, 20, 30, 40, 50]

# Accessing from the end
print("Last element:", nums[-1])  # 50
print("Second to last:", nums[-2]) # 40

# Adding / Removing
nums.append(60)       # O(1) add to end
last = nums.pop()     # O(1) remove from end
# nums.pop(0)         # O(n) remove from front. BAD! Use `collections.deque` instead.

# --- SLICING --- [start:stop:step]
# Stop index is EXCLUSIVE.
print("First three elements:", nums[:3])    # [10, 20, 30]
print("From index 2 to end:", nums[2:])     # [30, 40, 50]
print("Everything reversed:", nums[::-1])   # [50, 40, 30, 20, 10]
print("Every 2nd element:", nums[::2])      # [10, 30, 50]

# Creating 1D arrays
zeros = [0] * 5
print("Initialized array:", zeros)  # [0, 0, 0, 0, 0]

# ⚠️ DANGER: Creating 2D arrays
bad_2d = [[0] * 3] * 3
bad_2d[0][0] = 1
print("Bad 2D array (all rows change!):", bad_2d)  # [[1, 0, 0], [1, 0, 0], [1, 0, 0]]

# CORRECT way to create 2D arrays (List Comprehension)
good_2d = [[0] * 3 for _ in range(3)]
good_2d[0][0] = 1
print("Good 2D array:", good_2d)                   # [[1, 0, 0], [0, 0, 0], [0, 0, 0]]
```

## 3. Strings & Characters
Strings are immutable (cannot be changed in place).

```python
s = "competitive"

# Substrings (Slicing works exactly like lists)
print("Reverse string:", s[::-1])

# Converting between char and ASCII value
print("ASCII of 'a':", ord('a'))       # 97
print("Char from 98:", chr(98))        # 'b'

# Getting alphabet index (0-25) - Super common in Trie and Anagram problems
print("Index of 'c':", ord('c') - ord('a')) # 2

# Modifying a string? Convert to list first!
char_list = list(s)
char_list[0] = 'C'
modified_s = "".join(char_list)
print("Modified string:", modified_s)  # "Competitive"

# Helpful string methods
print("123".isdigit())   # True
print("abc".isalpha())   # True
print("a,b,c".split(',')) # ['a', 'b', 'c']
```

## 4. Loops, Enumerate, & Zip
Never use `range(len(nums))` if you actually want the items.

```python
fruits = ["apple", "banana", "cherry"]

# Standard iteration (Clean)
for fruit in fruits:
    pass

# Enumerate: Gives you both index AND value
print("Enumerate:")
for i, fruit in enumerate(fruits):
    print(f"  Index {i}: {fruit}")

# Zip: Iterate over two or more lists simultaneously
prices = [1.2, 0.5, 3.0]
print("\nZip:")
for fruit, price in zip(fruits, prices):
    print(f"  {fruit} costs ${price}")

# Reverse iteration
for i in range(5, 0, -1): # start=5, stop=0 (exclusive), step=-1
    pass # 5, 4, 3, 2, 1

# Reversing a list iterator
for fruit in reversed(fruits):
    pass # cherry, banana, apple
```

## 5. Dictionaries (Hash Maps) & Sets
The most important data structures for O(1) lookups.

```python
# SETS: Unordered, unique elements, O(1) insert/lookup
seen = set()
seen.add(1)
seen.add(2)
seen.add(1) # Ignored
print("Set:", seen)             # {1, 2}
print("Is 2 in set?", 2 in seen) # True, O(1) time!

a = {1, 2, 3}
b = {3, 4, 5}
print("Union:", a | b)          # {1, 2, 3, 4, 5}
print("Intersection:", a & b)   # {3}

# DICTIONARIES: Key-value pairs, O(1) insert/lookup
d = {"a": 1, "b": 2}
d["c"] = 3

# Safe lookup (avoids KeyError)
print("Value of 'z':", d.get("z", 0)) # Returns 0 if 'z' not found

# Iterating
for key, val in d.items():
    pass # loops over keys and values
```

## 6. The "collections" Module (Your Best Friend)
Built-in dicts and lists are good, but `collections` is god-tier.

```python
from collections import Counter, defaultdict, deque

# 1. Counter: Automatically counts frequencies of elements
count = Counter("mississippi")
print("Counter:", count)                 # Counter({'i': 4, 's': 4, 'p': 2, 'm': 1})
print("Most common:", count.most_common(2)) # [('i', 4), ('s', 4)]

# 2. defaultdict: Never throw a KeyError again
graph = defaultdict(list)
graph[0].append(1)  # Automatically created the empty list for key 0!
print("defaultdict graph:", dict(graph)) # {0: [1]}

freq = defaultdict(int)
freq['apples'] += 5 # Auto-initializes to 0, then adds 5
print("defaultdict freq:", dict(freq))   # {'apples': 5}

# 3. deque (Double Ended Queue): O(1) push/pop from BOTH ends
q = deque([1, 2, 3])
q.append(4)      # Add to right O(1)
q.appendleft(0)  # Add to left O(1)
q.popleft()      # Remove from left O(1)
print("Deque:", q)                       # deque([1, 2, 3, 4])
```

## 7. Comprehensions (Writing one-liners)
Comprehensions are faster and cleaner than standard `for` loops.

```python
nums = [1, 2, 3, 4, 5]

# List Comprehension: [expression for item in iterable if condition]
squares = [x * x for x in nums]
print("Squares:", squares)               # [1, 4, 9, 16, 25]

evens = [x for x in nums if x % 2 == 0]
print("Evens:", evens)                   # [2, 4]

# Set Comprehension
unique_squares = {x * x for x in [-2, -1, 0, 1, 2]}
print("Set comp:", unique_squares)       # {0, 1, 4}

# Dictionary Comprehension
val_to_index = {val: i for i, val in enumerate(nums)}
print("Dict comp:", val_to_index)        # {1: 0, 2: 1, 3: 2, 4: 3, 5: 4}
```

## 8. Custom Sorting & Lambdas
Sorting complex structures is a huge part of contests.

```python
intervals = [[1, 3], [15, 18], [8, 10], [2, 6]]

# Sort by default (sorts by 1st element, then 2nd element)
intervals.sort()
print("Default sort:", intervals)      # [[1, 3], [2, 6], [8, 10], [15, 18]]

# Sort by the SECOND element using a lambda (anonymous function)
# lambda arguments: expression
intervals.sort(key=lambda x: x[1])
print("Sort by 2nd element:", intervals) # [[1, 3], [2, 6], [8, 10], [15, 18]]

students = [("Alice", 90), ("Bob", 85), ("Charlie", 90)]
# Sort by grade DESCENDING, then name ASCENDING
# We negate the grade (-x[1]) to reverse the integer sort order!
students.sort(key=lambda x: (-x[1], x[0]))
print("Multi-condition sort:", students)
```

## 9. Advanced Built-in Functions

```python
# any() and all()
bools = [True, False, True]
print("any():", any(bools))  # True (at least one is True)
print("all():", all(bools))  # False (not all are True)

# Checking properties quickly
nums = [2, 4, 6, 8]
print("Are all even?", all(x % 2 == 0 for x in nums))

# map(): Apply a function to all items
string_nums = ["1", "2", "3"]
integers = list(map(int, string_nums))
print("Map to int:", integers)

# divmod(): Returns (quotient, remainder)
q, r = divmod(10, 3)
print("divmod(10, 3):", q, r) # 3, 1
```

## 10. `itertools` (The Magic Module)
Provides functions for efficient looping and combinatorics.

```python
import itertools

# Permutations (All orderings)
print("Permutations:", list(itertools.permutations([1, 2], 2))) # [(1, 2), (2, 1)]

# Combinations (Order doesn't matter)
print("Combinations:", list(itertools.combinations([1, 2, 3], 2))) # [(1, 2), (1, 3), (2, 3)]

# Product (Cartesian product / Nested loops)
print("Product:", list(itertools.product([1, 2], ['a', 'b']))) 
# [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b')]
```

## 12. Crucial Python "Gotchas" (Avoid these traps!)

```python
## Trap 1: Mutable Default Arguments
def bad_function(item, my_list=[]):  # The list is created ONCE at definition time!
    my_list.append(item)
    return my_list

print("Bad func 1:", bad_function(1)) # [1]
print("Bad func 2:", bad_function(2)) # [1, 2] !!! Wait, it kept the 1!

def good_function(item, my_list=None):
    if my_list is None:
        my_list = []
    my_list.append(item)
    return my_list

## Trap 2: Copying Lists
a = [1, 2, 3]
b = a         # b points to the SAME memory as a
b[0] = 99
print("\nList a changed too!", a) # [99, 2, 3]

# How to actually copy:
c = [1, 2, 3]
d = c[:]      # Slice creates a shallow copy
d[0] = 99
print("Safe copy:", c, d) # [1, 2, 3] [99, 2, 3]

## Trap 3: `is` vs `==`
# `==` checks for VALUE equality. `is` checks for MEMORY equality (same object).
x = [1, 2]
y = [1, 2]
print("\nx == y?", x == y) # True
print("x is y?", x is y) # False
```
