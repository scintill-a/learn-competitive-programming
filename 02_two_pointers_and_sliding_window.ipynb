## Part 1: Two Pointer Technique

### When to Use Two Pointers

| Scenario | Pattern |
|----------|--------|
| Sorted array + find pair with sum | Opposite pointers (left, right) |
| Check if palindrome | Opposite pointers (left, right) |
| Remove duplicates in-place | Slow/fast pointers |
| Linked list cycle | Fast/slow pointers |
| Merge two sorted arrays | Two read pointers |
| Partition (Dutch flag) | Three pointers |

````python
# Problem: Valid Palindrome
# Check if a string is a palindrome (ignoring non-alphanumeric, case-insensitive).

def is_palindrome(s):
    """Two pointers from opposite ends. O(n) time, O(1) space."""
    left, right = 0, len(s) - 1
    while left < right:
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

# Pythonic one-liner (but O(n) space)
def is_palindrome_oneliner(s):
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]

print("Valid Palindrome:")
print("  'A man, a plan, a canal: Panama':", is_palindrome("A man, a plan, a canal: Panama"))  # True
print("  'race a car':", is_palindrome("race a car"))  # False
````

````python
# Problem: Two Sum II — Input Array Is Sorted
# Given a sorted array, find two numbers that add up to target.

def two_sum_sorted(numbers, target):
    """Opposite-direction two pointers. O(n) time, O(1) space."""
    left, right = 0, len(numbers) - 1
    while left < right:
        current_sum = numbers[left] + numbers[right]
        if current_sum == target:
            return [left + 1, right + 1]  # 1-indexed
        elif current_sum < target:
            left += 1    # Need bigger sum → move left right
        else:
            right -= 1   # Need smaller sum → move right left
    return []

print("Two Sum II (Sorted):")
print("  [2,7,11,15], 9:", two_sum_sorted([2,7,11,15], 9))  # [1, 2]
print("  [2,3,4], 6:", two_sum_sorted([2,3,4], 6))            # [1, 3]
````

````python
# Problem: 3Sum
# Find all unique triplets that sum to zero.

def three_sum(nums):
    """Sort + fix one + two pointers. O(n²) time, O(1) extra space."""
    nums.sort()
    result = []
    for i in range(len(nums) - 2):
        if i > 0 and nums[i] == nums[i-1]:
            continue
        if nums[i] > 0:
            break
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                while left < right and nums[left] == nums[left+1]: left += 1
                while left < right and nums[right] == nums[right-1]: right -= 1
                left += 1; right -= 1
            elif total < 0:
                left += 1
            else:
                right -= 1
    return result

print("3Sum:")
print(" ", three_sum([-1,0,1,2,-1,-4]))  # [[-1,-1,2], [-1,0,1]]
````

````python
# Problem: Container With Most Water
# Given heights, find two lines that form a container holding the most water.

def max_area(height):
    """Two pointers. Always move the shorter line inward.
    Why? Moving the taller line can only decrease area (width shrinks, height capped by shorter).
    
    Time: O(n), Space: O(1)
    """
    left, right = 0, len(height) - 1
    best = 0
    while left < right:
        width = right - left
        h = min(height[left], height[right])
        best = max(best, width * h)
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return best

print("Container With Most Water:")
print("  [1,8,6,2,5,4,8,3,7]:", max_area([1,8,6,2,5,4,8,3,7]))  # 49
````

````python
# Problem: Trapping Rain Water
# Given elevation map, compute how much water it can trap.

def trap(height):
    """Two pointer approach. Water at position i = min(left_max, right_max) - height[i].
    
    Time: O(n), Space: O(1)
    """
    if not height:
        return 0
    left, right = 0, len(height) - 1
    left_max, right_max = height[left], height[right]
    water = 0
    while left < right:
        if left_max < right_max:
            left += 1
            left_max = max(left_max, height[left])
            water += left_max - height[left]
        else:
            right -= 1
            right_max = max(right_max, height[right])
            water += right_max - height[right]
    return water

print("Trapping Rain Water:")
print("  [0,1,0,2,1,0,1,3,2,1,2,1]:", trap([0,1,0,2,1,0,1,3,2,1,2,1]))  # 6
print("  [4,2,0,3,2,5]:", trap([4,2,0,3,2,5]))  # 9
````

## Part 2: Sliding Window

### Fixed Window vs Variable Window — Know the Difference!

| Type | Window Size | When to Use | Shrink Condition |
|------|------------|-------------|------------------|
| **Fixed** | Always K | "Sum of K elements", "anagram check" | After window reaches size K |
| **Variable** | Changes | "Longest/shortest with constraint" | When constraint is violated |

### The Variable Sliding Window Template

```python
def variable_sliding_window(arr):
    left = 0
    state = {}  # or set(), or sum, etc.
    best = 0
    
    for right in range(len(arr)):
        # 1. EXPAND: Add arr[right] to state
        update_state(arr[right])
        
        # 2. SHRINK: While state violates constraint
        while constraint_violated(state):
            remove_from_state(arr[left])
            left += 1
        
        # 3. UPDATE: Current window [left..right] is valid
        best = max(best, right - left + 1)
    
    return best
```

## 🔥🔥🔥 THE ANCHOR PROBLEM: `length_of_longest_substring(s)`

### Longest Substring Without Repeating Characters

> Given a string `s`, find the length of the longest substring without repeating characters.

**This is the #1 sliding window problem. Master this and all others become easy.**

````python
# ═══════════════════════════════════════════════════════════════
# VERSION 1: Set-based — The Clean Contest Solution
# ═══════════════════════════════════════════════════════════════

def length_of_longest_substring(s):
    """Set-based sliding window.
    
    Algorithm:
    - Expand right pointer each iteration
    - If s[right] already in window set → shrink from left
    - Update max length after each valid expansion
    
    Time:  O(n) — each char added/removed at most once
    Space: O(min(n, alphabet_size))
    """
    window = set()
    left = 0
    length = 0
    for right in range(len(s)):
        while s[right] in window:
            window.remove(s[left])
            left += 1
        window.add(s[right])
        length = max(length, right - left + 1)
    return length

# Test all cases
test_cases = [
    ("abcabcbb", 3),  # "abc"
    ("bbbbb", 1),     # "b"
    ("pwwkew", 3),    # "wke"
    ("", 0),
    (" ", 1),
    ("dvdf", 3),      # "vdf"
    ("abcdef", 6),    # entire string
    ("aab", 2),       # "ab"
]

print("Version 1: Set-based")
for s, expected in test_cases:
    result = length_of_longest_substring(s)
    status = "✅" if result == expected else "❌"
    print(f"  {status} '{s}' → {result} (expected {expected})")
````

````python
# ═══════════════════════════════════════════════════════════════
# VERSION 2: Dict-based — Faster in Practice (Jump Left Pointer)
# ═══════════════════════════════════════════════════════════════

def length_of_longest_substring_dict(s):
    """Dict version — stores last seen index for each character.
    
    Instead of shrinking one character at a time (while loop),
    JUMP left directly past the duplicate.
    
    Key check: `seen[ch] >= left` ensures the duplicate is actually
    within our current window, not from a previous window.
    
    Time: O(n), Space: O(min(n, alphabet))
    """
    seen = {}
    left = 0
    length = 0
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1  # Jump past the duplicate!
        seen[ch] = right
        length = max(length, right - left + 1)
    return length

print("Version 2: Dict-based")
for s, expected in test_cases:
    result = length_of_longest_substring_dict(s)
    status = "✅" if result == expected else "❌"
    print(f"  {status} '{s}' → {result} (expected {expected})")
````

````python
# ═══════════════════════════════════════════════════════════════
# VERSION 3: One-liner Style (for insight, not speed)
# ═══════════════════════════════════════════════════════════════

from functools import reduce

def length_of_longest_substring_oneliner(s):
    """Compact version using reduce. Not recommended for contests,
    but shows the power of Python's functional tools."""
    return reduce(
        lambda state, ch: (
            state[0][state[0].index(ch)+1:] + ch if ch in state[0] else state[0] + ch,
            max(state[1], len(state[0]) + (0 if ch in state[0] else 1))
        ),
        s,
        ('', 0)
    )[1] if s else 0

print("Version 3: One-liner style")
print(f"  'abcabcbb' → {length_of_longest_substring_oneliner('abcabcbb')}")
print(f"  'bbbbb' → {length_of_longest_substring_oneliner('bbbbb')}")
print(f"  'pwwkew' → {length_of_longest_substring_oneliner('pwwkew')}")
````

### 📊 Character-by-Character Trace Tables

````python
# Full trace function for visualization
def trace_longest_substring(s):
    """Visual trace of the set-based sliding window."""
    window = set()
    left = 0
    length = 0
    
    print(f"\n{'='*70}")
    print(f"Tracing: \"{s}\"")
    print(f"{'='*70}")
    print(f"{'right':<6} {'char':<6} {'action':<35} {'window':<20} {'left':<6} {'len':<5} {'max'}")
    print("-" * 85)
    
    for right in range(len(s)):
        actions = []
        while s[right] in window:
            actions.append(f"rm '{s[left]}' L→{left+1}")
            window.remove(s[left])
            left += 1
        window.add(s[right])
        cur_len = right - left + 1
        length = max(length, cur_len)
        
        action_str = "; ".join(actions) + (f"; add '{s[right]}'" if actions else f"add '{s[right]}'")
        window_str = "{" + ",".join(sorted(window)) + "}"
        print(f"{right:<6} '{s[right]}'   {action_str:<35} {window_str:<20} {left:<6} {cur_len:<5} {length}")
    
    print(f"\n🏆 Answer: {length}")
    return length

# Trace all three test strings
trace_longest_substring("abcabcbb")
trace_longest_substring("bbbbb")
trace_longest_substring("pwwkew")
````

### 💡 Why It's O(n)

Each character is:
- **Added** to the window exactly **once** (when `right` reaches it)
- **Removed** from the window at most **once** (when `left` passes it)

Both `left` and `right` only move forward (never backward).
Total operations: at most **2n**.

**O(2n) = O(n)** ✅

### ⚠️ Common Mistakes

1. **Off-by-one on left pointer:** `left = seen[ch] + 1`, not `seen[ch]`
2. **Forgetting `seen[ch] >= left` check:** Without this, stale entries in the dict cause wrong jumps
3. **Wrong max update position:** Update AFTER adding the character, not before
4. **Using `len(window)` instead of `right - left + 1`:** Same value, but `right - left + 1` is clearer

## 🎯 Sliding Window Variations

````python
# Problem: Best Time to Buy and Sell Stock
# This is actually a sliding window problem!

def max_profit(prices):
    """Track min price (left) and max profit (right - left).
    Time: O(n), Space: O(1)"""
    min_price = float('inf')
    max_profit = 0
    for price in prices:
        min_price = min(min_price, price)
        max_profit = max(max_profit, price - min_price)
    return max_profit

print("Buy/Sell Stock:")
print("  [7,1,5,3,6,4]:", max_profit([7,1,5,3,6,4]))  # 5
````

````python
from collections import defaultdict, Counter

# Problem: Longest Substring with At Most K Distinct Characters
def longest_k_distinct(s, k):
    """Variable window: shrink when distinct count > k.
    Time: O(n), Space: O(k)"""
    freq = defaultdict(int)
    left = 0
    max_len = 0
    for right in range(len(s)):
        freq[s[right]] += 1
        while len(freq) > k:
            freq[s[left]] -= 1
            if freq[s[left]] == 0:
                del freq[s[left]]
            left += 1
        max_len = max(max_len, right - left + 1)
    return max_len

print("At Most K Distinct:")
print("  'eceba', k=2:", longest_k_distinct("eceba", 2))  # 3 ("ece")
print("  'aa', k=1:", longest_k_distinct("aa", 1))        # 2
````

````python
# Problem: Longest Repeating Character Replacement
def character_replacement(s, k):
    """Window valid when (window_size - max_freq_in_window) <= k.
    Time: O(n), Space: O(1)"""
    freq = defaultdict(int)
    left = 0
    max_freq = 0
    max_len = 0
    for right in range(len(s)):
        freq[s[right]] += 1
        max_freq = max(max_freq, freq[s[right]])
        if (right - left + 1) - max_freq > k:
            freq[s[left]] -= 1
            left += 1
        max_len = max(max_len, right - left + 1)
    return max_len

print("Longest Repeating Character Replacement:")
print("  'ABAB', k=2:", character_replacement("ABAB", 2))       # 4
print("  'AABABBA', k=1:", character_replacement("AABABBA", 1))  # 4
````

````python
# Problem: Permutation in String
def check_inclusion(s1, s2):
    """Fixed window of size len(s1). Compare counters.
    Time: O(n), Space: O(1)"""
    if len(s1) > len(s2):
        return False
    need = Counter(s1)
    window = Counter(s2[:len(s1)])
    if window == need:
        return True
    for right in range(len(s1), len(s2)):
        window[s2[right]] += 1
        old = s2[right - len(s1)]
        window[old] -= 1
        if window[old] == 0:
            del window[old]
        if window == need:
            return True
    return False

print("Permutation in String:")
print("  'ab' in 'eidbaooo':", check_inclusion("ab", "eidbaooo"))  # True ("ba")
print("  'ab' in 'eidboaoo':", check_inclusion("ab", "eidboaoo"))  # False
````

````python
# Problem: Find All Anagrams in a String
def find_anagrams(s, p):
    """Fixed window + counter comparison.
    Time: O(n), Space: O(1)"""
    if len(p) > len(s):
        return []
    need = Counter(p)
    window = Counter()
    result = []
    for right in range(len(s)):
        window[s[right]] += 1
        if right >= len(p):
            old = s[right - len(p)]
            window[old] -= 1
            if window[old] == 0:
                del window[old]
        if window == need:
            result.append(right - len(p) + 1)
    return result

print("Find All Anagrams:")
print("  s='cbaebabacd', p='abc':", find_anagrams("cbaebabacd", "abc"))  # [0, 6]
````

````python
# Problem: Minimum Window Substring
def min_window(s, t):
    """Variable window: expand right until valid, shrink left to minimize.
    Time: O(|s| + |t|), Space: O(|s| + |t|)"""
    if not t or not s:
        return ""
    need = Counter(t)
    required = len(need)
    have = 0
    window_freq = defaultdict(int)
    left = 0
    best = (float('inf'), 0, 0)  # (length, left, right)
    
    for right in range(len(s)):
        ch = s[right]
        window_freq[ch] += 1
        if ch in need and window_freq[ch] == need[ch]:
            have += 1
        
        while have == required:
            if right - left + 1 < best[0]:
                best = (right - left + 1, left, right)
            left_ch = s[left]
            window_freq[left_ch] -= 1
            if left_ch in need and window_freq[left_ch] < need[left_ch]:
                have -= 1
            left += 1
    
    return s[best[1]:best[2]+1] if best[0] != float('inf') else ""

print("Minimum Window Substring:")
print("  s='ADOBECODEBANC', t='ABC':", min_window("ADOBECODEBANC", "ABC"))  # "BANC"
````

## 📋 Recognition Table: Which Keywords Map to Which Variant

| Keywords | Variant | Window Type | State |
|----------|---------|-------------|-------|
| "longest substring without repeating" | Unique chars | Variable | Set |
| "at most K distinct characters" | K distinct | Variable | Dict (freq) |
| "longest after K replacements" | Char replacement | Variable | Dict + max_freq |
| "find all anagrams" | Anagram | Fixed | Counter |
| "permutation in string" | Permutation | Fixed | Counter |
| "minimum window containing all" | Min window | Variable | Counter + have/need |
| "maximum sum of K elements" | Max sum | Fixed | Running sum |
| "shortest subarray with sum ≥ X" | Min length | Variable | Running sum |

### ⚡ Contest Speed Tricks

1. **Always use `defaultdict(int)` or `Counter`** — avoids KeyError boilerplate
2. **Delete keys when count reaches 0** — keeps `len(dict)` accurate for distinct counting
3. **For fixed window:** Use `right >= k` to know when to start removing old elements
4. **For variable window:** The while loop always checks the SAME character that was just added
5. **Window size formula:** Always `right - left + 1`
