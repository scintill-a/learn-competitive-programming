## ⏱️ Big-O Intuition: What O(n), O(log n), O(n²) FEEL Like in Practice

Let's make Big-O concrete. We'll time real Python code to build intuition.

````python
import time

def time_it(func, *args, label=""):
    """Time a function and print results."""
    start = time.perf_counter()
    result = func(*args)
    elapsed = time.perf_counter() - start
    print(f"  {label}: {elapsed:.6f} seconds")
    return result
````

````python
# O(n) — Linear: Feels fast. Even 10 million elements is fine.
def linear_search(arr, target):
    for x in arr:
        if x == target:
            return True
    return False

# O(n²) — Quadratic: Gets SLOW fast. 10K elements = noticeable, 100K = painful.
def quadratic_check(arr):
    count = 0
    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[i] + arr[j] == 0:
                count += 1
    return count

# O(log n) — Logarithmic: Feels instant. Even billions of elements.
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

# Let's feel the difference!
import random
random.seed(42)

sizes = [1000, 5000, 10000]
for n in sizes:
    arr = list(range(n))
    print(f"\nn = {n:,}:")
    time_it(linear_search, arr, n - 1, label=f"O(n) linear search")
    time_it(binary_search, arr, n - 1, label=f"O(log n) binary search")
    time_it(quadratic_check, arr[:n], label=f"O(n²) quadratic check")
````

### 💡 Key Takeaways

- **O(log n)** is basically instant — even for n = 10⁹, it's only ~30 operations
- **O(n)** is fast — 10⁶ elements in milliseconds
- **O(n²)** gets painful quickly — 10⁴ elements already takes seconds
- **O(n³)** is only viable for n ≤ 500

**THE CONTEST RULE:** Python does ~10⁷ operations per second. If your algorithm does more than 10⁸ operations, it will TLE (Time Limit Exceeded).

## 🐍 Python Performance Tips for Contests

````python
# Tip 1: Use sys.stdin for fast input
import sys
# In contests: input = sys.stdin.readline
# This is 3-5x faster than built-in input()

# Tip 2: set/dict lookup is O(1), list search is O(n)
big_list = list(range(100000))
big_set = set(range(100000))

print("Searching for 99999:")
time_it(lambda: 99999 in big_list, label="List search O(n)")
time_it(lambda: 99999 in big_set, label="Set search O(1)")
print("\n💡 Set lookup is 1000x+ faster. ALWAYS use sets/dicts for lookups.")
````

````python
# Tip 3: List comprehensions are faster than loops
n = 1000000

# Slow way
def loop_way():
    result = []
    for i in range(n):
        result.append(i * 2)
    return result

# Fast way
def comprehension_way():
    return [i * 2 for i in range(n)]

time_it(loop_way, label="Loop + append")
time_it(comprehension_way, label="List comprehension")
print("\n💡 Comprehensions are ~30% faster due to internal optimization.")
````

````python
# Tip 4: Use collections.Counter instead of manual counting
from collections import Counter

text = "abracadabra" * 10000

def manual_count(s):
    freq = {}
    for c in s:
        freq[c] = freq.get(c, 0) + 1
    return freq

def counter_count(s):
    return Counter(s)

time_it(manual_count, text, label="Manual dict counting")
time_it(counter_count, text, label="Counter (C-optimized)")
print("\n💡 Counter is implemented in C and is much faster.")
````

````python
# Tip 5: deque for BFS, not list
from collections import deque

n = 100000

def list_queue():
    q = list(range(n))
    while q:
        q.pop(0)  # O(n) per operation!

def deque_queue():
    q = deque(range(n))
    while q:
        q.popleft()  # O(1) per operation!

time_it(deque_queue, label="deque.popleft() — O(1)")
# time_it(list_queue, label="list.pop(0) — O(n)")  # Uncomment to see how slow it is
print("\n💡 NEVER use list.pop(0) for BFS. Use deque.popleft().")
print("   list.pop(0) is O(n) per call → O(n²) total for BFS. That's TLE!")
````

## 🧠 The "Brute Force First" Principle

**Always start with brute force.** Here's why:

1. It proves you understand the problem
2. It gives you a correct solution to test against
3. The brute force reveals what's SLOW, which tells you what to optimize

### The Optimization Pipeline

```
Brute Force → Identify Bottleneck → Apply Pattern → Optimal Solution
```

### Pattern: When You See a Loop That Searches → Think Hash Map

The most common optimization in competitive programming:

```python
# BEFORE (O(n²)): Inner loop searches for something
for i in range(n):
    for j in range(n):         # ← This loop SEARCHES
        if condition(arr[i], arr[j]):
            return True

# AFTER (O(n)): Hash map replaces the search
seen = {}  # or set()
for i in range(n):
    if need_value in seen:     # ← O(1) lookup replaces O(n) loop
        return True
    seen[arr[i]] = i
```

This single transformation covers Two Sum, Contains Duplicate, Two Sum II, and dozens more.

## 🔥🔥🔥 REAL CONTEST PROBLEM DEEP DIVE #1: Two Sum (No Loops)

### The Problem

> Given an array of integers `nums` and a target integer, return the indices of the two numbers that add up to the target. You may assume that each input has exactly one solution.

### Why Nested Loops Are WRONG in Contests

The naive approach:

````python
# ❌ WRONG — O(n²) — Will get TLE in any serious contest
def two_sum_brute(nums, target):
    """Check every pair. Time: O(n²), Space: O(1)."""
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

# It works, but it's O(n²). With n = 100,000 that's 10 BILLION operations.
# At Python's ~10⁷ ops/sec, that's ~1000 seconds. Contest time limit: 1-2 seconds.
print("Brute force result:", two_sum_brute([2, 7, 11, 15], 9))
print("Works... but too slow for contests.")
````

### 💡 The Complement Trick

The key mathematical insight:

```
If a + b = target
Then b = target - a
```

So for each number `a`, we don't need to search for `b` with a loop.
We just compute `b = target - a` and check if we've SEEN it before.

A hash map gives us O(1) lookup for "have we seen this number?".

````python
# ✅ OPTIMAL — O(n) — The hash map complement trick
def two_sum(nums, target):
    """Hash map complement lookup. Time: O(n), Space: O(n)."""
    seen = {}  # number → index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

# Test it
print("Test 1:", two_sum([2, 7, 11, 15], 9))   # [0, 1]
print("Test 2:", two_sum([3, 2, 4], 6))          # [1, 2]
print("Test 3:", two_sum([3, 3], 6))              # [0, 1]
````

````python
# 🔍 Step-by-step trace: two_sum([2, 7, 11, 15], 9)
def two_sum_traced(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        print(f"  Step {i}: num={num}, complement={complement}, seen={seen}")
        if complement in seen:
            print(f"  ✅ Found! complement {complement} is at index {seen[complement]}")
            return [seen[complement], i]
        seen[num] = i
        print(f"         Added {num}→{i} to seen")
    return []

print("Trace: two_sum([2, 7, 11, 15], 9)")
print("="*50)
result = two_sum_traced([2, 7, 11, 15], 9)
print(f"\nResult: {result}")
````

### 🔥 The Recursive Version (True No Loops)

Want to solve it with zero loops? Use recursion to pass the state forward:

````python
# Version 2: Recursion — no explicit loops
def two_sum_recursive(nums, target, index=0, seen=None):
    """Recursive Two Sum. Time: O(n), Space: O(n)."""
    if seen is None:
        seen = {}
    if index >= len(nums):
        return []

    curr = nums[index]
    comp = target - curr
    if comp in seen:
        return [seen[comp], index]

    seen[curr] = index
    return two_sum_recursive(nums, target, index + 1, seen)

print("Recursive version:")
print("  [2,7,11,15], target=9:", two_sum_recursive([2, 7, 11, 15], 9))  # [0, 1]
print("  [1,2,3], target=10:", two_sum_recursive([1, 2, 3], 10))          # []
print("  [3,3], target=6:", two_sum_recursive([3, 3], 6))                 # [0, 1]
print("  [3,5], target=6:", two_sum_recursive([3, 5], 6))                 # []
````

````python
# Version 3: Dict comprehension — build index map in one line
def two_sum_dictcomp(nums, target):
    """Dict comprehension approach."""
    d = {num: i for i, num in enumerate(nums)}
    return next(
        ([i, d[target - num]] for i, num in enumerate(nums)
         if target - num in d and d[target - num] != i),
        []
    )

print("Dict comprehension version:")
print("  [2,7,11,15], target=9:", two_sum_dictcomp([2, 7, 11, 15], 9))  # [0, 1]
print("  [3,2,4], target=6:", two_sum_dictcomp([3, 2, 4], 6))            # [1, 2]
````

### 🎯 5 Two Sum Variations (Increasing Difficulty)

````python
# Variation 1: Two Sum — check existence only (simplest)
def has_pair_with_sum(nums, target):
    seen = set()
    for num in nums:
        if target - num in seen:
            return True
        seen.add(num)
    return False

print("V1 - Has pair:", has_pair_with_sum([1, 4, 7, 2], 9))  # True (7+2)
print("V1 - No pair:", has_pair_with_sum([1, 4, 7, 2], 20))  # False
````

````python
# Variation 2: Two Sum on sorted array (Two Pointers — O(1) space)
def two_sum_sorted(nums, target):
    """Input is SORTED. Use two pointers instead of hash map."""
    left, right = 0, len(nums) - 1
    while left < right:
        s = nums[left] + nums[right]
        if s == target:
            return [left, right]
        elif s < target:
            left += 1
        else:
            right -= 1
    return []

print("V2 - Sorted:", two_sum_sorted([1, 2, 3, 4, 6], 6))  # [1, 3] (2+4)
````

````python
# Variation 3: Count pairs with target sum
def count_pairs(nums, target):
    """Count unique VALUE pairs that sum to target."""
    from collections import Counter
    freq = Counter(nums)
    count = 0
    seen = set()
    for num in freq:
        comp = target - num
        pair = (min(num, comp), max(num, comp))
        if comp in freq and pair not in seen:
            if num != comp or freq[num] >= 2:
                count += 1
                seen.add(pair)
    return count

print("V3 - Count pairs:", count_pairs([1, 5, 7, -1, 5], 6))  # 2: (1,5) and (7,-1)
````

````python
# Variation 4: Two Sum — return the VALUES (not indices)
def two_sum_values(nums, target):
    """Return the pair of values (not indices) that sum to target."""
    seen = set()
    for num in nums:
        comp = target - num
        if comp in seen:
            return (comp, num)
        seen.add(num)
    return None

print("V4 - Values:", two_sum_values([2, 7, 11, 15], 9))  # (2, 7)
````

````python
# Variation 5: Two Sum — difference instead of sum
def two_diff(nums, target):
    """Find pair where |a - b| = target. Uses same hash map trick!"""
    seen = set()
    for num in nums:
        if num + target in seen or num - target in seen:
            return True
        seen.add(num)
    return False

print("V5 - Difference:", two_diff([1, 5, 3, 4, 2], 2))  # True (5-3, 4-2, 3-1)
````

## 🔥🔥🔥 REAL CONTEST PROBLEM DEEP DIVE #2: Longest Substring Without Repeating Characters

### ⚠️ This exact problem appeared as HARD in a past contest. You WILL see this again.

### The Problem

> Given a string `s`, find the length of the **longest substring** without repeating characters.
>
> Write: `def length_of_longest_substring(s: str) -> int`

**Examples:**
- `"abcabcbb"` → `3` (substring `"abc"`)
- `"bbbbb"` → `1` (substring `"b"`)
- `"pwwkew"` → `3` (substring `"wke"`)
- `""` → `0`

### Recognition Trigger

When you see:
- "longest substring" + "without repeating" → **SLIDING WINDOW + SET**
- Any variation of "longest/shortest subarray/substring with constraint"

Instant pattern match. No thinking required.

````python
# ❌ BRUTE FORCE O(n³) — Check ALL substrings, verify uniqueness
# DO NOT use in contests. Shown only to understand what we're optimizing.

def length_of_longest_substring_brute(s):
    """Check every substring. O(n³) — TLE for sure."""
    n = len(s)
    max_len = 0
    for i in range(n):
        for j in range(i, n):
            substring = s[i:j+1]
            if len(set(substring)) == len(substring):  # All unique?
                max_len = max(max_len, j - i + 1)
    return max_len

print("Brute force O(n³):")
print("  'abcabcbb' →", length_of_longest_substring_brute("abcabcbb"))  # 3
print("  'bbbbb' →", length_of_longest_substring_brute("bbbbb"))        # 1
print("  'pwwkew' →", length_of_longest_substring_brute("pwwkew"))      # 3
print("\n  Works but O(n³) = TLE. We need O(n).")
````

````python
# ⚠️ BETTER BRUTE FORCE O(n²) — Check from each starting position
def length_of_longest_substring_better(s):
    """From each start, extend right until duplicate. O(n²) — still slow."""
    n = len(s)
    max_len = 0
    for i in range(n):
        seen = set()
        for j in range(i, n):
            if s[j] in seen:
                break
            seen.add(s[j])
            max_len = max(max_len, j - i + 1)
    return max_len

print("Better brute force O(n²):")
print("  'abcabcbb' →", length_of_longest_substring_better("abcabcbb"))  # 3
print("\n  Better, but still too slow for n = 10⁵.")
````

### ✅ OPTIMAL: Variable Sliding Window with a Set — O(n)

**The Algorithm:**

1. Use two pointers: `left` and `right`, both start at 0
2. Use a `set` to track characters in the current window
3. Expand `right` pointer one step at a time
4. If `s[right]` is already in the set → **collision!** Shrink from `left` until it's gone
5. Add `s[right]` to the set
6. Update `max_length = max(max_length, right - left + 1)`

**Why it's O(n):** Each character is added to the set at most once (when `right` passes it) and removed at most once (when `left` passes it). Total operations: at most 2n.

````python
# ✅ OPTIMAL — Set-based sliding window — O(n) time, O(min(n, alphabet)) space
# THIS IS THE CONTEST VERSION. MEMORIZE THIS.

def length_of_longest_substring(s):
    window = set()
    left = 0
    length = 0
    for right in range(len(s)):
        while s[right] in window:
            window.remove(s[left])
            left += 1
        window.add(s[right])
        length = max(length, right - left + 1)
    return length

# Test
print("Optimal O(n) — Set-based:")
print("  'abcabcbb' →", length_of_longest_substring("abcabcbb"))  # 3
print("  'bbbbb' →", length_of_longest_substring("bbbbb"))        # 1
print("  'pwwkew' →", length_of_longest_substring("pwwkew"))      # 3
print("  '' →", length_of_longest_substring(""))                   # 0
print("  ' ' →", length_of_longest_substring(" "))                 # 1
print("  'dvdf' →", length_of_longest_substring("dvdf"))          # 3
````

````python
# 🔍 FULL CHARACTER-BY-CHARACTER TRACE for "abcabcbb"

def length_of_longest_substring_traced(s):
    window = set()
    left = 0
    length = 0
    
    print(f"Tracing: \"{s}\"")
    print(f"{'Step':<6} {'right':<6} {'s[r]':<6} {'Action':<40} {'Window':<20} {'left':<6} {'max_len'}")
    print("=" * 90)
    
    for right in range(len(s)):
        step_actions = []
        while s[right] in window:
            step_actions.append(f"Remove '{s[left]}', left→{left+1}")
            window.remove(s[left])
            left += 1
        
        window.add(s[right])
        length = max(length, right - left + 1)
        
        action = "Add '" + s[right] + "'"
        if step_actions:
            action = "; ".join(step_actions) + "; " + action
        
        print(f"{right:<6} {right:<6} '{s[right]}'   {action:<40} {str(window):<20} {left:<6} {length}")
    
    print(f"\n🏆 Answer: {length}")
    return length

length_of_longest_substring_traced("abcabcbb")
````

````python
# ✅ ALTERNATIVE — Dict-based (jump left pointer directly, fewer iterations)
def length_of_longest_substring_dict(s):
    """Dict version — stores last seen index. Jumps left pointer directly."""
    seen = {}  # character → last index
    left = 0
    length = 0
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1  # Jump past the duplicate!
        seen[ch] = right
        length = max(length, right - left + 1)
    return length

print("Dict-based version:")
print("  'abcabcbb' →", length_of_longest_substring_dict("abcabcbb"))  # 3
print("  'bbbbb' →", length_of_longest_substring_dict("bbbbb"))        # 1
print("  'pwwkew' →", length_of_longest_substring_dict("pwwkew"))      # 3
print("  'dvdf' →", length_of_longest_substring_dict("dvdf"))          # 3

print("\n💡 The dict version avoids the inner while loop entirely.")
print("   Instead of shrinking one character at a time, it JUMPS left")
print("   directly to the position after the duplicate.")
````

### 🎯 Sliding Window Variations

Once you master the base problem, these variations become straightforward:

````python
# Variation 1: Longest Substring with At Most K Distinct Characters
# Same sliding window, but condition changes from "all unique" to "at most K distinct"

from collections import defaultdict

def longest_k_distinct(s, k):
    freq = defaultdict(int)
    left = 0
    max_len = 0
    for right in range(len(s)):
        freq[s[right]] += 1
        while len(freq) > k:  # More than K distinct → shrink
            freq[s[left]] -= 1
            if freq[s[left]] == 0:
                del freq[s[left]]
            left += 1
        max_len = max(max_len, right - left + 1)
    return max_len

print("Longest with at most K distinct:")
print("  'eceba', k=2:", longest_k_distinct("eceba", 2))        # 3 ("ece")
print("  'abcadcacacaca', k=3:", longest_k_distinct("abcadcacacaca", 3))  # 11
````

````python
# Variation 2: Longest Repeating Character Replacement
# Window valid when (window_size - max_frequency) <= k

def longest_repeating_replacement(s, k):
    freq = defaultdict(int)
    left = 0
    max_freq = 0
    max_len = 0
    for right in range(len(s)):
        freq[s[right]] += 1
        max_freq = max(max_freq, freq[s[right]])
        if (right - left + 1) - max_freq > k:
            freq[s[left]] -= 1
            left += 1
        max_len = max(max_len, right - left + 1)
    return max_len

print("Longest repeating char replacement:")
print("  'ABAB', k=2:", longest_repeating_replacement("ABAB", 2))     # 4
print("  'AABABBA', k=1:", longest_repeating_replacement("AABABBA", 1))  # 4
````

````python
# Problem 1: Contains Duplicate
# Given an array, return True if any value appears at least twice.
# Brute: O(n²) — compare all pairs. Optimal: O(n) — use a set.

def contains_duplicate(nums):
    return len(nums) != len(set(nums))

# Alternative with early exit:
def contains_duplicate_v2(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False

print("Contains Duplicate:")
print("  [1,2,3,1]:", contains_duplicate([1,2,3,1]))   # True
print("  [1,2,3,4]:", contains_duplicate([1,2,3,4]))   # False
# Time: O(n), Space: O(n)
````

````python
# Problem 2: Valid Anagram
# Given two strings s and t, return True if t is an anagram of s.

from collections import Counter

def is_anagram(s, t):
    return Counter(s) == Counter(t)

print("Valid Anagram:")
print("  'anagram', 'nagaram':", is_anagram("anagram", "nagaram"))  # True
print("  'rat', 'car':", is_anagram("rat", "car"))                  # False
# Time: O(n), Space: O(1) — at most 26 characters
````

````python
# Problem 3: Maximum Subarray (Kadane's Algorithm)
# Find the contiguous subarray with the largest sum.

def max_subarray(nums):
    max_sum = current = nums[0]
    for num in nums[1:]:
        current = max(num, current + num)  # Start fresh or extend?
        max_sum = max(max_sum, current)
    return max_sum

print("Maximum Subarray:")
print("  [-2,1,-3,4,-1,2,1,-5,4]:", max_subarray([-2,1,-3,4,-1,2,1,-5,4]))  # 6
print("  [1]:", max_subarray([1]))  # 1
# Time: O(n), Space: O(1)
````

````python
# Problem 4: Best Time to Buy and Sell Stock
# Find max profit from buying and selling one stock.

def max_profit(prices):
    min_price = float('inf')
    max_profit = 0
    for price in prices:
        min_price = min(min_price, price)
        max_profit = max(max_profit, price - min_price)
    return max_profit

print("Max Profit:")
print("  [7,1,5,3,6,4]:", max_profit([7,1,5,3,6,4]))  # 5 (buy at 1, sell at 6)
print("  [7,6,4,3,1]:", max_profit([7,6,4,3,1]))      # 0 (no profit possible)
# Time: O(n), Space: O(1)
````

````python
# Problem 5: Move Zeroes
# Move all 0's to end, maintaining relative order of non-zero elements.

def move_zeroes(nums):
    write = 0  # Position to write next non-zero
    for read in range(len(nums)):
        if nums[read] != 0:
            nums[write], nums[read] = nums[read], nums[write]
            write += 1
    return nums

print("Move Zeroes:")
print("  [0,1,0,3,12]:", move_zeroes([0,1,0,3,12]))  # [1,3,12,0,0]
# Time: O(n), Space: O(1)
````

````python
# Problem 6: Single Number
# Every element appears twice except one. Find the single one.

def single_number(nums):
    result = 0
    for num in nums:
        result ^= num  # XOR: a ^ a = 0, a ^ 0 = a
    return result

print("Single Number:")
print("  [2,2,1]:", single_number([2,2,1]))      # 1
print("  [4,1,2,1,2]:", single_number([4,1,2,1,2]))  # 4
# Time: O(n), Space: O(1) — XOR is the ultimate contest trick
````

````python
# Problem 7: Reverse String (in-place)
def reverse_string(s):
    left, right = 0, len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1
    return s

print("Reverse String:")
print("  ['h','e','l','l','o']:", reverse_string(list("hello")))  # ['o','l','l','e','h']
# Time: O(n), Space: O(1)
````

````python
# Problem 8: Valid Palindrome
def is_palindrome(s):
    cleaned = [c.lower() for c in s if c.isalnum()]
    return cleaned == cleaned[::-1]

# Contest one-liner:
def is_palindrome_oneliner(s):
    s = ''.join(c.lower() for c in s if c.isalnum())
    return s == s[::-1]

print("Valid Palindrome:")
print("  'A man, a plan, a canal: Panama':", is_palindrome("A man, a plan, a canal: Panama"))  # True
print("  'race a car':", is_palindrome("race a car"))  # False
# Time: O(n), Space: O(n)
````

````python
# Problem 9: Missing Number
# Given array containing n distinct numbers in [0, n], find missing one.

def missing_number(nums):
    n = len(nums)
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

# Alternative: XOR trick
def missing_number_xor(nums):
    result = len(nums)
    for i, num in enumerate(nums):
        result ^= i ^ num
    return result

print("Missing Number:")
print("  [3,0,1]:", missing_number([3,0,1]))    # 2
print("  [9,6,4,2,3,5,7,0,1]:", missing_number([9,6,4,2,3,5,7,0,1]))  # 8
# Time: O(n), Space: O(1)
````

````python
# Problem 10: Majority Element
# Find element that appears more than n/2 times.

def majority_element(nums):
    """Boyer-Moore Voting Algorithm — O(n) time, O(1) space."""
    candidate = nums[0]
    count = 0
    for num in nums:
        if count == 0:
            candidate = num
        count += (1 if num == candidate else -1)
    return candidate

# One-liner with Counter:
def majority_element_counter(nums):
    return Counter(nums).most_common(1)[0][0]

print("Majority Element:")
print("  [3,2,3]:", majority_element([3,2,3]))      # 3
print("  [2,2,1,1,1,2,2]:", majority_element([2,2,1,1,1,2,2]))  # 2
````

````python
# Problem 11: Intersection of Two Arrays
def intersection(nums1, nums2):
    return list(set(nums1) & set(nums2))  # Set intersection!

print("Intersection:")
print("  [1,2,2,1] & [2,2]:", intersection([1,2,2,1], [2,2]))  # [2]
print("  [4,9,5] & [9,4,9,8,4]:", intersection([4,9,5], [9,4,9,8,4]))  # [4, 9] (order may vary)
# Time: O(n + m), Space: O(n + m)
````

````python
# Problem 12: Climbing Stairs (Intro to DP)
# You can climb 1 or 2 steps. How many distinct ways to reach the top?

def climb_stairs(n):
    """This is just Fibonacci! dp[i] = dp[i-1] + dp[i-2]."""
    if n <= 2:
        return n
    a, b = 1, 2
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b

print("Climbing Stairs:")
print("  n=2:", climb_stairs(2))  # 2 (1+1 or 2)
print("  n=3:", climb_stairs(3))  # 3 (1+1+1 or 1+2 or 2+1)
print("  n=5:", climb_stairs(5))  # 8
# Time: O(n), Space: O(1)
````

5. **Always start with brute force**, then optimize.
6. **Python tips:** Use set for lookups, Counter for frequency, deque for BFS.
