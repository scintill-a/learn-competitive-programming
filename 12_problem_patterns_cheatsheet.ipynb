## ⚡ Decision Flowchart

### Step 1: What are you looking for?

```
Looking for a PAIR?              → Hash Map (Two Sum pattern)
Looking for OPTIMAL SUBARRAY?    → Sliding Window or Kadane's
Looking for ALL possibilities?   → Backtracking
Looking for COUNT or MIN/MAX?    → DP
Looking for SHORTEST PATH?       → BFS (unweighted) or Dijkstra (weighted)
```

### Step 2: What's the input?

```
Input is SORTED?                 → Binary Search or Two Pointers
Input is a TREE?                 → DFS/BFS
Input is a GRAPH?                → DFS/BFS/Dijkstra/Union-Find
Input is a GRID?                 → Treat as graph (4-directional neighbors)
Input is a STRING?               → Hash Map/Sliding Window/DP
Input is INTERVALS?              → Sort + merge/sweep
```

### Step 3: What's the constraint?

```
n ≤ 20     → Bitmask DP, Brute Force O(2ⁿ)
n ≤ 100    → O(n³) DP
n ≤ 1000   → O(n²) DP
n ≤ 10⁵    → O(n log n) Sorting/Binary Search
n ≤ 10⁶    → O(n) Hash Map/Sliding Window
n ≤ 10⁹    → O(log n) Binary Search or O(1) Math
```

## 🔥 The 10 Templates You Must Memorize

### Template 1: Hash Map Lookup

````python
# When: "find pair", "complement", "target sum"
def hash_map_lookup(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []
````

### Template 2: Variable Sliding Window

````python
# When: "longest/shortest subarray/substring with condition"
def sliding_window(arr):
    left = 0
    state = set()  # or dict, or counter
    best = 0
    for right in range(len(arr)):
        # EXPAND: add arr[right] to state
        state.add(arr[right])
        # SHRINK: while constraint is violated
        while False:  # constraint_violated(state)
            state.discard(arr[left])
            left += 1
        # UPDATE: current window [left..right] is valid
        best = max(best, right - left + 1)
    return best
````

### Template 3: Binary Search

````python
# When: sorted array, or monotonic answer space
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = (lo + hi) // 2
        if arr[mid] == target: return mid
        elif arr[mid] < target: lo = mid + 1
        else: hi = mid - 1
    return -1

def search_on_answer(lo, hi, predicate):
    while lo < hi:
        mid = (lo + hi) // 2
        if predicate(mid): hi = mid
        else: lo = mid + 1
    return lo
````

### Template 4: BFS (Shortest Path)

````python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    while queue:
        node = queue.popleft()
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
````

### Template 5: DFS (Graph/Grid)

````python
def dfs_grid(grid, r, c, visited):
    if r < 0 or r >= len(grid) or c < 0 or c >= len(grid[0]):
        return
    if (r, c) in visited or grid[r][c] == 0:
        return
    visited.add((r, c))
    for dr, dc in [(0,1),(0,-1),(1,0),(-1,0)]:
        dfs_grid(grid, r+dr, c+dc, visited)
````

### Template 6: Backtracking

````python
def backtrack(candidates, current, start, result):
    if len(current) == target_length:  # or sum == target, etc.
        result.append(current[:])
        return
    for i in range(start, len(candidates)):
        current.append(candidates[i])
        backtrack(candidates, current, i + 1, result)  # i+1 for no reuse, i for reuse
        current.pop()  # BACKTRACK

target_length = 2  # example
result = []
backtrack([1,2,3], [], 0, result)
print("Backtrack result:", result)
````

### Template 7: DP (1D Linear)

````python
# Bottom-up
def dp_linear(n):
    dp = [0] * (n + 1)
    dp[0] = 1  # base case
    for i in range(1, n + 1):
        dp[i] = dp[i-1] + dp[i-2] if i >= 2 else dp[i-1]
    return dp[n]

# Top-down (memoization)
from functools import lru_cache
@lru_cache(maxsize=None)
def dp_memo(n):
    if n <= 1: return 1
    return dp_memo(n-1) + dp_memo(n-2)
````

### Template 8: Dijkstra

````python
from heapq import heappush, heappop

def dijkstra(graph, start, n):
    dist = [float('inf')] * n
    dist[start] = 0
    heap = [(0, start)]
    while heap:
        d, u = heappop(heap)
        if d > dist[u]: continue
        for v, w in graph[u]:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                heappush(heap, (dist[v], v))
    return dist
````

### Template 9: Union-Find

````python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n
        self.components = n
    
    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]
    
    def union(self, x, y):
        px, py = self.find(x), self.find(y)
        if px == py: return False
        if self.rank[px] < self.rank[py]: px, py = py, px
        self.parent[py] = px
        if self.rank[px] == self.rank[py]: self.rank[px] += 1
        self.components -= 1
        return True

uf = UnionFind(5)
uf.union(0, 1); uf.union(2, 3); uf.union(1, 3)
print(f"Components: {uf.components}")  # 2
````

### Template 10: Topological Sort

````python
from collections import defaultdict, deque

def topo_sort(n, edges):
    graph = defaultdict(list)
    in_degree = [0] * n
    for u, v in edges:
        graph[u].append(v)
        in_degree[v] += 1
    queue = deque(i for i in range(n) if in_degree[i] == 0)
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for nb in graph[node]:
            in_degree[nb] -= 1
            if in_degree[nb] == 0:
                queue.append(nb)
    return order if len(order) == n else []  # Empty = cycle!

print("Topo sort:", topo_sort(4, [(0,1),(0,2),(1,3),(2,3)]))  # [0, 1, 2, 3] or [0, 2, 1, 3]
````

## 🧪 Edge Cases Checklist

Always test these before submitting:

| Input Type | Edge Cases |
|-----------|------------|
| Array | Empty `[]`, single element `[x]`, all same `[1,1,1]`, two elements |
| String | Empty `""`, single char `"a"`, all same `"aaa"` |
| Number | `0`, `1`, `-1`, `MAX_INT`, `MIN_INT` |
| Tree | `None`, single node, skewed (linked list), perfect binary |
| Graph | Disconnected, single node, self-loop, cycle |
| Binary Search | Not found, first element, last element, single element array |
| DP | Base cases, `n=0`, `n=1` |

## ⏱️ Complexity Quick Reference

| Complexity | n = 10⁶ operations | Typical Algorithm |

| O(1) | 1 | Hash lookup, math |
| O(log n) | 20 | Binary search |

| O(n log n) | 2×10⁷ | Sorting, heap |

**Rule of thumb:** Python can handle ~10⁷ operations per second.
For n = 10⁵, you need O(n log n) or better.
For n = 10⁶, you need O(n) or better.
