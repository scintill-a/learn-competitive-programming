## 📦 collections Cheat Sheet

```python
from collections import Counter, defaultdict, deque, OrderedDict, namedtuple

# === Counter ===
c = Counter("abracadabra")
print("Counter:", c)                           # Counter({'a': 5, 'b': 2, ...})
print("Most common 2:", c.most_common(2))      # [('a', 5), ('b', 2)]
print("Subtract:", c - Counter("abc"))          # Counter({'a': 4, 'b': 1, ...})
print("Elements:", list(c.elements())[:5])      # ['a', 'a', 'a', 'a', 'a']
print("Total:", c.total())                      # 11 (Python 3.10+)
c.update("xyz")                                 # Add counts
print("After update:", c['x'])                  # 1

# Counter arithmetic
c1 = Counter(a=3, b=1)
c2 = Counter(a=1, b=2)
print("c1 + c2:", c1 + c2)      # Counter({'a': 4, 'b': 3})
print("c1 - c2:", c1 - c2)      # Counter({'a': 2}) — only positive
print("c1 & c2:", c1 & c2)      # Counter({'a': 1, 'b': 1}) — min
print("c1 | c2:", c1 | c2)      # Counter({'a': 3, 'b': 2}) — max
```

```python
# === defaultdict ===
# No KeyError! Auto-creates missing keys.
graph = defaultdict(list)
graph[1].append(2)   # No need to check if key exists

freq = defaultdict(int)
for c in "hello": freq[c] += 1
print("defaultdict(int):", dict(freq))

groups = defaultdict(set)
groups['vowels'].add('a')
groups['vowels'].add('e')
print("defaultdict(set):", dict(groups))
```

```python
# === deque ===
# O(1) operations on both ends. ESSENTIAL for BFS.
q = deque([1, 2, 3])
q.appendleft(0)     # [0, 1, 2, 3] — O(1)
q.append(4)          # [0, 1, 2, 3, 4] — O(1)
q.popleft()          # 0 — O(1) (vs list.pop(0) which is O(n)!)
q.pop()              # 4 — O(1)
q.rotate(1)          # Right rotate: [3, 1, 2]
q.rotate(-1)         # Left rotate back: [1, 2, 3]
print("deque:", q)
```

```python
# === OrderedDict ===
# Dict that remembers insertion order (useful for LRU cache)
od = OrderedDict()
od['a'] = 1; od['b'] = 2; od['c'] = 3
od.move_to_end('a')           # Move to end
od.popitem(last=False)        # Pop from front (FIFO)
print("OrderedDict:", od)

# === namedtuple ===
Point = namedtuple('Point', ['x', 'y'])
p = Point(3, 4)
print(f"Point: ({p.x}, {p.y}), distance: {(p.x**2 + p.y**2)**0.5}")
```

## 📦 heapq Cheat Sheet

```python
import heapq

# Python heapq is MIN heap only
h = []
heapq.heappush(h, 3)    # Push
heapq.heappush(h, 1)
heapq.heappush(h, 2)
print("Min:", heapq.heappop(h))    # 1
print("Peek min:", h[0])            # 2 (don't pop)

# MAX heap trick: negate values
max_h = []
for v in [3, 1, 2]:
    heapq.heappush(max_h, -v)
print("Max:", -heapq.heappop(max_h))  # 3

# heapify — O(n) convert list to heap
arr = [5, 3, 1, 4, 2]
heapq.heapify(arr)

# heapreplace — pop min, push new (faster than pop+push)
heapq.heapreplace(arr, 10)  # Pop 1, push 10

# nlargest, nsmallest — O(n log k)
data = [15, 3, 7, 12, 1, 9]
print("3 largest:", heapq.nlargest(3, data))   # [15, 12, 9]
print("3 smallest:", heapq.nsmallest(3, data))  # [1, 3, 7]

# Merge sorted iterables
print("Merge:", list(heapq.merge([1,3,5], [2,4,6])))  # [1,2,3,4,5,6]
```

## 🔄 functools

```python
from functools import lru_cache, reduce, cmp_to_key, cache

# lru_cache — memoization for DP
@lru_cache(maxsize=None)
def fib(n):
    if n <= 1: return n
    return fib(n-1) + fib(n-2)
print("fib(50):", fib(50))  # Instant with memoization!

# cache (Python 3.9+) — simpler alias for lru_cache(maxsize=None)
# @cache
# def fib(n): ...

# reduce
product = reduce(lambda a, b: a * b, [1, 2, 3, 4, 5])
print("Product:", product)  # 120

# cmp_to_key — custom sorting comparator
nums = ['3', '30', '34', '5', '9']
nums.sort(key=cmp_to_key(lambda a, b: -1 if a+b > b+a else 1))
print("Largest number:", ''.join(nums))  # 9534330
```

## 🏆 One-Liner Solutions Hall of Fame

```python
from collections import Counter
from itertools import accumulate

# Contains Duplicate
contains_dup = lambda nums: len(nums) != len(set(nums))
print("Contains dup:", contains_dup([1,2,3,1]))  # True

# Valid Anagram
is_anagram = lambda s, t: Counter(s) == Counter(t)
print("Anagram:", is_anagram("listen", "silent"))  # True

# Single Number (XOR)
from functools import reduce
from operator import xor
single = lambda nums: reduce(xor, nums)
print("Single:", single([4,1,2,1,2]))  # 4

# Missing Number
missing = lambda nums: len(nums) * (len(nums) + 1) // 2 - sum(nums)
print("Missing:", missing([3,0,1]))  # 2

# Palindrome Check
is_pal = lambda s: s == s[::-1]
print("Palindrome:", is_pal("racecar"))  # True

# Flatten 2D List
flatten = lambda matrix: [x for row in matrix for x in row]
print("Flatten:", flatten([[1,2],[3,4]]))  # [1,2,3,4]

# Matrix Transpose
transpose = lambda m: list(map(list, zip(*m)))
print("Transpose:", transpose([[1,2],[3,4]]))  # [[1,3],[2,4]]

# Prefix Sums
prefix = lambda arr: [0] + list(accumulate(arr))
print("Prefix:", prefix([1,2,3,4]))  # [0,1,3,6,10]
```
