# 🔢 Arrays & Hashing — The Foundation of Everything

Arrays and hash maps are the two most important data structures in competitive programming.
If you master these, you'll solve 40% of all contest problems.

**Key Patterns:**
- Hash maps turn O(n²) searches into O(n)
- Sets give you O(1) membership testing
- Counter gives you frequency counting in one line
- defaultdict eliminates key-existence checks

**Recognition Triggers:**
- "find pair / complement / target" → Hash Map
- "frequency / count / group" → Counter / defaultdict
- "duplicate / seen before" → Set
- "product without self" → Prefix/Suffix products

### dict — The Swiss Army Knife

```python
# dict basics — O(1) insert, lookup, delete
d = {}
d['key'] = 'value'       # Insert: O(1)
'key' in d                # Lookup: O(1)
del d['key']              # Delete: O(1)
d.get('missing', 0)      # Safe get with default

# Dict comprehension
squares = {x: x**2 for x in range(10)}
print("Squares:", squares)

# Invert a dict
inverted = {v: k for k, v in squares.items()}
print("Inverted:", inverted)
```

```python
# set — O(1) membership test, union, intersection
s = set()
s.add(5)                  # Insert: O(1)
5 in s                    # Lookup: O(1)
s.discard(5)              # Delete (no error if missing): O(1)

# Set operations — incredibly useful in contests
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a | b)           # {1, 2, 3, 4, 5, 6}
print("Intersection:", a & b)    # {3, 4}
print("Difference:", a - b)      # {1, 2}
print("Symmetric diff:", a ^ b)  # {1, 2, 5, 6}
```

```python
# Frequency counting with Counter
from collections import Counter

# Count anything iterable in one line
freq = Counter("abracadabra")
print("Frequency:", freq)              # Counter({'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1})
print("Most common 3:", freq.most_common(3))  # [('a', 5), ('b', 2), ('r', 2)]
print("Count of 'a':", freq['a'])      # 5
print("Count of 'z':", freq['z'])      # 0 (no KeyError!)

# Counter arithmetic
c1 = Counter("aab")
c2 = Counter("abb")
print("Sum:", c1 + c2)       # Counter({'a': 3, 'b': 3})
print("Diff:", c1 - c2)      # Counter({'a': 1}) — only positive counts
print("Intersection:", c1 & c2)  # Counter({'a': 1, 'b': 1}) — min of each
```

```python
# defaultdict — auto-initializes missing keys
from collections import defaultdict

# Group items by key
words = ["eat", "tea", "tan", "ate", "nat", "bat"]
groups = defaultdict(list)
for word in words:
    key = ''.join(sorted(word))  # Sort letters to get anagram key
    groups[key].append(word)

print("Anagram groups:")
for key, group in groups.items():
    print(f"  {key}: {group}")
```

## 🔥 Problem 1: Two Sum

> Given an array of integers nums and a target, return indices of two numbers that add up to target.

```python
# Brute Force — O(n²)
def two_sum_brute(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
    return []

# Optimized — O(n) Hash Map
def two_sum(nums, target):
    seen = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    return []

# One-liner (for style points, not readability)
def two_sum_oneliner(nums, target):
    d = {num: i for i, num in enumerate(nums)}
    return next(([i, d[target-num]] for i, num in enumerate(nums) if target-num in d and d[target-num] != i), [])

print("Two Sum:")
print("  Brute:", two_sum_brute([2, 7, 11, 15], 9))  # [0, 1]
print("  Optimal:", two_sum([2, 7, 11, 15], 9))       # [0, 1]
print("  One-liner:", two_sum_oneliner([2, 7, 11, 15], 9))  # [0, 1]
print()
print("💡 Key Insight: complement = target - num. Hash map replaces inner loop.")
print("   Time: O(n²) → O(n). Space: O(1) → O(n). WORTH IT.")
```

## 🔥 Problem 2: Contains Duplicate

> Given an integer array nums, return True if any value appears at least twice.

```python
# Brute Force — O(n²)
def contains_duplicate_brute(nums):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return True
    return False

# Optimized — O(n) Set
def contains_duplicate(nums):
    return len(nums) != len(set(nums))

# With early exit
def contains_duplicate_v2(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False

print("Contains Duplicate:")
print("  [1,2,3,1]:", contains_duplicate([1,2,3,1]))  # True
print("  [1,2,3,4]:", contains_duplicate([1,2,3,4]))  # False
print("\n💡 Key Insight: Set gives O(1) lookup → entire solution is O(n).")
```

## 🔥 Problem 3: Group Anagrams

> Given an array of strings, group anagrams together.

```python
# Brute Force — O(n² * k) where k is max string length
# Compare every pair of strings to check if they're anagrams

# Optimized — O(n * k log k) Sorting-based key
def group_anagrams(strs):
    groups = defaultdict(list)
    for s in strs:
        key = ''.join(sorted(s))  # Sort letters → anagram key
        groups[key].append(s)
    return list(groups.values())

# Even better — O(n * k) Counting-based key
def group_anagrams_counting(strs):
    groups = defaultdict(list)
    for s in strs:
        # Count each letter → use as key (avoids sorting)
        count = [0] * 26
        for c in s:
            count[ord(c) - ord('a')] += 1
        groups[tuple(count)].append(s)
    return list(groups.values())

strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
print("Group Anagrams:")
for group in group_anagrams(strs):
    print(f"  {group}")
# Output: ['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']
print("\n💡 Key Insight: Anagrams have the same sorted letters. Use sorted string as dict key.")
```

## 🔥 Problem 4: Top K Frequent Elements

> Given an array and integer k, return the k most frequent elements.

```python
# Approach 1: Counter + most_common — O(n log n) but very Pythonic
def top_k_frequent(nums, k):
    return [x for x, _ in Counter(nums).most_common(k)]

# Approach 2: Bucket Sort — O(n)
def top_k_frequent_bucket(nums, k):
    freq = Counter(nums)
    # Bucket: index = frequency, value = list of numbers with that frequency
    buckets = [[] for _ in range(len(nums) + 1)]
    for num, count in freq.items():
        buckets[count].append(num)
    
    # Collect from highest frequency buckets
    result = []
    for i in range(len(buckets) - 1, -1, -1):
        result.extend(buckets[i])
        if len(result) >= k:
            return result[:k]
    return result[:k]

# Approach 3: Heap — O(n log k)
import heapq
def top_k_frequent_heap(nums, k):
    freq = Counter(nums)
    return heapq.nlargest(k, freq, key=freq.get)

print("Top K Frequent:")
print("  [1,1,1,2,2,3], k=2:", top_k_frequent([1,1,1,2,2,3], 2))  # [1, 2]
print("  Bucket sort:", top_k_frequent_bucket([1,1,1,2,2,3], 2))    # [1, 2]
print("  Heap:", top_k_frequent_heap([1,1,1,2,2,3], 2))             # [1, 2]
print("\n💡 Key Insight: Counter does the counting, then sort/bucket/heap picks top k.")
```

## 🔥 Problem 5: Product of Array Except Self

> Return array where `output[i]` = product of all elements except `nums[i]`. No division!

```python
# Brute Force — O(n²)
def product_except_self_brute(nums):
    n = len(nums)
    result = []
    for i in range(n):
        product = 1
        for j in range(n):
            if i != j:
                product *= nums[j]
        result.append(product)
    return result

# Optimized — O(n) with prefix/suffix products
def product_except_self(nums):
    n = len(nums)
    result = [1] * n
    
    # Left pass: result[i] = product of all elements to the LEFT of i
    prefix = 1
    for i in range(n):
        result[i] = prefix
        prefix *= nums[i]
    
    # Right pass: multiply by product of all elements to the RIGHT of i
    suffix = 1
    for i in range(n - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]
    
    return result

print("Product Except Self:")
print("  [1,2,3,4]:", product_except_self([1,2,3,4]))  # [24, 12, 8, 6]
print("  [-1,1,0,-3,3]:", product_except_self([-1,1,0,-3,3]))  # [0, 0, 9, 0, 0]
print("\n💡 Key Insight: result[i] = (product of left) × (product of right)")
print("   Two passes: left→right for prefix, right→left for suffix.")
print("   Time: O(n), Space: O(1) extra (output doesn't count).")
```

## 🔥 Problem 6: Valid Sudoku

> Determine if a 9x9 Sudoku board is valid. Only filled cells need to be validated.

```python
def is_valid_sudoku(board):
    """Check rows, columns, and 3x3 boxes using sets.
    
    Time: O(81) = O(1) — board is always 9x9
    Space: O(81) = O(1)
    """
    rows = defaultdict(set)
    cols = defaultdict(set)
    boxes = defaultdict(set)  # key: (row//3, col//3)
    
    for r in range(9):
        for c in range(9):
            val = board[r][c]
            if val == '.':
                continue
            
            box_key = (r // 3, c // 3)
            
            if val in rows[r] or val in cols[c] or val in boxes[box_key]:
                return False
            
            rows[r].add(val)
            cols[c].add(val)
            boxes[box_key].add(val)
    
    return True

# Test board
board = [
    ["5","3",".",".","7",".",".",".","."],
    ["6",".",".","1","9","5",".",".","."],
    [".","9","8",".",".",".",".","6","."],
    ["8",".",".",".","6",".",".",".","3"],
    ["4",".",".","8",".","3",".",".","1"],
    ["7",".",".",".","2",".",".",".","6"],
    [".","6",".",".",".",".","2","8","."],
    [".",".",".","4","1","9",".",".","5"],
    [".",".",".",".","8",".",".","7","9"]
]
print("Valid Sudoku:", is_valid_sudoku(board))  # True
print("\n💡 Key Insight: Three sets (row, col, box). Box index = (r//3, c//3).")
```

## 🔥 Problem 7: Longest Consecutive Sequence

> Given an unsorted array, find the length of the longest consecutive elements sequence. Must be O(n).

```python
# Brute Force — O(n log n) sort first, then scan
def longest_consecutive_sort(nums):
    if not nums:
        return 0
    nums = sorted(set(nums))
    max_len = current = 1
    for i in range(1, len(nums)):
        if nums[i] == nums[i-1] + 1:
            current += 1
            max_len = max(max_len, current)
        else:
            current = 1
    return max_len

# Optimized — O(n) with Hash Set
def longest_consecutive(nums):
    """Use a set. Only start counting from the BEGINNING of a sequence.
    A number starts a sequence if (num - 1) is NOT in the set.
    
    Time: O(n) — each number is visited at most twice
    Space: O(n)
    """
    num_set = set(nums)
    max_len = 0
    
    for num in num_set:
        # Only start counting from the beginning of a sequence
        if num - 1 not in num_set:  # This is the START of a sequence
            current = num
            length = 1
            while current + 1 in num_set:
                current += 1
                length += 1
            max_len = max(max_len, length)
    
    return max_len

print("Longest Consecutive Sequence:")
print("  [100,4,200,1,3,2]:", longest_consecutive([100,4,200,1,3,2]))  # 4 (1,2,3,4)
print("  [0,3,7,2,5,8,4,6,0,1]:", longest_consecutive([0,3,7,2,5,8,4,6,0,1]))  # 9
print("\n💡 Key Insight: Only start from sequence beginnings (num-1 not in set).")
print("   This ensures each number is visited at most twice → O(n).")
```

## 🔥 Problem 8: Three Sum

> Find all unique triplets in the array which sum to zero.

```python
def three_sum(nums):
    """Sort + fix one element + two pointers on the rest.
    
    Time: O(n²) — sort O(n log n) + for each element O(n) two-pointer scan
    Space: O(1) extra (excluding output)
    """
    nums.sort()
    result = []
    
    for i in range(len(nums) - 2):
        # Skip duplicates for first element
        if i > 0 and nums[i] == nums[i - 1]:
            continue
        # Early termination
        if nums[i] > 0:
            break
        
        target = -nums[i]
        left, right = i + 1, len(nums) - 1
        
        while left < right:
            current = nums[left] + nums[right]
            if current == target:
                result.append([nums[i], nums[left], nums[right]])
                while left < right and nums[left] == nums[left + 1]:
                    left += 1
                while left < right and nums[right] == nums[right - 1]:
                    right -= 1
                left += 1
                right -= 1
            elif current < target:
                left += 1
            else:
                right -= 1
    
    return result

print("Three Sum:")
print("  [-1,0,1,2,-1,-4]:", three_sum([-1,0,1,2,-1,-4]))
# Output: [[-1,-1,2], [-1,0,1]]
print("\n💡 Key Insight: Three Sum = fix one + Two Sum (sorted). Skip duplicates!")
```

## 🎯 How to Spot a Hashing Problem in 5 Seconds

### The Recognition Trigger Table

| You see... | You reach for... | Why |
|-----------|-----------------|-----|
| "find pair/complement" | `dict` (hash map) | O(1) complement lookup |
| "check if exists" / "duplicate" | `set` | O(1) membership test |
| "count frequency" / "most common" | `Counter` | One-line frequency counting |
| "group by property" | `defaultdict(list)` | Auto-creates empty lists |
| "consecutive sequence" | `set` | O(1) neighbor check |
| "product except self" | Prefix/Suffix arrays | Build up from both directions |

### The Decision Process

1. **Do I need to search for something inside a loop?** → Hash map/set
2. **Do I need to count things?** → Counter
3. **Do I need to group things?** → defaultdict(list)
4. **Do I need uniqueness?** → set
5. **Do I need index lookup?** → dict (value → index)

**Master takeaway:** The hash map is your #1 weapon. When in doubt, hash it out. 🔥
